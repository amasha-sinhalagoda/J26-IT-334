```python

# Project paths
PROJECT_DIR = Path.cwd().resolve()

while not (PROJECT_DIR / "data" / "raw").is_dir():
    if PROJECT_DIR.parent == PROJECT_DIR:
        raise FileNotFoundError("Project directory not found")
    PROJECT_DIR = PROJECT_DIR.parent

input_file = (
    PROJECT_DIR / "data" / "raw" / "rainfall" / "kelani"
    / "2015 - 2025 daily rainfall.xlsx"
)

output_dir = PROJECT_DIR / "data" / "processed" / "kelani"
output_dir.mkdir(parents=True, exist_ok=True)

output_file = output_dir / "kelani_daily_rainfall.xlsx"


def preprocess_rainfall(input_file, output_file):
    """
    Process all worksheets:
    - Extract location from the worksheet name
    - Skip the title and metadata rows
    - Rename Year/Month to Date
    - Add Location as the first column
    - Save all processed worksheets into a new Excel file
    """
    with pd.ExcelFile(input_file) as workbook:
        with pd.ExcelWriter(output_file, engine="openpyxl") as writer:

            for sheet_name in workbook.sheet_names:
                # Row 3 (Excel row number) contains the column headers
                df = pd.read_excel(
                    workbook,
                    sheet_name=sheet_name,
                    header=2
                )

                # Remove completely empty columns
                df = df.dropna(axis=1, how="all")

                # Clean column names
                df.columns = df.columns.astype(str).str.strip()

                # Remove empty rows
                df = df.dropna(how="all").copy()

                # Rename the date column
                df = df.rename(columns={"Year/Month": "Date"})

                # Extract station name from worksheet name
                location = sheet_name.split("-", 1)[-1].strip()

                # Add location as the first column
                df.insert(0, "Location", location)

                # Save each processed worksheet
                df.to_excel(
                    writer,
                    sheet_name=sheet_name[:31],
                    index=False
                )

                print(
                    f"Processed: {sheet_name} | "
                    f"Location: {location} | "
                    f"Rows: {len(df)}"
                )

    print(f"\nPreprocessing completed.")
    print(f"Saved file: {output_file}")


preprocess_rainfall(input_file, output_file)
```

```python

# Reprocess all sheets with the correct column headers
with pd.ExcelFile(input_file) as workbook:
    with pd.ExcelWriter(output_file, engine="openpyxl") as writer:

        for sheet_name in workbook.sheet_names:
            df = pd.read_excel(
                workbook,
                sheet_name=sheet_name,
                header=2
            )

            # Remove completely empty columns and rows
            df = df.dropna(axis=1, how="all")
            df = df.dropna(how="all").copy()
            df.columns = df.columns.astype(str).str.strip()

            # Rename the first date column
            df = df.rename(columns={"Year/Month": "Date"})

            # Extract station name
            location = sheet_name.split("-", 1)[-1].strip()

            # Add Location as the first column
            df.insert(0, "Location", location)

            # Save the corrected sheet
            df.to_excel(
                writer,
                sheet_name=sheet_name[:31],
                index=False
            )

print("All worksheets corrected and saved.")
print("Output:", output_file)
```

```python
# Update the existing processed Excel file
temp_file = output_file.with_name("kelani_temp.xlsx")

with pd.ExcelFile(output_file, engine="openpyxl") as workbook:
    with pd.ExcelWriter(temp_file, engine="openpyxl") as writer:
        for sheet_name in workbook.sheet_names:
            df = pd.read_excel(workbook, sheet_name=sheet_name, header=None)

            # Delete the first row and promote the second row to headers
            df = df.iloc[1:].reset_index(drop=True)
            df.columns = df.iloc[0].astype(str).str.strip()
            df = df.iloc[1:].reset_index(drop=True)

            # Save the updated worksheet
            df.to_excel(writer, sheet_name=sheet_name, index=False)

# Replace the existing processed file
temp_file.replace(output_file)

print("Updated file:", output_file)

# Display the top 5 rows of every worksheet
with pd.ExcelFile(output_file, engine="openpyxl") as workbook:
    for sheet_name in workbook.sheet_names:
        print(f"\nWorksheet: {sheet_name}")
        display(pd.read_excel(workbook, sheet_name=sheet_name, nrows=5))
        
```

```python
# Update the existing processed Excel file
temp_file = output_file.with_name("kelani_temp.xlsx")

with pd.ExcelFile(output_file, engine="openpyxl") as workbook:
    with pd.ExcelWriter(temp_file, engine="openpyxl") as writer:
        for sheet_name in workbook.sheet_names:
            df = pd.read_excel(workbook, sheet_name=sheet_name)

            # Rename the first column to Location
            df.rename(columns={df.columns[0]: "Location"}, inplace=True)

            df.to_excel(writer, sheet_name=sheet_name, index=False)

# Replace the existing processed file
temp_file.replace(output_file)

print("Updated file:", output_file)

# Display the top 5 rows of every worksheet
with pd.ExcelFile(output_file, engine="openpyxl") as workbook:
    for sheet_name in workbook.sheet_names:
        print(f"\nWorksheet: {sheet_name}")
        display(pd.read_excel(workbook, sheet_name=sheet_name, nrows=5))
```

```python
# Input: existing processed Kelani rainfall workbook
input_file = output_file

# Output: new workbook with all sheets combined
combined_file = input_file.with_name("kelani_daily_rainfall_combined.xlsx")

# Read every worksheet
with pd.ExcelFile(input_file, engine="openpyxl") as workbook:
    all_sheets = {
        sheet_name: pd.read_excel(workbook, sheet_name=sheet_name)
        for sheet_name in workbook.sheet_names
    }

# Combine vertically without dropping rows
combined_df = pd.concat(
    list(all_sheets.values()),
    ignore_index=True,
    sort=False
)

# Save to a new Excel file
combined_df.to_excel(
    combined_file,
    sheet_name="Kelani_Rainfall",
    index=False,
    engine="openpyxl"
)

print("Combination completed successfully!")
print("Worksheets combined:", len(all_sheets))
print("Total rows:", len(combined_df))
print("Total columns:", len(combined_df.columns))
print("Saved to:", combined_file)

# Verify the first 5 rows
display(combined_df.head(5))

# Show the number of rows contributed by each original worksheet
print("\nRows from each worksheet:")
for sheet_name, df in all_sheets.items():
    print(f"{sheet_name}: {len(df)} rows")
```

```python

# Path to the existing combined workbook
combined_file = (
    PROJECT_DIR / "data" / "processed" / "kelani"
    / "kelani_daily_rainfall_combined.xlsx"
)

temp_file = combined_file.with_name("kelani_combined_temp.xlsx")

with pd.ExcelFile(combined_file, engine="openpyxl") as workbook:
    df = pd.read_excel(workbook, sheet_name=workbook.sheet_names[0])

# Identify the two Year/Month columns
print("Current columns:", df.columns.tolist())

year_month_cols = [
    col for col in df.columns
    if str(col).strip().upper() in ["YEARMONTH", "YEAR/MONTH"]
]

if len(year_month_cols) != 2:
    raise ValueError(
        f"Expected 2 Year/Month columns, found {len(year_month_cols)}: "
        f"{year_month_cols}. Check the printed column names."
    )

# Keep the first column's values where present; otherwise use the second
first_col, second_col = year_month_cols

df[first_col] = df[first_col].combine_first(df[second_col])

# Remove the duplicate column and standardize the name
df.drop(columns=[second_col], inplace=True)
df.rename(columns={first_col: "YEARMONTH"}, inplace=True)

# Save to a temporary file, then replace the existing combined workbook
with pd.ExcelWriter(temp_file, engine="openpyxl") as writer:
    df.to_excel(writer, sheet_name="Kelani_Rainfall", index=False)

temp_file.replace(combined_file)

print("\nUpdated file:", combined_file)
print("Total rows:", len(df))
print("Total columns:", len(df.columns))
print("\nTop 5 rows:")
display(df.head(5))
```
