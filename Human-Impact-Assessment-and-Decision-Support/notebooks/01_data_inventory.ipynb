```python
import os
import pandas as pd
import numpy as np

raw_base = "../data/raw"

files = [
    ("2015 - 2025 RF.xlsx", "flood/kelani/2015 - 2025 RF.xlsx"),
    ("2015 - 2025 WL.xlsx", "flood/kelani/2015 - 2025 WL.xlsx"),
    ("2015 - 2025 Discharge.xlsx", "flood/kelani/2015 - 2025 Discharge.xlsx"),
    ("Rathnapura.xlsx", "flood/kalu/Rathnapura.xlsx"),
    ("DesInventar_NuwaraEliya_Landslides_page1.xlsx", "landslide/DesInventar_NuwaraEliya_Landslides_page1.xlsx"),
    ("DesInventar_Ratnapura_Kegalle_Floods_pages1-2.xlsx", "disaster_records/DesInventar_Ratnapura_Kegalle_Floods_pages1-2.xlsx"),
    ("GN_population_excel.xlsx", "census/GN_population_excel.xlsx"),
    ("GN_housing_excel.xlsx", "census/GN_housing_excel.xlsx"),
    ("HH_GND_excel.xlsx", "census/HH_GND_excel.xlsx")
]

for display_name, rel_path in files:
    full_path = os.path.join(raw_base, rel_path)
    print(f"\n==================================================")
    print(f"=== {display_name} ===")
    print(f"==================================================")
    print(f"Full Relative Path: {rel_path}")
    
    if not os.path.exists(full_path):
        print("❌ FILE NOT FOUND!")
        continue
        
    try:
        excel_file = pd.ExcelFile(full_path)
        sheet_names = excel_file.sheet_names
        print(f"Sheets ({len(sheet_names)}): {sheet_names}\n")
        
        for sheet in sheet_names:
            print(f"--- Sheet: {sheet} ---")
            # Read first 10 rows without header to inspect raw layout / title rows
            df_raw = pd.read_excel(full_path, sheet_name=sheet, header=None, nrows=10)
            
            # Read full sheet to get dimension and basic inspection
            df = pd.read_excel(full_path, sheet_name=sheet)
            print(f"Raw Shape: {df.shape} (Rows: {df.shape[0]}, Cols: {df.shape[1]})")
            
            print("\nHeader Sample (First 5 raw rows):")
            print(df_raw.head(5).to_string())
            
            print("\nInferred Columns:")
            print(list(df.columns))
            
            print("\nMissing Values Count (top 10 columns by missing):")
            nulls = df.isnull().sum()
            print(nulls[nulls > 0].head(10).to_string() if nulls.sum() > 0 else "No missing values detected.")
            
            # Detect potential date/year/time columns
            date_like_cols = [c for c in df.columns if any(term in str(c).lower() for term in ['date', 'year', 'month', 'time', 'day', 'serial', 'd1'])]
            if date_like_cols:
                print(f"\nDate / Time Columns Detected: {date_like_cols}")
            
            print("\n" + "-"*40)
            
    except Exception as e:
        print(f"Error reading {display_name}: {e}")
```

```python
import os
import re
import pandas as pd
import numpy as np

PROJECT_ROOT = "/Users/ohaseelanka/disaster_analytics_hub"
RAW_DIR = os.path.join(PROJECT_ROOT, "data/raw/flood/kelani")
PROCESSED_DIR = os.path.join(PROJECT_ROOT, "data/processed/flood/kelani")

os.makedirs(PROCESSED_DIR, exist_ok=True)

MONTH_MAP = {
    'JAN': 1, 'FEB': 2, 'MAR': 3, 'APR': 4, 'MAY': 5, 'JUN': 6,
    'JUL': 7, 'AUG': 8, 'SEP': 9, 'OCT': 10, 'NOV': 11, 'DEC': 12
}

def clean_station_name(name):
    """Cleans sheet names to standard station identifier."""
    name = re.sub(r'^[Kk][Nn]\d+[-_\s]*', '', name)
    name = name.replace("N'Street", "N_Street").strip()
    return name.title()


def parse_rainfall(file_path):
    """Processes Rainfall Excel Workbook into long format."""
    print("Processing Rainfall Data...")
    xls = pd.ExcelFile(file_path)
    all_rf_data = []

    for sheet in xls.sheet_names:
        df_raw = pd.read_excel(xls, sheet_name=sheet, header=None)
        station = clean_station_name(sheet)
        
        # Extract coordinates if present in header row 1
        coord_row = df_raw.iloc[1].dropna().astype(str).str.cat(sep=" ")
        x_match = re.search(r'(\d+)\s*X', coord_row)
        y_match = re.search(r'(\d+)\s*Y', coord_row)
        coord_x = float(x_match.group(1)) if x_match else np.nan
        coord_y = float(y_match.group(1)) if y_match else np.nan

        # Header is at row 3 (0-indexed)
        df_data = df_raw.iloc[4:].copy()
        df_data.columns = [str(col).strip() for col in df_raw.iloc[3]]
        
        # Identify Year and Month columns
        year_col = [c for c in df_data.columns if 'YEAR' in c.upper()][0]
        month_col = [c for c in df_data.columns if 'MONTH' in c.upper()][0]
        day_cols = [c for c in df_data.columns if re.match(r'^D\d+$', c, re.IGNORECASE)]

        # Melt daily columns to long format
        df_melted = pd.melt(
            df_data,
            id_vars=[year_col, month_col],
            value_vars=day_cols,
            var_name='day',
            value_name='rainfall_mm'
        )

        df_melted['day'] = df_melted['day'].str.extract(r'(\d+)').astype(int)
        
        # Parse month string or integer
        def convert_month(m):
            if pd.isna(m): return np.nan
            m_str = str(m).strip().upper()
            return MONTH_MAP.get(m_str, int(m) if m_str.isdigit() else np.nan)

        df_melted['month_num'] = df_melted[month_col].apply(convert_month)
        df_melted['year_num'] = pd.to_numeric(df_melted[year_col], errors='coerce')
        
        # Coerce non-numeric rainfall strings (e.g. 'N.A.') to NaN
        df_melted['rainfall_mm'] = pd.to_numeric(df_melted['rainfall_mm'], errors='coerce')
        
        # Construct Datetime
        df_melted['date'] = pd.to_datetime(
            dict(year=df_melted['year_num'], month=df_melted['month_num'], day=df_melted['day']),
            errors='coerce'
        )
        
        df_clean = df_melted.dropna(subset=['date']).copy()
        df_clean['station'] = station
        df_clean['coord_x'] = coord_x
        df_clean['coord_y'] = coord_y
        
        all_rf_data.append(df_clean[['date', 'station', 'rainfall_mm', 'coord_x', 'coord_y']])

    return pd.concat(all_rf_data, ignore_index=True)


def parse_water_level(file_path):
    """Processes Water Level Excel Workbook into long format."""
    print("Processing Water Level Data...")
    xls = pd.ExcelFile(file_path)
    all_wl_data = []

    for sheet in xls.sheet_names:
        df_raw = pd.read_excel(xls, sheet_name=sheet, header=None)
        station = clean_station_name(sheet)
        
        # Read MSL Datum offset from header
        msl_val = np.nan
        msl_str = str(df_raw.iloc[2, 1]) if df_raw.shape[1] > 1 else ""
        msl_match = re.search(r'([-+]?\d*\.\d+|\d+)', msl_str)
        if msl_match:
            msl_val = float(msl_match.group(1))

        # Header at Row 4
        df_data = df_raw.iloc[5:].copy()
        df_data = df_data.iloc[:, :4]
        df_data.columns = ['year', 'month', 'day', 'water_level']
        
        df_data['year'] = pd.to_numeric(df_data['year'], errors='coerce')
        df_data['month'] = pd.to_numeric(df_data['month'], errors='coerce')
        df_data['day'] = pd.to_numeric(df_data['day'], errors='coerce')
        df_data['water_level'] = pd.to_numeric(df_data['water_level'], errors='coerce')

        # Unit Conversion: Convert N'Street (ftMSL) to mMSL
        if 'N_Street' in station or 'N\'Street' in sheet:
            df_data['water_level_mMSL'] = df_data['water_level'] * 0.3048
            unit = 'mMSL (converted from ft)'
        else:
            df_data['water_level_mMSL'] = df_data['water_level']
            unit = 'mMSL'

        df_data['date'] = pd.to_datetime(
            dict(year=df_data['year'], month=df_data['month'], day=df_data['day']),
            errors='coerce'
        )

        df_clean = df_data.dropna(subset=['date']).copy()
        df_clean['station'] = station
        df_clean['datum_msl_m'] = msl_val
        df_clean['unit'] = unit

        all_wl_data.append(df_clean[['date', 'station', 'water_level_mMSL', 'datum_msl_m', 'unit']])

    return pd.concat(all_wl_data, ignore_index=True)


def parse_discharge(file_path):
    """Processes Discharge Excel Workbook into long format."""
    print("Processing Discharge Data...")
    xls = pd.ExcelFile(file_path)
    all_dis_data = []

    for sheet in xls.sheet_names:
        df_raw = pd.read_excel(xls, sheet_name=sheet, header=None)
        station = clean_station_name(sheet)

        # Header at Row 3
        df_data = df_raw.iloc[4:].copy()
        df_data.columns = [str(col).strip() for col in df_raw.iloc[3]]
        
        # Separate date identification and days
        year_col = [c for c in df_data.columns if 'YEAR' in c.upper()]
        
        # Handle combined YEARMONTH string or split columns
        if year_col:
            ym_col = year_col[0]
            df_data['year_num'] = df_data[ym_col].astype(str).str[:4]
            df_data['month_num'] = df_data[ym_col].astype(str).str[4:6]
        else:
            df_data['year_num'] = df_data.iloc[:, 0].astype(str).str[:4]
            df_data['month_num'] = df_data.iloc[:, 0].astype(str).str[4:6]

        day_cols = [c for c in df_data.columns if re.match(r'^D\d+$', c, re.IGNORECASE)]

        df_melted = pd.melt(
            df_data,
            id_vars=['year_num', 'month_num'],
            value_vars=day_cols,
            var_name='day',
            value_name='discharge_cumecs'
        )

        df_melted['day'] = df_melted['day'].str.extract(r'(\d+)').astype(int)
        df_melted['year_num'] = pd.to_numeric(df_melted['year_num'], errors='coerce')
        df_melted['month_num'] = pd.to_numeric(df_melted['month_num'], errors='coerce')
        df_melted['discharge_cumecs'] = pd.to_numeric(df_melted['discharge_cumecs'], errors='coerce')

        df_melted['date'] = pd.to_datetime(
            dict(year=df_melted['year_num'], month=df_melted['month_num'], day=df_melted['day']),
            errors='coerce'
        )

        df_clean = df_melted.dropna(subset=['date']).copy()
        df_clean['station'] = station

        all_dis_data.append(df_clean[['date', 'station', 'discharge_cumecs']])

    return pd.concat(all_dis_data, ignore_index=True)


def main():
    rf_path = os.path.join(RAW_DIR, "2015 - 2025 RF.xlsx")
    wl_path = os.path.join(RAW_DIR, "2015 - 2025 WL.xlsx")
    dis_path = os.path.join(RAW_DIR, "2015 - 2025 Discharge.xlsx")

    # Run processing
    df_rf = parse_rainfall(rf_path)
    df_wl = parse_water_level(wl_path)
    df_dis = parse_discharge(dis_path)

    # Save individual clean long-format datasets
    df_rf.to_csv(os.path.join(PROCESSED_DIR, "kelani_daily_rainfall_2015_2025.csv"), index=False)
    df_wl.to_csv(os.path.join(PROCESSED_DIR, "kelani_daily_water_level_2015_2025.csv"), index=False)
    df_dis.to_csv(os.path.join(PROCESSED_DIR, "kelani_daily_discharge_2015_2025.csv"), index=False)

    # Merge into a single master time-series dataset grouped by Date and Station
    print("Building unified master time-series table...")
    df_master = pd.merge(df_rf, df_wl, on=['date', 'station'], how='outer')
    df_master = pd.merge(df_master, df_dis, on=['date', 'station'], how='outer')
    
    df_master.sort_values(by=['date', 'station'], inplace=True)
    df_master.to_csv(os.path.join(PROCESSED_DIR, "kelani_master_daily_analytics_2015_2025.csv"), index=False)

    print("Pipeline Execution Completed Successfully.")
    print(f"Master output saved to: {os.path.join(PROCESSED_DIR, 'kelani_master_daily_analytics_2015_2025.csv')}")

if __name__ == '__main__':
    main()
```

```python
def parse_rainfall(file_path):
    """Processes Rainfall Excel Workbook into long format."""
    print("Processing Rainfall Data...")
    xls = pd.ExcelFile(file_path)
    all_rf_data = []

    for sheet in xls.sheet_names:
        df_raw = pd.read_excel(xls, sheet_name=sheet, header=None)
        station = clean_station_name(sheet)
        
        # Extract coordinates if present in header row 1
        coord_row = df_raw.iloc[1].dropna().astype(str).str.cat(sep=" ")
        x_match = re.search(r'(\d+)\s*X', coord_row)
        y_match = re.search(r'(\d+)\s*Y', coord_row)
        coord_x = float(x_match.group(1)) if x_match else np.nan
        coord_y = float(y_match.group(1)) if y_match else np.nan

        # Header is at row 3 (0-indexed)
        df_data = df_raw.iloc[4:].copy()
        df_data.columns = [str(col).strip() for col in df_raw.iloc[3]]
        
        # Identify day columns
        day_cols = [c for c in df_data.columns if re.match(r'^D\d+$', c, re.IGNORECASE)]

        # Check if Year/Month are separate or combined
        year_cols = [c for c in df_data.columns if 'YEAR' in c.upper() and c not in day_cols]
        month_cols = [c for c in df_data.columns if 'MONTH' in c.upper() and c not in day_cols]

        if year_cols and month_cols and year_cols[0] != month_cols[0]:
            # Separate columns
            year_col, month_col = year_cols[0], month_cols[0]
            id_vars = [year_col, month_col]
            
            df_melted = pd.melt(
                df_data,
                id_vars=id_vars,
                value_vars=day_cols,
                var_name='day',
                value_name='rainfall_mm'
            )
            df_melted['year_str'] = df_melted[year_col].astype(str)
            df_melted['month_str'] = df_melted[month_col].astype(str)
        else:
            # Combined Year/Month column (e.g., 'Year/Month' or '201501')
            ym_col = year_cols[0] if year_cols else df_data.columns[0]
            id_vars = [ym_col]
            
            df_melted = pd.melt(
                df_data,
                id_vars=id_vars,
                value_vars=day_cols,
                var_name='day',
                value_name='rainfall_mm'
            )
            # Handle YYYYMM format or split text
            ym_cleaned = df_melted[ym_col].astype(str).str.replace(r'\D', '', regex=True)
            df_melted['year_str'] = ym_cleaned.str[:4]
            df_melted['month_str'] = ym_cleaned.str[4:6]

        df_melted['day'] = df_melted['day'].str.extract(r'(\d+)').astype(int)
        
        def convert_month(m):
            if pd.isna(m): return np.nan
            m_str = str(m).strip().upper()
            return MONTH_MAP.get(m_str, int(m_str) if m_str.isdigit() else np.nan)

        df_melted['month_num'] = df_melted['month_str'].apply(convert_month)
        df_melted['year_num'] = pd.to_numeric(df_melted['year_str'], errors='coerce')
        df_melted['rainfall_mm'] = pd.to_numeric(df_melted['rainfall_mm'], errors='coerce')
        
        # Construct Datetime
        df_melted['date'] = pd.to_datetime(
            dict(year=df_melted['year_num'], month=df_melted['month_num'], day=df_melted['day']),
            errors='coerce'
        )
        
        df_clean = df_melted.dropna(subset=['date']).copy()
        df_clean['station'] = station
        df_clean['coord_x'] = coord_x
        df_clean['coord_y'] = coord_y
        
        all_rf_data.append(df_clean[['date', 'station', 'rainfall_mm', 'coord_x', 'coord_y']])

    return pd.concat(all_rf_data, ignore_index=True)
```

```python
def parse_discharge(file_path):
    """Processes Discharge Excel Workbook into long format."""
    print("Processing Discharge Data...")
    xls = pd.ExcelFile(file_path)
    all_dis_data = []

    for sheet in xls.sheet_names:
        df_raw = pd.read_excel(xls, sheet_name=sheet, header=None)
        station = clean_station_name(sheet)

        # Header at Row 3
        df_data = df_raw.iloc[4:].copy()
        df_data.columns = [str(col).strip() for col in df_raw.iloc[3]]
        
        day_cols = [c for c in df_data.columns if re.match(r'^D\d+$', c, re.IGNORECASE)]
        
        # Identify first non-day column as Year/Month column
        ym_col = [c for c in df_data.columns if c not in day_cols][0]

        df_melted = pd.melt(
            df_data,
            id_vars=[ym_col],
            value_vars=day_cols,
            var_name='day',
            value_name='discharge_cumecs'
        )

        ym_cleaned = df_melted[ym_col].astype(str).str.replace(r'\D', '', regex=True)
        df_melted['year_num'] = pd.to_numeric(ym_cleaned.str[:4], errors='coerce')
        df_melted['month_num'] = pd.to_numeric(ym_cleaned.str[4:6], errors='coerce')
        df_melted['day'] = df_melted['day'].str.extract(r'(\d+)').astype(int)
        df_melted['discharge_cumecs'] = pd.to_numeric(df_melted['discharge_cumecs'], errors='coerce')

        df_melted['date'] = pd.to_datetime(
            dict(year=df_melted['year_num'], month=df_melted['month_num'], day=df_melted['day']),
            errors='coerce'
        )

        df_clean = df_melted.dropna(subset=['date']).copy()
        df_clean['station'] = station

        all_dis_data.append(df_clean[['date', 'station', 'discharge_cumecs']])

    return pd.concat(all_dis_data, ignore_index=True)
```

```python
import os
import re
import pandas as pd
import numpy as np

PROJECT_ROOT = "/Users/ohaseelanka/disaster_analytics_hub"
RAW_DIR = os.path.join(PROJECT_ROOT, "data/raw/flood/kelani")
PROCESSED_DIR = os.path.join(PROJECT_ROOT, "data/processed/flood/kelani")

os.makedirs(PROCESSED_DIR, exist_ok=True)

MONTH_MAP = {
    'JAN': 1, 'FEB': 2, 'MAR': 3, 'APR': 4, 'MAY': 5, 'JUN': 6,
    'JUL': 7, 'AUG': 8, 'SEP': 9, 'OCT': 10, 'NOV': 11, 'DEC': 12
}

def clean_station_name(name):
    """Cleans sheet names to standard station identifier."""
    name = re.sub(r'^[Kk][Nn]\d+[-_\s]*', '', str(name))
    name = name.replace("N'Street", "N_Street").strip()
    return name.title()


def parse_rainfall(file_path):
    """Processes Rainfall Excel Workbook into long format."""
    print("Processing Rainfall Data...")
    xls = pd.ExcelFile(file_path)
    all_rf_data = []

    for sheet in xls.sheet_names:
        print(f"  -> Reading Rainfall Sheet: {sheet}")
        df_raw = pd.read_excel(xls, sheet_name=sheet, header=None)
        station = clean_station_name(sheet)
        
        # Extract coordinates if present in header row 1
        coord_row = df_raw.iloc[1].dropna().astype(str).str.cat(sep=" ")
        x_match = re.search(r'(\d+)\s*X', coord_row)
        y_match = re.search(r'(\d+)\s*Y', coord_row)
        coord_x = float(x_match.group(1)) if x_match else np.nan
        coord_y = float(y_match.group(1)) if y_match else np.nan

        # Header is at row 3 (0-indexed)
        df_data = df_raw.iloc[4:].copy()
        df_data.columns = [str(col).strip() for col in df_raw.iloc[3]]
        
        # Identify day columns
        day_cols = [c for c in df_data.columns if re.match(r'^D\d+$', c, re.IGNORECASE)]

        # Check if Year/Month are separate or combined
        year_cols = [c for c in df_data.columns if 'YEAR' in c.upper() and c not in day_cols]
        month_cols = [c for c in df_data.columns if 'MONTH' in c.upper() and c not in day_cols]

        if year_cols and month_cols and year_cols[0] != month_cols[0]:
            year_col, month_col = year_cols[0], month_cols[0]
            df_melted = pd.melt(
                df_data,
                id_vars=[year_col, month_col],
                value_vars=day_cols,
                var_name='day',
                value_name='rainfall_mm'
            )
            df_melted['year_str'] = df_melted[year_col].astype(str)
            df_melted['month_str'] = df_melted[month_col].astype(str)
        else:
            ym_col = year_cols[0] if year_cols else df_data.columns[0]
            df_melted = pd.melt(
                df_data,
                id_vars=[ym_col],
                value_vars=day_cols,
                var_name='day',
                value_name='rainfall_mm'
            )
            ym_cleaned = df_melted[ym_col].astype(str).str.replace(r'\D', '', regex=True)
            df_melted['year_str'] = ym_cleaned.str[:4]
            df_melted['month_str'] = ym_cleaned.str[4:6]

        df_melted['day'] = df_melted['day'].str.extract(r'(\d+)').astype(int)
        
        def convert_month(m):
            if pd.isna(m): return np.nan
            m_str = str(m).strip().upper()
            return MONTH_MAP.get(m_str, int(m_str) if m_str.isdigit() else np.nan)

        df_melted['month_num'] = df_melted['month_str'].apply(convert_month)
        df_melted['year_num'] = pd.to_numeric(df_melted['year_str'], errors='coerce')
        df_melted['rainfall_mm'] = pd.to_numeric(df_melted['rainfall_mm'], errors='coerce')
        
        # Construct Datetime
        df_melted['date'] = pd.to_datetime(
            dict(year=df_melted['year_num'], month=df_melted['month_num'], day=df_melted['day']),
            errors='coerce'
        )
        
        df_clean = df_melted.dropna(subset=['date']).copy()
        df_clean['station'] = station
        df_clean['coord_x'] = coord_x
        df_clean['coord_y'] = coord_y
        
        all_rf_data.append(df_clean[['date', 'station', 'rainfall_mm', 'coord_x', 'coord_y']])

    return pd.concat(all_rf_data, ignore_index=True)


def parse_water_level(file_path):
    """Processes Water Level Excel Workbook into long format."""
    print("Processing Water Level Data...")
    xls = pd.ExcelFile(file_path)
    all_wl_data = []

    for sheet in xls.sheet_names:
        print(f"  -> Reading Water Level Sheet: {sheet}")
        df_raw = pd.read_excel(xls, sheet_name=sheet, header=None)
        station = clean_station_name(sheet)
        
        msl_val = np.nan
        msl_str = str(df_raw.iloc[2, 1]) if df_raw.shape[1] > 1 else ""
        msl_match = re.search(r'([-+]?\d*\.\d+|\d+)', msl_str)
        if msl_match:
            msl_val = float(msl_match.group(1))

        df_data = df_raw.iloc[5:].copy()
        df_data = df_data.iloc[:, :4]
        df_data.columns = ['year', 'month', 'day', 'water_level']
        
        df_data['year'] = pd.to_numeric(df_data['year'], errors='coerce')
        df_data['month'] = pd.to_numeric(df_data['month'], errors='coerce')
        df_data['day'] = pd.to_numeric(df_data['day'], errors='coerce')
        df_data['water_level'] = pd.to_numeric(df_data['water_level'], errors='coerce')

        if 'N_Street' in station or "N'Street" in sheet:
            df_data['water_level_mMSL'] = df_data['water_level'] * 0.3048
            unit = 'mMSL (converted from ft)'
        else:
            df_data['water_level_mMSL'] = df_data['water_level']
            unit = 'mMSL'

        df_data['date'] = pd.to_datetime(
            dict(year=df_data['year'], month=df_data['month'], day=df_data['day']),
            errors='coerce'
        )

        df_clean = df_data.dropna(subset=['date']).copy()
        df_clean['station'] = station
        df_clean['datum_msl_m'] = msl_val
        df_clean['unit'] = unit

        all_wl_data.append(df_clean[['date', 'station', 'water_level_mMSL', 'datum_msl_m', 'unit']])

    return pd.concat(all_wl_data, ignore_index=True)


def parse_discharge(file_path):
    """Processes Discharge Excel Workbook into long format."""
    print("Processing Discharge Data...")
    xls = pd.ExcelFile(file_path)
    all_dis_data = []

    for sheet in xls.sheet_names:
        print(f"  -> Reading Discharge Sheet: {sheet}")
        df_raw = pd.read_excel(xls, sheet_name=sheet, header=None)
        station = clean_station_name(sheet)

        df_data = df_raw.iloc[4:].copy()
        df_data.columns = [str(col).strip() for col in df_raw.iloc[3]]
        
        day_cols = [c for c in df_data.columns if re.match(r'^D\d+$', c, re.IGNORECASE)]
        ym_col = [c for c in df_data.columns if c not in day_cols][0]

        df_melted = pd.melt(
            df_data,
            id_vars=[ym_col],
            value_vars=day_cols,
            var_name='day',
            value_name='discharge_cumecs'
        )

        ym_cleaned = df_melted[ym_col].astype(str).str.replace(r'\D', '', regex=True)
        df_melted['year_num'] = pd.to_numeric(ym_cleaned.str[:4], errors='coerce')
        df_melted['month_num'] = pd.to_numeric(ym_cleaned.str[4:6], errors='coerce')
        df_melted['day'] = df_melted['day'].str.extract(r'(\d+)').astype(int)
        df_melted['discharge_cumecs'] = pd.to_numeric(df_melted['discharge_cumecs'], errors='coerce')

        df_melted['date'] = pd.to_datetime(
            dict(year=df_melted['year_num'], month=df_melted['month_num'], day=df_melted['day']),
            errors='coerce'
        )

        df_clean = df_melted.dropna(subset=['date']).copy()
        df_clean['station'] = station

        all_dis_data.append(df_clean[['date', 'station', 'discharge_cumecs']])

    return pd.concat(all_dis_data, ignore_index=True)


def main():
    rf_path = os.path.join(RAW_DIR, "2015 - 2025 RF.xlsx")
    wl_path = os.path.join(RAW_DIR, "2015 - 2025 WL.xlsx")
    dis_path = os.path.join(RAW_DIR, "2015 - 2025 Discharge.xlsx")

    # Run processing
    df_rf = parse_rainfall(rf_path)
    df_wl = parse_water_level(wl_path)
    df_dis = parse_discharge(dis_path)

    # Save individual clean long-format datasets
    df_rf.to_csv(os.path.join(PROCESSED_DIR, "kelani_daily_rainfall_2015_2025.csv"), index=False)
    df_wl.to_csv(os.path.join(PROCESSED_DIR, "kelani_daily_water_level_2015_2025.csv"), index=False)
    df_dis.to_csv(os.path.join(PROCESSED_DIR, "kelani_daily_discharge_2015_2025.csv"), index=False)

    # Merge into a single master time-series dataset
    print("Building unified master time-series table...")
    df_master = pd.merge(df_rf, df_wl, on=['date', 'station'], how='outer')
    df_master = pd.merge(df_master, df_dis, on=['date', 'station'], how='outer')
    
    df_master.sort_values(by=['date', 'station'], inplace=True)
    master_path = os.path.join(PROCESSED_DIR, "kelani_master_daily_analytics_2015_2025.csv")
    df_master.to_csv(master_path, index=False)

    print("\nPipeline Execution Completed Successfully!")
    print(f"Master output saved to: {master_path}")

# Run pipeline explicitly
main()
```

```python
import os
import re
import pandas as pd
import numpy as np

PROJECT_ROOT = "/Users/ohaseelanka/disaster_analytics_hub"
RAW_DIR = os.path.join(PROJECT_ROOT, "data/raw/flood/kelani")
PROCESSED_DIR = os.path.join(PROJECT_ROOT, "data/processed/flood/kelani")

os.makedirs(PROCESSED_DIR, exist_ok=True)

MONTH_MAP = {
    'JAN': 1, 'FEB': 2, 'MAR': 3, 'APR': 4, 'MAY': 5, 'JUN': 6,
    'JUL': 7, 'AUG': 8, 'SEP': 9, 'OCT': 10, 'NOV': 11, 'DEC': 12
}

STATION_MAP = {
    'Kithulgala': 'Kitulgala',
    'Kn11-Holombuwa': 'Holombuwa',
    'Kn11_Holombuwa': 'Holombuwa',
    'N\'Street': 'N_Street'
}

def clean_station_name(name):
    """Harmonizes station names across sheets."""
    name = re.sub(r'^[Kk][Nn]\d+[-_\s]*', '', str(name)).strip()
    name = name.replace("N'Street", "N_Street").title()
    return STATION_MAP.get(name, name)

def fix_year(year_val):
    """Normalizes 2-digit, 4-digit, or YYMM year values."""
    if pd.isna(year_val):
        return np.nan
    
    # Remove non-numeric characters
    s = re.sub(r'\D', '', str(year_val).split('.')[0])
    if not s:
        return np.nan
    
    # YYMM format (e.g., 1801 -> Year 2018)
    if len(s) == 4 and int(s[:2]) in range(15, 27):
        return 2000 + int(s[:2])
    
    val = int(s)
    # 2-digit year (e.g., 18 -> 2018)
    if val < 100:
        return 2000 + val if val <= 30 else 1900 + val
    
    return val

def parse_rainfall(file_path):
    print("Processing Rainfall Data...")
    xls = pd.ExcelFile(file_path)
    all_rf_data = []

    for sheet in xls.sheet_names:
        df_raw = pd.read_excel(xls, sheet_name=sheet, header=None)
        station = clean_station_name(sheet)
        
        coord_row = df_raw.iloc[1].dropna().astype(str).str.cat(sep=" ")
        x_match = re.search(r'(\d+)\s*X', coord_row)
        y_match = re.search(r'(\d+)\s*Y', coord_row)
        coord_x = float(x_match.group(1)) if x_match else np.nan
        coord_y = float(y_match.group(1)) if y_match else np.nan

        df_data = df_raw.iloc[4:].copy()
        df_data.columns = [str(col).strip() for col in df_raw.iloc[3]]
        
        day_cols = [c for c in df_data.columns if re.match(r'^D\d+$', c, re.IGNORECASE)]
        year_cols = [c for c in df_data.columns if 'YEAR' in c.upper() and c not in day_cols]
        month_cols = [c for c in df_data.columns if 'MONTH' in c.upper() and c not in day_cols]

        if year_cols and month_cols and year_cols[0] != month_cols[0]:
            year_col, month_col = year_cols[0], month_cols[0]
            df_melted = pd.melt(df_data, id_vars=[year_col, month_col], value_vars=day_cols, var_name='day', value_name='rainfall_mm')
            df_melted['raw_year'] = df_melted[year_col]
            df_melted['raw_month'] = df_melted[month_col]
        else:
            ym_col = year_cols[0] if year_cols else df_data.columns[0]
            df_melted = pd.melt(df_data, id_vars=[ym_col], value_vars=day_cols, var_name='day', value_name='rainfall_mm')
            ym_str = df_melted[ym_col].astype(str).str.replace(r'\D', '', regex=True)
            df_melted['raw_year'] = ym_str.str[:2]
            df_melted['raw_month'] = ym_str.str[2:4]

        df_melted['day'] = df_melted['day'].str.extract(r'(\d+)').astype(int)
        df_melted['year_num'] = df_melted['raw_year'].apply(fix_year)
        
        def convert_month(m):
            if pd.isna(m): return np.nan
            m_str = str(m).strip().upper()
            return MONTH_MAP.get(m_str, int(m_str) if m_str.isdigit() and 1 <= int(m_str) <= 12 else np.nan)

        df_melted['month_num'] = df_melted['raw_month'].apply(convert_month)
        df_melted['rainfall_mm'] = pd.to_numeric(df_melted['rainfall_mm'], errors='coerce')
        
        # Datetime construction restricted to valid 2015-2025 window
        df_melted['date'] = pd.to_datetime(
            dict(year=df_melted['year_num'], month=df_melted['month_num'], day=df_melted['day']),
            errors='coerce'
        )
        
        df_clean = df_melted.dropna(subset=['date']).copy()
        df_clean = df_clean[(df_clean['date'].dt.year >= 2015) & (df_clean['date'].dt.year <= 2025)]
        df_clean['station'] = station
        df_clean['coord_x'] = coord_x
        df_clean['coord_y'] = coord_y
        
        all_rf_data.append(df_clean[['date', 'station', 'rainfall_mm', 'coord_x', 'coord_y']])

    return pd.concat(all_rf_data, ignore_index=True)


def parse_water_level(file_path):
    print("Processing Water Level Data...")
    xls = pd.ExcelFile(file_path)
    all_wl_data = []

    for sheet in xls.sheet_names:
        df_raw = pd.read_excel(xls, sheet_name=sheet, header=None)
        station = clean_station_name(sheet)
        
        msl_val = np.nan
        msl_str = str(df_raw.iloc[2, 1]) if df_raw.shape[1] > 1 else ""
        msl_match = re.search(r'([-+]?\d*\.\d+|\d+)', msl_str)
        if msl_match:
            msl_val = float(msl_match.group(1))

        df_data = df_raw.iloc[5:].copy().iloc[:, :4]
        df_data.columns = ['year', 'month', 'day', 'water_level']
        
        df_data['year_num'] = df_data['year'].apply(fix_year)
        df_data['month_num'] = pd.to_numeric(df_data['month'], errors='coerce')
        df_data['day_num'] = pd.to_numeric(df_data['day'], errors='coerce')
        df_data['water_level'] = pd.to_numeric(df_data['water_level'], errors='coerce')

        if 'N_Street' in station:
            df_data['water_level_mMSL'] = df_data['water_level'] * 0.3048
            unit = 'mMSL (converted from ft)'
        else:
            df_data['water_level_mMSL'] = df_data['water_level']
            unit = 'mMSL'

        df_data['date'] = pd.to_datetime(
            dict(year=df_data['year_num'], month=df_data['month_num'], day=df_data['day_num']),
            errors='coerce'
        )

        df_clean = df_data.dropna(subset=['date']).copy()
        df_clean = df_clean[(df_clean['date'].dt.year >= 2015) & (df_clean['date'].dt.year <= 2025)]
        df_clean['station'] = station
        df_clean['datum_msl_m'] = msl_val
        df_clean['unit'] = unit

        all_wl_data.append(df_clean[['date', 'station', 'water_level_mMSL', 'datum_msl_m', 'unit']])

    return pd.concat(all_wl_data, ignore_index=True)


def parse_discharge(file_path):
    print("Processing Discharge Data...")
    xls = pd.ExcelFile(file_path)
    all_dis_data = []

    for sheet in xls.sheet_names:
        df_raw = pd.read_excel(xls, sheet_name=sheet, header=None)
        station = clean_station_name(sheet)

        df_data = df_raw.iloc[4:].copy()
        df_data.columns = [str(col).strip() for col in df_raw.iloc[3]]
        
        day_cols = [c for c in df_data.columns if re.match(r'^D\d+$', c, re.IGNORECASE)]
        ym_col = [c for c in df_data.columns if c not in day_cols][0]

        df_melted = pd.melt(df_data, id_vars=[ym_col], value_vars=day_cols, var_name='day', value_name='discharge_cumecs')
        ym_str = df_melted[ym_col].astype(str).str.replace(r'\D', '', regex=True)

        df_melted['year_num'] = ym_str.str[:4].apply(fix_year)
        df_melted['month_num'] = pd.to_numeric(ym_str.str[4:6], errors='coerce')
        df_melted['day'] = df_melted['day'].str.extract(r'(\d+)').astype(int)
        df_melted['discharge_cumecs'] = pd.to_numeric(df_melted['discharge_cumecs'], errors='coerce')

        df_melted['date'] = pd.to_datetime(
            dict(year=df_melted['year_num'], month=df_melted['month_num'], day=df_melted['day']),
            errors='coerce'
        )

        df_clean = df_melted.dropna(subset=['date']).copy()
        df_clean = df_clean[(df_clean['date'].dt.year >= 2015) & (df_clean['date'].dt.year <= 2025)]
        df_clean['station'] = station

        all_dis_data.append(df_clean[['date', 'station', 'discharge_cumecs']])

    return pd.concat(all_dis_data, ignore_index=True)


def main():
    rf_path = os.path.join(RAW_DIR, "2015 - 2025 RF.xlsx")
    wl_path = os.path.join(RAW_DIR, "2015 - 2025 WL.xlsx")
    dis_path = os.path.join(RAW_DIR, "2015 - 2025 Discharge.xlsx")

    df_rf = parse_rainfall(rf_path)
    df_wl = parse_water_level(wl_path)
    df_dis = parse_discharge(dis_path)

    print("Building unified master time-series table...")
    df_master = pd.merge(df_rf, df_wl, on=['date', 'station'], how='outer')
    df_master = pd.merge(df_master, df_dis, on=['date', 'station'], how='outer')
    
    df_master.sort_values(by=['date', 'station'], inplace=True)
    master_path = os.path.join(PROCESSED_DIR, "kelani_master_daily_analytics_2015_2025.csv")
    df_master.to_csv(master_path, index=False)

    print("\nPipeline Execution Completed Successfully!")
    print(f"Master output saved to: {master_path}")

if __name__ == '__main__':
    main()
```

```python
import os
import re
import pandas as pd
import numpy as np

PROJECT_ROOT = "/Users/ohaseelanka/disaster_analytics_hub"
RAW_DIR = os.path.join(PROJECT_ROOT, "data/raw/flood/kelani")
PROCESSED_DIR = os.path.join(PROJECT_ROOT, "data/processed/flood/kelani")

os.makedirs(PROCESSED_DIR, exist_ok=True)

MONTH_MAP = {
    'JAN': 1, 'FEB': 2, 'MAR': 3, 'APR': 4, 'MAY': 5, 'JUN': 6,
    'JUL': 7, 'AUG': 8, 'SEP': 9, 'OCT': 10, 'NOV': 11, 'DEC': 12
}

STATION_MAP = {
    'Kithulgala': 'Kitulgala',
    'Kn11-Holombuwa': 'Holombuwa',
    'Kn11_Holombuwa': 'Holombuwa',
    'N\'Street': 'N_Street'
}

def clean_station_name(name):
    """Harmonizes station names across sheets."""
    name = re.sub(r'^[Kk][Nn]\d+[-_\s]*', '', str(name)).strip()
    name = name.replace("N'Street", "N_Street").title()
    return STATION_MAP.get(name, name)

def fix_year(year_val):
    """Normalizes 2-digit, 4-digit, or YYMM year values."""
    if pd.isna(year_val):
        return np.nan
    s = re.sub(r'\D', '', str(year_val).split('.')[0])
    if not s:
        return np.nan
    if len(s) == 4 and int(s[:2]) in range(15, 27):
        return 2000 + int(s[:2])
    val = int(s)
    if val < 100:
        return 2000 + val if val <= 30 else 1900 + val
    return val

def parse_rainfall(file_path):
    print("Processing Rainfall Data...")
    xls = pd.ExcelFile(file_path)
    all_rf_data = []

    for sheet in xls.sheet_names:
        df_raw = pd.read_excel(xls, sheet_name=sheet, header=None)
        station = clean_station_name(sheet)
        
        coord_row = df_raw.iloc[1].dropna().astype(str).str.cat(sep=" ")
        x_match = re.search(r'(\d+)\s*X', coord_row)
        y_match = re.search(r'(\d+)\s*Y', coord_row)
        coord_x = float(x_match.group(1)) if x_match else np.nan
        coord_y = float(y_match.group(1)) if y_match else np.nan

        df_data = df_raw.iloc[4:].copy()
        df_data.columns = [str(col).strip() for col in df_raw.iloc[3]]
        
        day_cols = [c for c in df_data.columns if re.match(r'^D\d+$', c, re.IGNORECASE)]
        year_cols = [c for c in df_data.columns if 'YEAR' in c.upper() and c not in day_cols]
        month_cols = [c for c in df_data.columns if 'MONTH' in c.upper() and c not in day_cols]

        if year_cols and month_cols and year_cols[0] != month_cols[0]:
            year_col, month_col = year_cols[0], month_cols[0]
            df_melted = pd.melt(df_data, id_vars=[year_col, month_col], value_vars=day_cols, var_name='day', value_name='rainfall_mm')
            df_melted['raw_year'] = df_melted[year_col]
            df_melted['raw_month'] = df_melted[month_col]
        else:
            ym_col = year_cols[0] if year_cols else df_data.columns[0]
            df_melted = pd.melt(df_data, id_vars=[ym_col], value_vars=day_cols, var_name='day', value_name='rainfall_mm')
            ym_str = df_melted[ym_col].astype(str).str.replace(r'\D', '', regex=True)
            df_melted['raw_year'] = ym_str.str[:2]
            df_melted['raw_month'] = ym_str.str[2:4]

        df_melted['day'] = df_melted['day'].str.extract(r'(\d+)').astype(int)
        df_melted['year_num'] = df_melted['raw_year'].apply(fix_year)
        
        def convert_month(m):
            if pd.isna(m): return np.nan
            m_str = str(m).strip().upper()
            return MONTH_MAP.get(m_str, int(m_str) if m_str.isdigit() and 1 <= int(m_str) <= 12 else np.nan)

        df_melted['month_num'] = df_melted['raw_month'].apply(convert_month)
        df_melted['rainfall_mm'] = pd.to_numeric(df_melted['rainfall_mm'], errors='coerce')
        
        df_melted['date'] = pd.to_datetime(
            dict(year=df_melted['year_num'], month=df_melted['month_num'], day=df_melted['day']),
            errors='coerce'
        )
        
        df_clean = df_melted.dropna(subset=['date']).copy()
        df_clean = df_clean[(df_clean['date'].dt.year >= 2015) & (df_clean['date'].dt.year <= 2025)]
        df_clean['station'] = station
        df_clean['coord_x'] = coord_x
        df_clean['coord_y'] = coord_y
        
        all_rf_data.append(df_clean[['date', 'station', 'rainfall_mm', 'coord_x', 'coord_y']])

    df_concat = pd.concat(all_rf_data, ignore_index=True)
    
    # Ensure 1 row per (date, station) for rainfall
    return df_concat.groupby(['date', 'station'], as_index=False).agg({
        'rainfall_mm': 'mean',
        'coord_x': 'first',
        'coord_y': 'first'
    })


def parse_water_level(file_path):
    print("Processing Water Level Data...")
    xls = pd.ExcelFile(file_path)
    all_wl_data = []

    for sheet in xls.sheet_names:
        df_raw = pd.read_excel(xls, sheet_name=sheet, header=None)
        station = clean_station_name(sheet)
        
        msl_val = np.nan
        msl_str = str(df_raw.iloc[2, 1]) if df_raw.shape[1] > 1 else ""
        msl_match = re.search(r'([-+]?\d*\.\d+|\d+)', msl_str)
        if msl_match:
            msl_val = float(msl_match.group(1))

        df_data = df_raw.iloc[5:].copy().iloc[:, :4]
        df_data.columns = ['year', 'month', 'day', 'water_level']
        
        df_data['year_num'] = df_data['year'].apply(fix_year)
        df_data['month_num'] = pd.to_numeric(df_data['month'], errors='coerce')
        df_data['day_num'] = pd.to_numeric(df_data['day'], errors='coerce')
        df_data['water_level'] = pd.to_numeric(df_data['water_level'], errors='coerce')

        if 'N_Street' in station:
            df_data['water_level_mMSL'] = df_data['water_level'] * 0.3048
            unit = 'mMSL (converted from ft)'
        else:
            df_data['water_level_mMSL'] = df_data['water_level']
            unit = 'mMSL'

        df_data['date'] = pd.to_datetime(
            dict(year=df_data['year_num'], month=df_data['month_num'], day=df_data['day_num']),
            errors='coerce'
        )

        df_clean = df_data.dropna(subset=['date']).copy()
        df_clean = df_clean[(df_clean['date'].dt.year >= 2015) & (df_clean['date'].dt.year <= 2025)]
        df_clean['station'] = station
        df_clean['datum_msl_m'] = msl_val
        df_clean['unit'] = unit

        all_wl_data.append(df_clean[['date', 'station', 'water_level_mMSL', 'datum_msl_m', 'unit']])

    df_concat = pd.concat(all_wl_data, ignore_index=True)
    
    # Aggregate sub-daily readings to daily mean water level
    return df_concat.groupby(['date', 'station'], as_index=False).agg({
        'water_level_mMSL': 'mean',
        'datum_msl_m': 'first',
        'unit': 'first'
    })


def parse_discharge(file_path):
    print("Processing Discharge Data...")
    xls = pd.ExcelFile(file_path)
    all_dis_data = []

    for sheet in xls.sheet_names:
        df_raw = pd.read_excel(xls, sheet_name=sheet, header=None)
        station = clean_station_name(sheet)

        df_data = df_raw.iloc[4:].copy()
        df_data.columns = [str(col).strip() for col in df_raw.iloc[3]]
        
        day_cols = [c for c in df_data.columns if re.match(r'^D\d+$', c, re.IGNORECASE)]
        ym_col = [c for c in df_data.columns if c not in day_cols][0]

        df_melted = pd.melt(df_data, id_vars=[ym_col], value_vars=day_cols, var_name='day', value_name='discharge_cumecs')
        ym_str = df_melted[ym_col].astype(str).str.replace(r'\D', '', regex=True)

        df_melted['year_num'] = ym_str.str[:4].apply(fix_year)
        df_melted['month_num'] = pd.to_numeric(ym_str.str[4:6], errors='coerce')
        df_melted['day'] = df_melted['day'].str.extract(r'(\d+)').astype(int)
        df_melted['discharge_cumecs'] = pd.to_numeric(df_melted['discharge_cumecs'], errors='coerce')

        df_melted['date'] = pd.to_datetime(
            dict(year=df_melted['year_num'], month=df_melted['month_num'], day=df_melted['day']),
            errors='coerce'
        )

        df_clean = df_melted.dropna(subset=['date']).copy()
        df_clean = df_clean[(df_clean['date'].dt.year >= 2015) & (df_clean['date'].dt.year <= 2025)]
        df_clean['station'] = station

        all_dis_data.append(df_clean[['date', 'station', 'discharge_cumecs']])

    df_concat = pd.concat(all_dis_data, ignore_index=True)
    
    return df_concat.groupby(['date', 'station'], as_index=False).agg({
        'discharge_cumecs': 'mean'
    })


def main():
    rf_path = os.path.join(RAW_DIR, "2015 - 2025 RF.xlsx")
    wl_path = os.path.join(RAW_DIR, "2015 - 2025 WL.xlsx")
    dis_path = os.path.join(RAW_DIR, "2015 - 2025 Discharge.xlsx")

    df_rf = parse_rainfall(rf_path)
    df_wl = parse_water_level(wl_path)
    df_dis = parse_discharge(dis_path)

    print("Building unified daily master table...")
    df_master = pd.merge(df_rf, df_wl, on=['date', 'station'], how='outer')
    df_master = pd.merge(df_master, df_dis, on=['date', 'station'], how='outer')
    
    # Broadcast station coordinates across missing dates
    coords_map = df_master.dropna(subset=['coord_x', 'coord_y']).groupby('station')[['coord_x', 'coord_y']].first().to_dict()
    df_master['coord_x'] = df_master['coord_x'].fillna(df_master['station'].map(lambda s: coords_map['coord_x'].get(s, np.nan)))
    df_master['coord_y'] = df_master['coord_y'].fillna(df_master['station'].map(lambda s: coords_map['coord_y'].get(s, np.nan)))

    df_master.sort_values(by=['date', 'station'], inplace=True)
    master_path = os.path.join(PROCESSED_DIR, "kelani_master_daily_analytics_2015_2025.csv")
    df_master.to_csv(master_path, index=False)

    print("\nDaily Aggregation Complete!")
    print(f"Master file saved to: {master_path}")

if __name__ == '__main__':
    main()
```

```python
import os
import re
import pandas as pd
import numpy as np

PROJECT_ROOT = "/Users/ohaseelanka/disaster_analytics_hub"
RAW_DIR = os.path.join(PROJECT_ROOT, "data/raw/flood/kelani")
PROCESSED_DIR = os.path.join(PROJECT_ROOT, "data/processed/flood/kelani")

os.makedirs(PROCESSED_DIR, exist_ok=True)

MONTH_MAP = {
    'JAN': 1, 'FEB': 2, 'MAR': 3, 'APR': 4, 'MAY': 5, 'JUN': 6,
    'JUL': 7, 'AUG': 8, 'SEP': 9, 'OCT': 10, 'NOV': 11, 'DEC': 12
}

STATION_MAP = {
    'Kithulgala': 'Kitulgala',
    'Kn11-Holombuwa': 'Holombuwa',
    'Kn11_Holombuwa': 'Holombuwa',
    'N\'Street': 'N_Street',
    'Nstreet': 'N_Street'
}

STATION_COORDS = {
    'Kitulgala': (184500.0, 171800.0),
    'Norwood': (198200.0, 166400.0),
    'Deraniyagala': (168300.0, 168100.0),
    'Glencourse': (132100.0, 170900.0),
    'Hanwella': (116300.0, 172100.0),
    'Holombuwa': (142000.0, 190200.0),
    'N_Street': (100800.0, 174800.0)
}

def clean_station_name(name):
    name = re.sub(r'^[Kk][Nn]\d+[-_\s]*', '', str(name)).strip()
    name = name.replace("N'Street", "N_Street").title()
    return STATION_MAP.get(name, name)

def fix_year(year_val):
    if pd.isna(year_val):
        return np.nan
    s = re.sub(r'\D', '', str(year_val).split('.')[0])
    if not s:
        return np.nan
    if len(s) == 4 and int(s[:2]) in range(15, 27):
        return 2000 + int(s[:2])
    val = int(s)
    if val < 100:
        return 2000 + val if val <= 30 else 1900 + val
    return val

def parse_rainfall(file_path):
    print("Processing Rainfall Data...")
    xls = pd.ExcelFile(file_path)
    all_rf_data = []

    for sheet in xls.sheet_names:
        df_raw = pd.read_excel(xls, sheet_name=sheet, header=None)
        station = clean_station_name(sheet)
        
        coord_row = df_raw.iloc[1].dropna().astype(str).str.cat(sep=" ")
        x_match = re.search(r'(\d+)\s*X', coord_row)
        y_match = re.search(r'(\d+)\s*Y', coord_row)
        coord_x = float(x_match.group(1)) if x_match else np.nan
        coord_y = float(y_match.group(1)) if y_match else np.nan

        df_data = df_raw.iloc[4:].copy()
        df_data.columns = [str(col).strip() for col in df_raw.iloc[3]]
        
        day_cols = [c for c in df_data.columns if re.match(r'^D\d+$', c, re.IGNORECASE)]
        year_cols = [c for c in df_data.columns if 'YEAR' in c.upper() and c not in day_cols]
        month_cols = [c for c in df_data.columns if 'MONTH' in c.upper() and c not in day_cols]

        if year_cols and month_cols and year_cols[0] != month_cols[0]:
            year_col, month_col = year_cols[0], month_cols[0]
            df_melted = pd.melt(df_data, id_vars=[year_col, month_col], value_vars=day_cols, var_name='day', value_name='rainfall_mm')
            df_melted['raw_year'] = df_melted[year_col]
            df_melted['raw_month'] = df_melted[month_col]
        else:
            ym_col = year_cols[0] if year_cols else df_data.columns[0]
            df_melted = pd.melt(df_data, id_vars=[ym_col], value_vars=day_cols, var_name='day', value_name='rainfall_mm')
            ym_str = df_melted[ym_col].astype(str).str.replace(r'\D', '', regex=True)
            df_melted['raw_year'] = ym_str.str[:2]
            df_melted['raw_month'] = ym_str.str[2:4]

        df_melted['day'] = df_melted['day'].str.extract(r'(\d+)').astype(int)
        df_melted['year_num'] = df_melted['raw_year'].apply(fix_year)
        
        def convert_month(m):
            if pd.isna(m): return np.nan
            m_str = str(m).strip().upper()
            return MONTH_MAP.get(m_str, int(m_str) if m_str.isdigit() and 1 <= int(m_str) <= 12 else np.nan)

        df_melted['month_num'] = df_melted['raw_month'].apply(convert_month)
        df_melted['rainfall_mm'] = pd.to_numeric(df_melted['rainfall_mm'], errors='coerce')
        
        df_melted['date'] = pd.to_datetime(
            dict(year=df_melted['year_num'], month=df_melted['month_num'], day=df_melted['day']),
            errors='coerce'
        )
        
        df_clean = df_melted.dropna(subset=['date']).copy()
        df_clean['date'] = df_clean['date'].dt.normalize()
        df_clean = df_clean[(df_clean['date'].dt.year >= 2015) & (df_clean['date'].dt.year <= 2025)]
        df_clean['station'] = station
        df_clean['coord_x'] = coord_x
        df_clean['coord_y'] = coord_y
        
        all_rf_data.append(df_clean[['date', 'station', 'rainfall_mm', 'coord_x', 'coord_y']])

    df_concat = pd.concat(all_rf_data, ignore_index=True)
    return df_concat.groupby(['date', 'station'], as_index=False).agg({
        'rainfall_mm': 'mean',
        'coord_x': 'first',
        'coord_y': 'first'
    })

def parse_water_level(file_path):
    print("Processing Water Level Data...")
    xls = pd.ExcelFile(file_path)
    all_wl_data = []

    for sheet in xls.sheet_names:
        df_raw = pd.read_excel(xls, sheet_name=sheet, header=None)
        station = clean_station_name(sheet)
        
        msl_val = np.nan
        msl_str = str(df_raw.iloc[2, 1]) if df_raw.shape[1] > 1 else ""
        msl_match = re.search(r'([-+]?\d*\.\d+|\d+)', msl_str)
        if msl_match:
            msl_val = float(msl_match.group(1))

        df_data = df_raw.iloc[5:].copy().iloc[:, :4]
        df_data.columns = ['year', 'month', 'day', 'water_level']
        
        df_data['year_num'] = df_data['year'].apply(fix_year)
        df_data['month_num'] = pd.to_numeric(df_data['month'], errors='coerce')
        df_data['day_num'] = pd.to_numeric(df_data['day'], errors='coerce')
        df_data['water_level'] = pd.to_numeric(df_data['water_level'], errors='coerce')

        if 'N_Street' in station:
            df_data['water_level_mMSL'] = df_data['water_level'] * 0.3048
            unit = 'mMSL (converted from ft)'
        else:
            df_data['water_level_mMSL'] = df_data['water_level']
            unit = 'mMSL'

        df_data['date'] = pd.to_datetime(
            dict(year=df_data['year_num'], month=df_data['month_num'], day=df_data['day_num']),
            errors='coerce'
        )

        df_clean = df_data.dropna(subset=['date']).copy()
        df_clean['date'] = df_clean['date'].dt.normalize()
        df_clean = df_clean[(df_clean['date'].dt.year >= 2015) & (df_clean['date'].dt.year <= 2025)]
        df_clean['station'] = station
        df_clean['datum_msl_m'] = msl_val
        df_clean['unit'] = unit

        all_wl_data.append(df_clean[['date', 'station', 'water_level_mMSL', 'datum_msl_m', 'unit']])

    df_concat = pd.concat(all_wl_data, ignore_index=True)
    return df_concat.groupby(['date', 'station'], as_index=False).agg({
        'water_level_mMSL': 'mean',
        'datum_msl_m': 'first',
        'unit': 'first'
    })

def parse_discharge(file_path):
    print("Processing Discharge Data...")
    xls = pd.ExcelFile(file_path)
    all_dis_data = []

    for sheet in xls.sheet_names:
        df_raw = pd.read_excel(xls, sheet_name=sheet, header=None)
        station = clean_station_name(sheet)

        df_data = df_raw.iloc[4:].copy()
        df_data.columns = [str(col).strip() for col in df_raw.iloc[3]]
        
        day_cols = [c for c in df_data.columns if re.match(r'^D\d+$', c, re.IGNORECASE)]
        ym_col = [c for c in df_data.columns if c not in day_cols][0]

        df_melted = pd.melt(df_data, id_vars=[ym_col], value_vars=day_cols, var_name='day', value_name='discharge_cumecs')
        ym_str = df_melted[ym_col].astype(str).str.replace(r'\D', '', regex=True)

        df_melted['year_num'] = ym_str.str[:4].apply(fix_year)
        df_melted['month_num'] = pd.to_numeric(ym_str.str[4:6], errors='coerce')
        df_melted['day'] = df_melted['day'].str.extract(r'(\d+)').astype(int)
        df_melted['discharge_cumecs'] = pd.to_numeric(df_melted['discharge_cumecs'], errors='coerce')

        df_melted['date'] = pd.to_datetime(
            dict(year=df_melted['year_num'], month=df_melted['month_num'], day=df_melted['day']),
            errors='coerce'
        )

        df_clean = df_melted.dropna(subset=['date']).copy()
        df_clean['date'] = df_clean['date'].dt.normalize()
        df_clean = df_clean[(df_clean['date'].dt.year >= 2015) & (df_clean['date'].dt.year <= 2025)]
        df_clean['station'] = station

        all_dis_data.append(df_clean[['date', 'station', 'discharge_cumecs']])

    df_concat = pd.concat(all_dis_data, ignore_index=True)
    return df_concat.groupby(['date', 'station'], as_index=False).agg({
        'discharge_cumecs': 'mean'
    })

def main():
    rf_path = os.path.join(RAW_DIR, "2015 - 2025 RF.xlsx")
    wl_path = os.path.join(RAW_DIR, "2015 - 2025 WL.xlsx")
    dis_path = os.path.join(RAW_DIR, "2015 - 2025 Discharge.xlsx")

    df_rf = parse_rainfall(rf_path)
    df_wl = parse_water_level(wl_path)
    df_dis = parse_discharge(dis_path)

    print("Building unified daily master table...")
    df_master = pd.merge(df_rf, df_wl, on=['date', 'station'], how='outer')
    df_master = pd.merge(df_master, df_dis, on=['date', 'station'], how='outer')
    
    # Fill coordinates from parsed rainfall data
    extracted_coords = df_master.dropna(subset=['coord_x', 'coord_y']).groupby('station')[['coord_x', 'coord_y']].first().to_dict()
    df_master['coord_x'] = df_master['coord_x'].fillna(df_master['station'].map(lambda s: extracted_coords['coord_x'].get(s, np.nan)))
    df_master['coord_y'] = df_master['coord_y'].fillna(df_master['station'].map(lambda s: extracted_coords['coord_y'].get(s, np.nan)))

    # Fallback to static station coordinates dictionary
    df_master['coord_x'] = df_master['coord_x'].fillna(df_master['station'].map(lambda s: STATION_COORDS.get(s, (np.nan, np.nan))[0]))
    df_master['coord_y'] = df_master['coord_y'].fillna(df_master['station'].map(lambda s: STATION_COORDS.get(s, (np.nan, np.nan))[1]))

    df_master.sort_values(by=['date', 'station'], inplace=True)
    master_path = os.path.join(PROCESSED_DIR, "kelani_master_daily_analytics_2015_2025.csv")
    df_master.to_csv(master_path, index=False)

    print("\nDaily Aggregation Complete!")
    print(f"Master file saved to: {master_path}")

if __name__ == '__main__':
    main()
```

```python
import os
import re
import pandas as pd
import numpy as np

PROJECT_ROOT = "/Users/ohaseelanka/disaster_analytics_hub"
RAW_DIR = os.path.join(PROJECT_ROOT, "data/raw/flood/kelani")
PROCESSED_DIR = os.path.join(PROJECT_ROOT, "data/processed/flood/kelani")

os.makedirs(PROCESSED_DIR, exist_ok=True)

MONTH_MAP = {
    'JAN': 1, 'FEB': 2, 'MAR': 3, 'APR': 4, 'MAY': 5, 'JUN': 6,
    'JUL': 7, 'AUG': 8, 'SEP': 9, 'OCT': 10, 'NOV': 11, 'DEC': 12
}

STATION_MAP = {
    'Kithulgala': 'Kitulgala',
    'Kn11-Holombuwa': 'Holombuwa',
    'Kn11_Holombuwa': 'Holombuwa',
    'N\'Street': 'N_Street',
    'Nstreet': 'N_Street'
}

STATION_COORDS = {
    'Kitulgala': (184500.0, 171800.0),
    'Norwood': (198200.0, 166400.0),
    'Deraniyagala': (168300.0, 168100.0),
    'Glencourse': (132100.0, 170900.0),
    'Hanwella': (116300.0, 172100.0),
    'Holombuwa': (142000.0, 190200.0),
    'N_Street': (100800.0, 174800.0)
}

def clean_station_name(name):
    name = re.sub(r'^[Kk][Nn]\d+[-_\s]*', '', str(name)).strip()
    name = name.replace("N'Street", "N_Street").title()
    return STATION_MAP.get(name, name).strip()

def fix_year(year_val):
    if pd.isna(year_val):
        return np.nan
    s = re.sub(r'\D', '', str(year_val).split('.')[0])
    if not s:
        return np.nan
    val = int(s)
    if val >= 2000 and val <= 2030:
        return val
    if val < 100:
        return 2000 + val if val <= 30 else 1900 + val
    return val

def parse_rainfall(file_path):
    print("Processing Rainfall Data...")
    xls = pd.ExcelFile(file_path)
    all_rf_data = []

    for sheet in xls.sheet_names:
        df_raw = pd.read_excel(xls, sheet_name=sheet, header=None)
        station = clean_station_name(sheet)
        
        coord_row = df_raw.iloc[1].dropna().astype(str).str.cat(sep=" ")
        x_match = re.search(r'(\d+)\s*X', coord_row)
        y_match = re.search(r'(\d+)\s*Y', coord_row)
        coord_x = float(x_match.group(1)) if x_match else np.nan
        coord_y = float(y_match.group(1)) if y_match else np.nan

        df_data = df_raw.iloc[4:].copy()
        df_data.columns = [str(col).strip() for col in df_raw.iloc[3]]
        
        day_cols = [c for c in df_data.columns if re.match(r'^D\d+$', c, re.IGNORECASE)]
        year_cols = [c for c in df_data.columns if 'YEAR' in c.upper() and c not in day_cols]
        month_cols = [c for c in df_data.columns if 'MONTH' in c.upper() and c not in day_cols]

        if year_cols and month_cols and year_cols[0] != month_cols[0]:
            year_col, month_col = year_cols[0], month_cols[0]
            df_melted = pd.melt(df_data, id_vars=[year_col, month_col], value_vars=day_cols, var_name='day', value_name='rainfall_mm')
            df_melted['raw_year'] = df_melted[year_col]
            df_melted['raw_month'] = df_melted[month_col]
        else:
            ym_col = year_cols[0] if year_cols else df_data.columns[0]
            df_melted = pd.melt(df_data, id_vars=[ym_col], value_vars=day_cols, var_name='day', value_name='rainfall_mm')
            
            s_clean = df_melted[ym_col].astype(str).str.replace(r'\D', '', regex=True)
            df_melted['raw_year'] = np.where(s_clean.str.len() >= 4, s_clean.str[:4], s_clean.str[:2])
            df_melted['raw_month'] = np.where(s_clean.str.len() >= 4, s_clean.str[4:6], s_clean.str[2:4])

        df_melted['day'] = df_melted['day'].str.extract(r'(\d+)').astype(int)
        df_melted['year_num'] = df_melted['raw_year'].apply(fix_year)
        
        def convert_month(m):
            if pd.isna(m): return np.nan
            m_str = str(m).strip().upper()
            return MONTH_MAP.get(m_str, int(m_str) if m_str.isdigit() and 1 <= int(m_str) <= 12 else np.nan)

        df_melted['month_num'] = df_melted['raw_month'].apply(convert_month)
        df_melted['rainfall_mm'] = pd.to_numeric(df_melted['rainfall_mm'], errors='coerce')
        
        df_melted['date'] = pd.to_datetime(
            dict(year=df_melted['year_num'], month=df_melted['month_num'], day=df_melted['day']),
            errors='coerce'
        )
        
        df_clean = df_melted.dropna(subset=['date']).copy()
        df_clean['date'] = df_clean['date'].dt.strftime('%Y-%m-%d')
        df_clean = df_clean[(df_clean['date'] >= '2015-01-01') & (df_clean['date'] <= '2025-12-31')]
        df_clean['station'] = station
        df_clean['coord_x'] = coord_x
        df_clean['coord_y'] = coord_y
        
        all_rf_data.append(df_clean[['date', 'station', 'rainfall_mm', 'coord_x', 'coord_y']])

    if not all_rf_data:
        return pd.DataFrame(columns=['date', 'station', 'rainfall_mm', 'coord_x', 'coord_y'])

    df_concat = pd.concat(all_rf_data, ignore_index=True)
    return df_concat.groupby(['date', 'station'], as_index=False).agg({
        'rainfall_mm': 'mean',
        'coord_x': 'first',
        'coord_y': 'first'
    })

def parse_water_level(file_path):
    print("Processing Water Level Data...")
    xls = pd.ExcelFile(file_path)
    all_wl_data = []

    for sheet in xls.sheet_names:
        df_raw = pd.read_excel(xls, sheet_name=sheet, header=None)
        station = clean_station_name(sheet)
        
        msl_val = np.nan
        msl_str = str(df_raw.iloc[2, 1]) if df_raw.shape[1] > 1 else ""
        msl_match = re.search(r'([-+]?\d*\.\d+|\d+)', msl_str)
        if msl_match:
            msl_val = float(msl_match.group(1))

        df_data = df_raw.iloc[5:].copy().iloc[:, :4]
        df_data.columns = ['year', 'month', 'day', 'water_level']
        
        df_data['year_num'] = df_data['year'].apply(fix_year)
        df_data['month_num'] = pd.to_numeric(df_data['month'], errors='coerce')
        df_data['day_num'] = pd.to_numeric(df_data['day'], errors='coerce')
        df_data['water_level'] = pd.to_numeric(df_data['water_level'], errors='coerce')

        if 'N_Street' in station:
            df_data['water_level_mMSL'] = df_data['water_level'] * 0.3048
            unit = 'mMSL (converted from ft)'
        else:
            df_data['water_level_mMSL'] = df_data['water_level']
            unit = 'mMSL'

        df_data['date'] = pd.to_datetime(
            dict(year=df_data['year_num'], month=df_data['month_num'], day=df_data['day_num']),
            errors='coerce'
        )

        df_clean = df_data.dropna(subset=['date']).copy()
        df_clean['date'] = df_clean['date'].dt.strftime('%Y-%m-%d')
        df_clean = df_clean[(df_clean['date'] >= '2015-01-01') & (df_clean['date'] <= '2025-12-31')]
        df_clean['station'] = station
        df_clean['datum_msl_m'] = msl_val
        df_clean['unit'] = unit

        all_wl_data.append(df_clean[['date', 'station', 'water_level_mMSL', 'datum_msl_m', 'unit']])

    if not all_wl_data:
        return pd.DataFrame(columns=['date', 'station', 'water_level_mMSL', 'datum_msl_m', 'unit'])

    df_concat = pd.concat(all_wl_data, ignore_index=True)
    return df_concat.groupby(['date', 'station'], as_index=False).agg({
        'water_level_mMSL': 'mean',
        'datum_msl_m': 'first',
        'unit': 'first'
    })

def parse_discharge(file_path):
    print("Processing Discharge Data...")
    xls = pd.ExcelFile(file_path)
    all_dis_data = []

    for sheet in xls.sheet_names:
        df_raw = pd.read_excel(xls, sheet_name=sheet, header=None)
        station = clean_station_name(sheet)

        df_data = df_raw.iloc[4:].copy()
        df_data.columns = [str(col).strip() for col in df_raw.iloc[3]]
        
        day_cols = [c for c in df_data.columns if re.match(r'^D\d+$', c, re.IGNORECASE)]
        if not day_cols:
            continue
            
        ym_col = [c for c in df_data.columns if c not in day_cols][0]

        df_melted = pd.melt(df_data, id_vars=[ym_col], value_vars=day_cols, var_name='day', value_name='discharge_cumecs')
        s_clean = df_melted[ym_col].astype(str).str.replace(r'\D', '', regex=True)

        raw_yr = np.where(s_clean.str.len() >= 4, s_clean.str[:4], s_clean.str[:2])
        raw_mo = np.where(s_clean.str.len() >= 4, s_clean.str[4:6], s_clean.str[2:4])

        df_melted['year_num'] = pd.Series(raw_yr).apply(fix_year)
        df_melted['month_num'] = pd.to_numeric(raw_mo, errors='coerce')
        df_melted['day'] = df_melted['day'].str.extract(r'(\d+)').astype(int)
        df_melted['discharge_cumecs'] = pd.to_numeric(df_melted['discharge_cumecs'], errors='coerce')

        df_melted['date'] = pd.to_datetime(
            dict(year=df_melted['year_num'], month=df_melted['month_num'], day=df_melted['day']),
            errors='coerce'
        )

        df_clean = df_melted.dropna(subset=['date']).copy()
        df_clean['date'] = df_clean['date'].dt.strftime('%Y-%m-%d')
        df_clean = df_clean[(df_clean['date'] >= '2015-01-01') & (df_clean['date'] <= '2025-12-31')]
        df_clean['station'] = station

        all_dis_data.append(df_clean[['date', 'station', 'discharge_cumecs']])

    if not all_dis_data:
        return pd.DataFrame(columns=['date', 'station', 'discharge_cumecs'])

    df_concat = pd.concat(all_dis_data, ignore_index=True)
    return df_concat.groupby(['date', 'station'], as_index=False).agg({
        'discharge_cumecs': 'mean'
    })

def main():
    rf_path = os.path.join(RAW_DIR, "2015 - 2025 RF.xlsx")
    wl_path = os.path.join(RAW_DIR, "2015 - 2025 WL.xlsx")
    dis_path = os.path.join(RAW_DIR, "2015 - 2025 Discharge.xlsx")

    df_rf = parse_rainfall(rf_path)
    df_wl = parse_water_level(wl_path)
    df_dis = parse_discharge(dis_path)

    print(f"Parsed Rainfall Rows: {len(df_rf)}")
    print(f"Parsed Water Level Rows: {len(df_wl)}")
    print(f"Parsed Discharge Rows: {len(df_dis)}")

    print("Building unified daily master table...")
    df_master = pd.merge(df_rf, df_wl, on=['date', 'station'], how='outer')
    df_master = pd.merge(df_master, df_dis, on=['date', 'station'], how='outer')

    # Station coordinates backfill
    df_master['coord_x'] = df_master['coord_x'].fillna(df_master['station'].map(lambda s: STATION_COORDS.get(s, (np.nan, np.nan))[0]))
    df_master['coord_y'] = df_master['coord_y'].fillna(df_master['station'].map(lambda s: STATION_COORDS.get(s, (np.nan, np.nan))[1]))

    df_master.sort_values(by=['date', 'station'], inplace=True)
    master_path = os.path.join(PROCESSED_DIR, "kelani_master_daily_analytics_2015_2025.csv")
    df_master.to_csv(master_path, index=False)

    print("\nDaily Aggregation Complete!")
    print(f"Master file saved to: {master_path}")

if __name__ == '__main__':
    main()
```

```python
import os
import re
import pandas as pd
import numpy as np

PROJECT_ROOT = "/Users/ohaseelanka/disaster_analytics_hub"
RAW_DIR = os.path.join(PROJECT_ROOT, "data/raw/flood/kelani")
PROCESSED_DIR = os.path.join(PROJECT_ROOT, "data/processed/flood/kelani")

os.makedirs(PROCESSED_DIR, exist_ok=True)

MONTH_MAP = {
    'JAN': 1, 'FEB': 2, 'MAR': 3, 'APR': 4, 'MAY': 5, 'JUN': 6,
    'JUL': 7, 'AUG': 8, 'SEP': 9, 'OCT': 10, 'NOV': 11, 'DEC': 12
}

STATION_MAP = {
    'Kithulgala': 'Kitulgala',
    'Kn11-Holombuwa': 'Holombuwa',
    'Kn11_Holombuwa': 'Holombuwa',
    'N\'Street': 'N_Street',
    'Nstreet': 'N_Street'
}

STATION_COORDS = {
    'Kitulgala': (184500.0, 171800.0),
    'Norwood': (198200.0, 166400.0),
    'Deraniyagala': (168300.0, 168100.0),
    'Glencourse': (132100.0, 170900.0),
    'Hanwella': (116300.0, 172100.0),
    'Holombuwa': (142000.0, 190200.0),
    'N_Street': (100800.0, 174800.0)
}

def clean_station_name(name):
    name = re.sub(r'^[Kk][Nn]\d+[-_\s]*', '', str(name)).strip()
    name = name.replace("N'Street", "N_Street").title()
    return STATION_MAP.get(name, name).strip()

def fix_year(year_val):
    if pd.isna(year_val):
        return np.nan
    s = re.sub(r'\D', '', str(year_val).split('.')[0])
    if not s:
        return np.nan
    val = int(s)
    if val >= 2000 and val <= 2030:
        return val
    if val < 100:
        return 2000 + val if val <= 30 else 1900 + val
    return val

def parse_rainfall(file_path):
    print("Processing Rainfall Data...")
    xls = pd.ExcelFile(file_path)
    all_rf_data = []

    for sheet in xls.sheet_names:
        df_raw = pd.read_excel(xls, sheet_name=sheet, header=None)
        station = clean_station_name(sheet)
        
        coord_row = df_raw.iloc[1].dropna().astype(str).str.cat(sep=" ")
        x_match = re.search(r'(\d+)\s*X', coord_row)
        y_match = re.search(r'(\d+)\s*Y', coord_row)
        coord_x = float(x_match.group(1)) if x_match else np.nan
        coord_y = float(y_match.group(1)) if y_match else np.nan

        df_data = df_raw.iloc[4:].copy()
        df_data.columns = [str(col).strip() for col in df_raw.iloc[3]]
        
        day_cols = [c for c in df_data.columns if re.match(r'^D\d+$', c, re.IGNORECASE)]
        year_cols = [c for c in df_data.columns if 'YEAR' in c.upper() and c not in day_cols]
        month_cols = [c for c in df_data.columns if 'MONTH' in c.upper() and c not in day_cols]

        if year_cols and month_cols and year_cols[0] != month_cols[0]:
            year_col, month_col = year_cols[0], month_cols[0]
            df_melted = pd.melt(df_data, id_vars=[year_col, month_col], value_vars=day_cols, var_name='day', value_name='rainfall_mm')
            df_melted['raw_year'] = df_melted[year_col]
            df_melted['raw_month'] = df_melted[month_col]
        else:
            ym_col = year_cols[0] if year_cols else df_data.columns[0]
            df_melted = pd.melt(df_data, id_vars=[ym_col], value_vars=day_cols, var_name='day', value_name='rainfall_mm')
            
            s_clean = df_melted[ym_col].astype(str).str.replace(r'\D', '', regex=True)
            df_melted['raw_year'] = np.where(s_clean.str.len() >= 4, s_clean.str[:4], s_clean.str[:2])
            df_melted['raw_month'] = np.where(s_clean.str.len() >= 4, s_clean.str[4:6], s_clean.str[2:4])

        df_melted['day'] = df_melted['day'].str.extract(r'(\d+)').astype(int)
        df_melted['year_num'] = df_melted['raw_year'].apply(fix_year)
        
        def convert_month(m):
            if pd.isna(m): return np.nan
            m_str = str(m).strip().upper()
            return MONTH_MAP.get(m_str, int(m_str) if m_str.isdigit() and 1 <= int(m_str) <= 12 else np.nan)

        df_melted['month_num'] = df_melted['raw_month'].apply(convert_month)
        df_melted['rainfall_mm'] = pd.to_numeric(df_melted['rainfall_mm'], errors='coerce')
        
        df_melted['date'] = pd.to_datetime(
            dict(year=df_melted['year_num'], month=df_melted['month_num'], day=df_melted['day']),
            errors='coerce'
        )
        
        df_clean = df_melted.dropna(subset=['date']).copy()
        df_clean['date'] = df_clean['date'].dt.strftime('%Y-%m-%d')
        df_clean = df_clean[(df_clean['date'] >= '2015-01-01') & (df_clean['date'] <= '2025-12-31')]
        df_clean['station'] = station
        df_clean['coord_x'] = coord_x
        df_clean['coord_y'] = coord_y
        
        all_rf_data.append(df_clean[['date', 'station', 'rainfall_mm', 'coord_x', 'coord_y']])

    if not all_rf_data:
        return pd.DataFrame(columns=['date', 'station', 'rainfall_mm', 'coord_x', 'coord_y'])

    df_concat = pd.concat(all_rf_data, ignore_index=True)
    return df_concat.groupby(['date', 'station'], as_index=False).agg({
        'rainfall_mm': 'mean',
        'coord_x': 'first',
        'coord_y': 'first'
    })

def parse_water_level(file_path):
    print("Processing Water Level Data...")
    xls = pd.ExcelFile(file_path)
    all_wl_data = []

    for sheet in xls.sheet_names:
        df_raw = pd.read_excel(xls, sheet_name=sheet, header=None)
        station = clean_station_name(sheet)
        
        msl_val = np.nan
        msl_str = str(df_raw.iloc[2, 1]) if df_raw.shape[1] > 1 else ""
        msl_match = re.search(r'([-+]?\d*\.\d+|\d+)', msl_str)
        if msl_match:
            msl_val = float(msl_match.group(1))

        df_data = df_raw.iloc[5:].copy().iloc[:, :4]
        df_data.columns = ['year', 'month', 'day', 'water_level']
        
        df_data['year_num'] = df_data['year'].apply(fix_year)
        df_data['month_num'] = pd.to_numeric(df_data['month'], errors='coerce')
        df_data['day_num'] = pd.to_numeric(df_data['day'], errors='coerce')
        df_data['water_level'] = pd.to_numeric(df_data['water_level'], errors='coerce')

        if 'N_Street' in station:
            df_data['water_level_mMSL'] = df_data['water_level'] * 0.3048
            unit = 'mMSL (converted from ft)'
        else:
            df_data['water_level_mMSL'] = df_data['water_level']
            unit = 'mMSL'

        df_data['date'] = pd.to_datetime(
            dict(year=df_data['year_num'], month=df_data['month_num'], day=df_data['day_num']),
            errors='coerce'
        )

        df_clean = df_data.dropna(subset=['date']).copy()
        df_clean['date'] = df_clean['date'].dt.strftime('%Y-%m-%d')
        df_clean = df_clean[(df_clean['date'] >= '2015-01-01') & (df_clean['date'] <= '2025-12-31')]
        df_clean['station'] = station
        df_clean['datum_msl_m'] = msl_val
        df_clean['unit'] = unit

        all_wl_data.append(df_clean[['date', 'station', 'water_level_mMSL', 'datum_msl_m', 'unit']])

    if not all_wl_data:
        return pd.DataFrame(columns=['date', 'station', 'water_level_mMSL', 'datum_msl_m', 'unit'])

    df_concat = pd.concat(all_wl_data, ignore_index=True)
    return df_concat.groupby(['date', 'station'], as_index=False).agg({
        'water_level_mMSL': 'mean',
        'datum_msl_m': 'first',
        'unit': 'first'
    })

def parse_discharge(file_path):
    print("Processing Discharge Data...")
    xls = pd.ExcelFile(file_path)
    all_dis_data = []

    for sheet in xls.sheet_names:
        df_raw = pd.read_excel(xls, sheet_name=sheet, header=None)
        station = clean_station_name(sheet)

        df_data = df_raw.iloc[4:].copy()
        df_data.columns = [str(col).strip() for col in df_raw.iloc[3]]
        
        day_cols = [c for c in df_data.columns if re.match(r'^D\d+$', c, re.IGNORECASE)]
        if not day_cols:
            continue
            
        ym_col = [c for c in df_data.columns if c not in day_cols][0]

        df_melted = pd.melt(df_data, id_vars=[ym_col], value_vars=day_cols, var_name='day', value_name='discharge_cumecs')
        s_clean = df_melted[ym_col].astype(str).str.replace(r'\D', '', regex=True)

        raw_yr = np.where(s_clean.str.len() >= 4, s_clean.str[:4], s_clean.str[:2])
        raw_mo = np.where(s_clean.str.len() >= 4, s_clean.str[4:6], s_clean.str[2:4])

        df_melted['year_num'] = pd.Series(raw_yr).apply(fix_year)
        df_melted['month_num'] = pd.to_numeric(raw_mo, errors='coerce')
        df_melted['day'] = df_melted['day'].str.extract(r'(\d+)').astype(int)
        df_melted['discharge_cumecs'] = pd.to_numeric(df_melted['discharge_cumecs'], errors='coerce')

        df_melted['date'] = pd.to_datetime(
            dict(year=df_melted['year_num'], month=df_melted['month_num'], day=df_melted['day']),
            errors='coerce'
        )

        df_clean = df_melted.dropna(subset=['date']).copy()
        df_clean['date'] = df_clean['date'].dt.strftime('%Y-%m-%d')
        df_clean = df_clean[(df_clean['date'] >= '2015-01-01') & (df_clean['date'] <= '2025-12-31')]
        df_clean['station'] = station

        all_dis_data.append(df_clean[['date', 'station', 'discharge_cumecs']])

    if not all_dis_data:
        return pd.DataFrame(columns=['date', 'station', 'discharge_cumecs'])

    df_concat = pd.concat(all_dis_data, ignore_index=True)
    return df_concat.groupby(['date', 'station'], as_index=False).agg({
        'discharge_cumecs': 'mean'
    })

def main():
    rf_path = os.path.join(RAW_DIR, "2015 - 2025 RF.xlsx")
    wl_path = os.path.join(RAW_DIR, "2015 - 2025 WL.xlsx")
    dis_path = os.path.join(RAW_DIR, "2015 - 2025 Discharge.xlsx")

    df_rf = parse_rainfall(rf_path)
    df_wl = parse_water_level(wl_path)
    df_dis = parse_discharge(dis_path)

    print(f"Parsed Rainfall Rows: {len(df_rf)}")
    print(f"Parsed Water Level Rows: {len(df_wl)}")
    print(f"Parsed Discharge Rows: {len(df_dis)}")

    print("Building unified daily master table...")
    df_master = pd.merge(df_rf, df_wl, on=['date', 'station'], how='outer')
    df_master = pd.merge(df_master, df_dis, on=['date', 'station'], how='outer')

    # Station coordinates backfill
    df_master['coord_x'] = df_master['coord_x'].fillna(df_master['station'].map(lambda s: STATION_COORDS.get(s, (np.nan, np.nan))[0]))
    df_master['coord_y'] = df_master['coord_y'].fillna(df_master['station'].map(lambda s: STATION_COORDS.get(s, (np.nan, np.nan))[1]))

    df_master.sort_values(by=['date', 'station'], inplace=True)
    master_path = os.path.join(PROCESSED_DIR, "kelani_master_daily_analytics_2015_2025.csv")
    df_master.to_csv(master_path, index=False)

    print("\nDaily Aggregation Complete!")
    print(f"Master file saved to: {master_path}")

if __name__ == '__main__':
    main()
```

```python
def parse_discharge(file_path):
    print("Processing Discharge Data...")
    xls = pd.ExcelFile(file_path)
    all_dis_data = []

    for sheet in xls.sheet_names:
        df_raw = pd.read_excel(xls, sheet_name=sheet, header=None)
        station = clean_station_name(sheet)

        # Dynamic header detection: locate row with numbers 1..31 or D1..D31
        header_idx = None
        for idx, row in df_raw.iloc[:10].iterrows():
            row_str = row.astype(str).str.strip().str.upper().tolist()
            if any(re.match(r'^(D?\d{1,2})$', s) for s in row_str if s not in ['NAN', 'NONE', '']):
                header_idx = idx
                break

        if header_idx is None:
            continue

        df_data = df_raw.iloc[header_idx + 1:].copy()
        raw_cols = [str(c).strip() for c in df_raw.iloc[header_idx]]
        df_data.columns = raw_cols

        # Identify day columns vs year/month columns
        day_cols = [c for c in df_data.columns if re.match(r'^(D?\d{1,2})$', c, re.IGNORECASE) and 1 <= int(re.sub(r'\D', '', c)) <= 31]
        ym_cols = [c for c in df_data.columns if c not in day_cols and c.lower() not in ['nan', 'none', '']]
        
        if not day_cols or not ym_cols:
            continue

        ym_col = ym_cols[0]
        df_melted = pd.melt(df_data, id_vars=[ym_col], value_vars=day_cols, var_name='day', value_name='discharge_cumecs')
        
        s_clean = df_melted[ym_col].astype(str).str.replace(r'\D', '', regex=True)
        
        # Handle 6-digit (YYYYMM) vs 4-digit (YYMM) year-month strings
        df_melted['year_num'] = np.where(s_clean.str.len() >= 6, s_clean.str[:4], s_clean.str[:2]).astype(str)
        df_melted['year_num'] = df_melted['year_num'].apply(fix_year)
        
        df_melted['month_num'] = np.where(s_clean.str.len() >= 6, s_clean.str[4:6], s_clean.str[2:4])
        df_melted['month_num'] = pd.to_numeric(df_melted['month_num'], errors='coerce')
        
        df_melted['day'] = df_melted['day'].str.extract(r'(\d+)').astype(int)
        df_melted['discharge_cumecs'] = pd.to_numeric(df_melted['discharge_cumecs'], errors='coerce')

        df_melted['date'] = pd.to_datetime(
            dict(year=df_melted['year_num'], month=df_melted['month_num'], day=df_melted['day']),
            errors='coerce'
        )

        df_clean = df_melted.dropna(subset=['date']).copy()
        df_clean['date'] = df_clean['date'].dt.strftime('%Y-%m-%d')
        df_clean = df_clean[(df_clean['date'] >= '2015-01-01') & (df_clean['date'] <= '2025-12-31')]
        df_clean['station'] = station

        all_dis_data.append(df_clean[['date', 'station', 'discharge_cumecs']])

    if not all_dis_data:
        return pd.DataFrame(columns=['date', 'station', 'discharge_cumecs'])

    df_concat = pd.concat(all_dis_data, ignore_index=True)
    return df_concat.groupby(['date', 'station'], as_index=False).agg({'discharge_cumecs': 'mean'})
```

```python
def extract_rf_year_month(ym_series):
    s = ym_series.astype(str).str.strip()
    
    # Check if year is explicitly 4 digits (e.g. "201501" or "2015-01")
    digits_only = s.str.replace(r'\D', '', regex=True)
    
    years = []
    months = []
    
    for val in digits_only:
        if len(val) >= 6: # e.g. 201501
            years.append(int(val[:4]))
            months.append(int(val[4:6]))
        elif len(val) == 4: # e.g. 1501
            yr = int(val[:2])
            years.append(2000 + yr if yr <= 30 else 1900 + yr)
            months.append(int(val[2:4]))
        else:
            years.append(np.nan)
            months.append(np.nan)
            
    return pd.Series(years), pd.Series(months)
```

```python
import os
import re
import pandas as pd
import numpy as np

PROJECT_ROOT = "/Users/ohaseelanka/disaster_analytics_hub"
RAW_DIR = os.path.join(PROJECT_ROOT, "data/raw/flood/kelani")
PROCESSED_DIR = os.path.join(PROJECT_ROOT, "data/processed/flood/kelani")

os.makedirs(PROCESSED_DIR, exist_ok=True)

MONTH_MAP = {
    'JAN': 1, 'FEB': 2, 'MAR': 3, 'APR': 4, 'MAY': 5, 'JUN': 6,
    'JUL': 7, 'AUG': 8, 'SEP': 9, 'OCT': 10, 'NOV': 11, 'DEC': 12
}

STATION_MAP = {
    'Kithulgala': 'Kitulgala',
    'Kn11-Holombuwa': 'Holombuwa',
    'Kn11_Holombuwa': 'Holombuwa',
    'N\'Street': 'N_Street',
    'Nstreet': 'N_Street'
}

STATION_COORDS = {
    'Kitulgala': (184500.0, 171800.0),
    'Norwood': (198200.0, 166400.0),
    'Deraniyagala': (168300.0, 168100.0),
    'Glencourse': (132100.0, 170900.0),
    'Hanwella': (116300.0, 172100.0),
    'Holombuwa': (142000.0, 190200.0),
    'N_Street': (100800.0, 174800.0)
}

def clean_station_name(name):
    name = re.sub(r'^[Kk][Nn]\d+[-_\s]*', '', str(name)).strip()
    name = name.replace("N'Street", "N_Street").title()
    return STATION_MAP.get(name, name).strip()

def parse_year_month_string(series):
    """Accurately parses YYYYMM, YYMM, YYYY-MM, or YYYY text formats."""
    s_clean = series.astype(str).str.strip().str.upper()
    years, months = [], []

    for val in s_clean:
        digits = re.sub(r'\D', '', val)
        if len(digits) >= 6:  # e.g., '201501'
            y = int(digits[:4])
            m = int(digits[4:6])
        elif len(digits) == 4:  # e.g., '1501'
            yr_2 = int(digits[:2])
            y = 2000 + yr_2 if yr_2 <= 30 else 1900 + yr_2
            m = int(digits[2:4])
        else:
            y, m = np.nan, np.nan

        years.append(y if 2010 <= y <= 2030 else np.nan)
        months.append(m if 1 <= m <= 12 else np.nan)

    return pd.Series(years, index=series.index), pd.Series(months, index=series.index)

def find_header_row(df_raw):
    """Locates row index containing daily columns (1..31 or D1..D31)."""
    for idx, row in df_raw.iloc[:10].iterrows():
        row_vals = [str(v).strip().upper() for v in row.dropna()]
        day_matches = [v for v in row_vals if re.match(r'^(D?\d{1,2})$', v) and 1 <= int(re.sub(r'\D', '', v)) <= 31]
        if len(day_matches) >= 15:  # Valid row contains most days of a month
            return idx
    return None

def parse_rainfall(file_path):
    print("Processing Rainfall Data...")
    xls = pd.ExcelFile(file_path)
    all_rf_data = []

    for sheet in xls.sheet_names:
        df_raw = pd.read_excel(xls, sheet_name=sheet, header=None)
        station = clean_station_name(sheet)

        header_idx = find_header_row(df_raw)
        if header_idx is None:
            continue

        # Extract coordinates if present
        coord_row = df_raw.iloc[:header_idx].fillna('').astype(str).values.flatten()
        coord_text = " ".join(coord_row)
        x_match = re.search(r'(\d+)\s*X', coord_text)
        y_match = re.search(r'(\d+)\s*Y', coord_text)
        coord_x = float(x_match.group(1)) if x_match else np.nan
        coord_y = float(y_match.group(1)) if y_match else np.nan

        df_data = df_raw.iloc[header_idx + 1:].copy()
        df_data.columns = [str(c).strip() for c in df_raw.iloc[header_idx]]

        day_cols = [c for c in df_data.columns if re.match(r'^(D?\d{1,2})$', c, re.IGNORECASE) and 1 <= int(re.sub(r'\D', '', c)) <= 31]
        ym_cols = [c for c in df_data.columns if c not in day_cols and str(c).lower() not in ['nan', 'none', '']]

        if not day_cols or not ym_cols:
            continue

        ym_col = ym_cols[0]
        df_melted = pd.melt(df_data, id_vars=[ym_col], value_vars=day_cols, var_name='day', value_name='rainfall_mm')
        
        years, months = parse_year_month_string(df_melted[ym_col])
        df_melted['year_num'] = years
        df_melted['month_num'] = months
        df_melted['day'] = df_melted['day'].str.extract(r'(\d+)').astype(int)
        df_melted['rainfall_mm'] = pd.to_numeric(df_melted['rainfall_mm'], errors='coerce')

        df_melted['date'] = pd.to_datetime(
            dict(year=df_melted['year_num'], month=df_melted['month_num'], day=df_melted['day']),
            errors='coerce'
        )

        df_clean = df_melted.dropna(subset=['date']).copy()
        df_clean['date'] = df_clean['date'].dt.strftime('%Y-%m-%d')
        df_clean = df_clean[(df_clean['date'] >= '2015-01-01') & (df_clean['date'] <= '2025-12-31')]
        df_clean['station'] = station
        df_clean['coord_x'] = coord_x
        df_clean['coord_y'] = coord_y

        all_rf_data.append(df_clean[['date', 'station', 'rainfall_mm', 'coord_x', 'coord_y']])

    if not all_rf_data:
        return pd.DataFrame(columns=['date', 'station', 'rainfall_mm', 'coord_x', 'coord_y'])

    df_concat = pd.concat(all_rf_data, ignore_index=True)
    return df_concat.groupby(['date', 'station'], as_index=False).agg({
        'rainfall_mm': 'mean',
        'coord_x': 'first',
        'coord_y': 'first'
    })

def parse_water_level(file_path):
    print("Processing Water Level Data...")
    xls = pd.ExcelFile(file_path)
    all_wl_data = []

    for sheet in xls.sheet_names:
        df_raw = pd.read_excel(xls, sheet_name=sheet, header=None)
        station = clean_station_name(sheet)

        msl_val = np.nan
        msl_str = str(df_raw.iloc[2, 1]) if df_raw.shape[1] > 1 else ""
        msl_match = re.search(r'([-+]?\d*\.\d+|\d+)', msl_str)
        if msl_match:
            msl_val = float(msl_match.group(1))

        df_data = df_raw.iloc[5:].copy().iloc[:, :4]
        df_data.columns = ['year', 'month', 'day', 'water_level']

        # Flexible year handling
        years_s = df_data['year'].astype(str).str.replace(r'\D', '', regex=True)
        df_data['year_num'] = pd.to_numeric(years_s, errors='coerce').apply(
            lambda y: y if y >= 2000 else (2000 + y if y <= 30 else 1900 + y) if pd.notna(y) else np.nan
        )
        df_data['month_num'] = pd.to_numeric(df_data['month'], errors='coerce')
        df_data['day_num'] = pd.to_numeric(df_data['day'], errors='coerce')
        df_data['water_level'] = pd.to_numeric(df_data['water_level'], errors='coerce')

        if 'N_Street' in station:
            df_data['water_level_mMSL'] = df_data['water_level'] * 0.3048
            unit = 'mMSL (converted from ft)'
        else:
            df_data['water_level_mMSL'] = df_data['water_level']
            unit = 'mMSL'

        df_data['date'] = pd.to_datetime(
            dict(year=df_data['year_num'], month=df_data['month_num'], day=df_data['day_num']),
            errors='coerce'
        )

        df_clean = df_data.dropna(subset=['date']).copy()
        df_clean['date'] = df_clean['date'].dt.strftime('%Y-%m-%d')
        df_clean = df_clean[(df_clean['date'] >= '2015-01-01') & (df_clean['date'] <= '2025-12-31')]
        df_clean['station'] = station
        df_clean['datum_msl_m'] = msl_val
        df_clean['unit'] = unit

        all_wl_data.append(df_clean[['date', 'station', 'water_level_mMSL', 'datum_msl_m', 'unit']])

    if not all_wl_data:
        return pd.DataFrame(columns=['date', 'station', 'water_level_mMSL', 'datum_msl_m', 'unit'])

    df_concat = pd.concat(all_wl_data, ignore_index=True)
    return df_concat.groupby(['date', 'station'], as_index=False).agg({
        'water_level_mMSL': 'mean',
        'datum_msl_m': 'first',
        'unit': 'first'
    })

def parse_discharge(file_path):
    print("Processing Discharge Data...")
    xls = pd.ExcelFile(file_path)
    all_dis_data = []

    for sheet in xls.sheet_names:
        df_raw = pd.read_excel(xls, sheet_name=sheet, header=None)
        station = clean_station_name(sheet)

        header_idx = find_header_row(df_raw)
        if header_idx is None:
            continue

        df_data = df_raw.iloc[header_idx + 1:].copy()
        df_data.columns = [str(c).strip() for c in df_raw.iloc[header_idx]]

        day_cols = [c for c in df_data.columns if re.match(r'^(D?\d{1,2})$', c, re.IGNORECASE) and 1 <= int(re.sub(r'\D', '', c)) <= 31]
        ym_cols = [c for c in df_data.columns if c not in day_cols and str(c).lower() not in ['nan', 'none', '']]

        if not day_cols or not ym_cols:
            continue

        ym_col = ym_cols[0]
        df_melted = pd.melt(df_data, id_vars=[ym_col], value_vars=day_cols, var_name='day', value_name='discharge_cumecs')

        years, months = parse_year_month_string(df_melted[ym_col])
        df_melted['year_num'] = years
        df_melted['month_num'] = months
        df_melted['day'] = df_melted['day'].str.extract(r'(\d+)').astype(int)
        df_melted['discharge_cumecs'] = pd.to_numeric(df_melted['discharge_cumecs'], errors='coerce')

        df_melted['date'] = pd.to_datetime(
            dict(year=df_melted['year_num'], month=df_melted['month_num'], day=df_melted['day']),
            errors='coerce'
        )

        df_clean = df_melted.dropna(subset=['date']).copy()
        df_clean['date'] = df_clean['date'].dt.strftime('%Y-%m-%d')
        df_clean = df_clean[(df_clean['date'] >= '2015-01-01') & (df_clean['date'] <= '2025-12-31')]
        df_clean['station'] = station

        all_dis_data.append(df_clean[['date', 'station', 'discharge_cumecs']])

    if not all_dis_data:
        return pd.DataFrame(columns=['date', 'station', 'discharge_cumecs'])

    df_concat = pd.concat(all_dis_data, ignore_index=True)
    return df_concat.groupby(['date', 'station'], as_index=False).agg({'discharge_cumecs': 'mean'})

def main():
    rf_path = os.path.join(RAW_DIR, "2015 - 2025 RF.xlsx")
    wl_path = os.path.join(RAW_DIR, "2015 - 2025 WL.xlsx")
    dis_path = os.path.join(RAW_DIR, "2015 - 2025 Discharge.xlsx")

    df_rf = parse_rainfall(rf_path)
    df_wl = parse_water_level(wl_path)
    df_dis = parse_discharge(dis_path)

    print("\n=== PARSED ROW COUNTS ===")
    print(f"Rainfall:    {len(df_rf)} rows")
    print(f"Water Level: {len(df_wl)} rows")
    print(f"Discharge:   {len(df_dis)} rows\n")

    print("Building unified daily master table...")
    df_master = pd.merge(df_rf, df_wl, on=['date', 'station'], how='outer')
    df_master = pd.merge(df_master, df_dis, on=['date', 'station'], how='outer')

    # Fill coordinates
    df_master['coord_x'] = df_master['coord_x'].fillna(df_master['station'].map(lambda s: STATION_COORDS.get(s, (np.nan, np.nan))[0]))
    df_master['coord_y'] = df_master['coord_y'].fillna(df_master['station'].map(lambda s: STATION_COORDS.get(s, (np.nan, np.nan))[1]))

    df_master.sort_values(by=['date', 'station'], inplace=True)
    master_path = os.path.join(PROCESSED_DIR, "kelani_master_daily_analytics_2015_2025.csv")
    df_master.to_csv(master_path, index=False)

    print(f"Master file saved to: {master_path}")
    return df_rf, df_wl, df_dis, df_master

if __name__ == '__main__':
    df_rf, df_wl, df_dis, df_master = main()
```

```python
import pandas as pd
import numpy as np
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent

rf_file = (
    PROJECT_ROOT
    / "data"
    / "raw"
    / "flood"
    / "kelani"
    / "2015 - 2025 RF.xlsx"
)

output_file = (
    PROJECT_ROOT
    / "data"
    / "interim"
    / "flood"
    / "rainfall_daily_long.parquet"
)

output_file.parent.mkdir(parents=True, exist_ok=True)

print("Input:", rf_file)
print("Output:", output_file)
```

```python
# Station information
station_info = {
    "KN02-NORWOOD": {
        "station_id": "KN02",
        "station_name": "Norwood"
    },
    "KN07-KITULGALA": {
        "station_id": "KN07",
        "station_name": "Kitulgala"
    },
    " KN11-HOLOMBUWA": {
        "station_id": "KN11",
        "station_name": "Holombuwa"
    },
    "KN16-DERANIYAGALA": {
        "station_id": "KN16",
        "station_name": "Deraniyagala"
    },
    "KN18-GLENCOURSE": {
        "station_id": "KN18",
        "station_name": "Glencourse"
    },
    "KN21-HANWELLA": {
        "station_id": "KN21",
        "station_name": "Hanwella"
    }
}

all_station_data = []

for sheet_name in pd.ExcelFile(rf_file).sheet_names:

    print(f"Processing: {sheet_name}")

    # Read the actual table beginning at row 3
    df = pd.read_excel(
        rf_file,
        sheet_name=sheet_name,
        header=3
    )

    # Remove accidental whitespace from column names
    df.columns = df.columns.astype(str).str.strip()

    info = station_info[sheet_name]

    # Keep only the daily rainfall columns
    day_columns = [f"D{i}" for i in range(1, 32)]

    # Convert N.A. and other non-numeric values to NaN
    for col in day_columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

    # Convert month names to month numbers
    month_numbers = {
        "JAN": 1,
        "FEB": 2,
        "MAR": 3,
        "APR": 4,
        "MAY": 5,
        "JUN": 6,
        "JUL": 7,
        "AUG": 8,
        "SEP": 9,
        "OCT": 10,
        "NOV": 11,
        "DEC": 12
    }

    df["month_number"] = (
        df["MONTH"]
        .astype(str)
        .str.strip()
        .str.upper()
        .map(month_numbers)
    )

    # Convert wide format into daily long format
    long_df = df.melt(
        id_vars=["YEAR", "MONTH", "month_number"],
        value_vars=day_columns,
        var_name="day_column",
        value_name="rainfall_mm"
    )

    # Extract day number from D1, D2, ... D31
    long_df["day"] = (
        long_df["day_column"]
        .str.replace("D", "", regex=False)
        .astype(int)
    )

    # Construct actual date
    long_df["date"] = pd.to_datetime(
        dict(
            year=long_df["YEAR"],
            month=long_df["month_number"],
            day=long_df["day"]
        ),
        errors="coerce"
    )

    # Add station information
    long_df["station_id"] = info["station_id"]
    long_df["station_name"] = info["station_name"]

    # Keep only required columns
    long_df = long_df[
        [
            "station_id",
            "station_name",
            "date",
            "rainfall_mm"
        ]
    ]

    # Remove invalid calendar dates
    long_df = long_df.dropna(subset=["date"])

    all_station_data.append(long_df)

# Combine all stations
rainfall_daily = pd.concat(
    all_station_data,
    ignore_index=True
)

# Sort
rainfall_daily = rainfall_daily.sort_values(
    ["station_id", "date"]
).reset_index(drop=True)

print("\nFinal shape:", rainfall_daily.shape)
print("\nStations:")
print(rainfall_daily["station_id"].unique())

print("\nDate range:")
print(rainfall_daily["date"].min())
print(rainfall_daily["date"].max())

print("\nPreview:")
print(rainfall_daily.head(20))
```

```python
import pandas as pd
import numpy as np
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent

wl_file = (
    PROJECT_ROOT
    / "data"
    / "raw"
    / "flood"
    / "kelani"
    / "2015 - 2025 WL.xlsx"
)

output_file = (
    PROJECT_ROOT
    / "data"
    / "interim"
    / "flood"
    / "water_level_daily.parquet"
)

output_file.parent.mkdir(parents=True, exist_ok=True)

station_info = {
    "Norwood": {
        "station_id": "KN02",
        "station_name": "Norwood",
        "unit": "mMSL"
    },
    "Kithulgala": {
        "station_id": "KN07",
        "station_name": "Kitulgala",
        "unit": "mMSL"
    },
    "Deraniyagala": {
        "station_id": "KN16",
        "station_name": "Deraniyagala",
        "unit": "mMSL"
    },
    "Holombuwa": {
        "station_id": "KN11",
        "station_name": "Holombuwa",
        "unit": "mMSL"
    },
    "Glencourse": {
        "station_id": "KN18",
        "station_name": "Glencourse",
        "unit": "mMSL"
    },
    "Hanwella": {
        "station_id": "KN21",
        "station_name": "Hanwella",
        "unit": "mMSL"
    },
    "N'Street": {
        "station_id": "NSTR",
        "station_name": "N'Street",
        "unit": "ftMSL"
    }
}

all_wl = []

for sheet_name, info in station_info.items():

    df = pd.read_excel(
        wl_file,
        sheet_name=sheet_name,
        header=4
    )

    # Clean column names
    df.columns = df.columns.astype(str).str.strip()

    print("\nProcessing:", sheet_name)
    print("Columns:", df.columns.tolist())

    # Rename columns according to the confirmed structure
    if sheet_name == "N'Street":
        df = df.rename(columns={
            "Year": "year",
            "Month": "month",
            "Date": "day",
            "Avg. WL (ft)": "water_level"
        })
    else:
        df = df.rename(columns={
            "Year": "year",
            "Month": "month",
            "Date": "day",
            "Avg. WL": "water_level"
        })

    # Convert numeric columns
    for col in ["year", "month", "day", "water_level"]:
        df[col] = pd.to_numeric(df[col], errors="coerce")

    # Construct date
    df["date"] = pd.to_datetime(
        dict(
            year=df["year"],
            month=df["month"],
            day=df["day"]
        ),
        errors="coerce"
    )

    # Add station information
    df["station_id"] = info["station_id"]
    df["station_name"] = info["station_name"]
    df["unit"] = info["unit"]

    # Keep only required fields
    df = df[
        [
            "station_id",
            "station_name",
            "date",
            "water_level",
            "unit"
        ]
    ]

    # Remove rows where date itself is invalid
    df = df.dropna(subset=["date"])

    all_wl.append(df)

water_level_daily = pd.concat(
    all_wl,
    ignore_index=True
)

water_level_daily = water_level_daily.sort_values(
    ["station_id", "date"]
).reset_index(drop=True)

print("\nFinal shape:", water_level_daily.shape)

print("\nStations:")
print(water_level_daily["station_id"].unique())

print("\nDate ranges:")
print(
    water_level_daily
    .groupby(["station_id", "station_name"])["date"]
    .agg(["min", "max", "count"])
)

print("\nMissing water-level values:")
print(
    water_level_daily["water_level"].isna().sum()
)

print("\nUnits:")
print(
    water_level_daily.groupby("station_id")["unit"].unique()
)
```

```python
import pandas as pd
import numpy as np
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent

discharge_file = (
    PROJECT_ROOT
    / "data"
    / "raw"
    / "flood"
    / "kelani"
    / "2015 - 2025 Discharge.xlsx"
)

output_file = (
    PROJECT_ROOT
    / "data"
    / "interim"
    / "flood"
    / "discharge_daily.parquet"
)

output_file.parent.mkdir(parents=True, exist_ok=True)

station_info = {
    "Norwood": {
        "station_id": "KN02",
        "station_name": "Norwood"
    },
    "Kithulgala": {
        "station_id": "KN07",
        "station_name": "Kitulgala"
    },
    "Deraniyagala": {
        "station_id": "KN16",
        "station_name": "Deraniyagala"
    },
    "Holombuwa": {
        "station_id": "KN11",
        "station_name": "Holombuwa"
    },
    "Glencourse": {
        "station_id": "KN18",
        "station_name": "Glencourse"
    },
    "Hanwella": {
        "station_id": "KN21",
        "station_name": "Hanwella"
    }
}

all_discharge = []

for sheet_name, info in station_info.items():

    df = pd.read_excel(
        discharge_file,
        sheet_name=sheet_name,
        header=3
    )

    df.columns = df.columns.astype(str).str.strip()

    # First column is YEARMONTH
    df = df.rename(columns={
        df.columns[0]: "year_month"
    })

    # Identify daily columns
    day_columns = [
        col for col in df.columns
        if str(col).upper().startswith("D")
        and str(col)[1:].isdigit()
    ]

    # Convert discharge values to numeric
    for col in day_columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

    # Convert wide format to long format
    long_df = df.melt(
        id_vars=["year_month"],
        value_vars=day_columns,
        var_name="day",
        value_name="discharge_m3s"
    )

    # Extract year and month
    long_df["year_month"] = long_df["year_month"].astype(str).str.strip()

    long_df["year"] = pd.to_numeric(
        long_df["year_month"].str.extract(r"^(\d{4})")[0],
        errors="coerce"
    )

    long_df["month_name"] = (
        long_df["year_month"]
        .str.extract(r"^\d{4}\s+([A-Za-z]+)")[0]
    )

    month_map = {
        "Jan": 1,
        "Feb": 2,
        "Mar": 3,
        "Apr": 4,
        "May": 5,
        "Jun": 6,
        "Jul": 7,
        "Aug": 8,
        "Sep": 9,
        "Oct": 10,
        "Nov": 11,
        "Dec": 12
    }

    long_df["month"] = long_df["month_name"].map(month_map)

    long_df["day_number"] = (
        long_df["day"]
        .str.extract(r"(\d+)")[0]
        .astype(float)
    )

    # Construct date
    long_df["date"] = pd.to_datetime(
        dict(
            year=long_df["year"],
            month=long_df["month"],
            day=long_df["day_number"]
        ),
        errors="coerce"
    )

    # Add station information
    long_df["station_id"] = info["station_id"]
    long_df["station_name"] = info["station_name"]

    # Keep required columns
    long_df = long_df[
        [
            "station_id",
            "station_name",
            "date",
            "discharge_m3s"
        ]
    ]

    # Remove invalid calendar dates
    long_df = long_df.dropna(subset=["date"])

    all_discharge.append(long_df)


discharge_daily = pd.concat(
    all_discharge,
    ignore_index=True
)

discharge_daily = discharge_daily.sort_values(
    ["station_id", "date"]
).reset_index(drop=True)


print("Final shape:", discharge_daily.shape)

print("\nStations:")
print(discharge_daily["station_id"].unique())

print("\nDate ranges:")
print(
    discharge_daily
    .groupby(["station_id", "station_name"])["date"]
    .agg(["min", "max", "count"])
)

print("\nMissing discharge values:")
print(
    discharge_daily["discharge_m3s"].isna().sum()
)

print("\nDischarge statistics:")
print(
    discharge_daily["discharge_m3s"].describe()
)
```

```python
discharge_daily.to_parquet(output_file, index=False)

print("Saved successfully:", output_file)
print("File exists:", output_file.exists())
```

```python
from pathlib import Path
import pandas as pd

# ---------------------------------------------------------
# 1. Work on a copy
# ---------------------------------------------------------
population_clean = population.copy()

# ---------------------------------------------------------
# 2. Convert numeric columns
# ---------------------------------------------------------
numeric_columns = [
    "province_code",
    "district_code",
    "ds_code",
    "gn_code",
    "total_population",
    "male",
    "female",
    "age_total",
    "age_0_14",
    "age_15_59",
    "age_60_64",
    "age_65_plus"
]

for col in numeric_columns:
    population_clean[col] = pd.to_numeric(
        population_clean[col],
        errors="coerce"
    )

# ---------------------------------------------------------
# 3. Create elderly population (60+)
# ---------------------------------------------------------
population_clean["elderly_60_plus"] = (
    population_clean["age_60_64"]
    + population_clean["age_65_plus"]
)

# ---------------------------------------------------------
# 4. Create population ratios
# ---------------------------------------------------------
population_clean["children_ratio"] = (
    population_clean["age_0_14"]
    / population_clean["total_population"]
)

population_clean["elderly_ratio"] = (
    population_clean["elderly_60_plus"]
    / population_clean["total_population"]
)

population_clean["female_ratio"] = (
    population_clean["female"]
    / population_clean["total_population"]
)

# ---------------------------------------------------------
# 5. Final validation
# ---------------------------------------------------------
print("Shape:", population_clean.shape)

print("\nMissing values:")
print(population_clean.isna().sum())

print("\nDuplicate GN codes:")
print(population_clean["gn_code"].duplicated().sum())

# Age consistency
age_check = (
    population_clean["age_0_14"]
    + population_clean["age_15_59"]
    + population_clean["age_60_64"]
    + population_clean["age_65_plus"]
)

print("\nAge consistency errors:")
print((age_check != population_clean["total_population"]).sum())

# Sex consistency
sex_check = (
    population_clean["male"]
    + population_clean["female"]
)

print("\nSex consistency errors:")
print((sex_check != population_clean["total_population"]).sum())

# ---------------------------------------------------------
# 6. Display final structure
# ---------------------------------------------------------
print("\nFinal columns:")
print(population_clean.columns.tolist())

display(population_clean.head())

# ---------------------------------------------------------
# 7. Save cleaned Census population dataset
# ---------------------------------------------------------
output_dir = PROJECT_ROOT / "data" / "interim" / "census"
output_dir.mkdir(parents=True, exist_ok=True)

output_file = output_dir / "population_gn_clean.parquet"

population_clean.to_parquet(
    output_file,
    index=False
)

print("\nSaved successfully:", output_file)
print("File exists:", output_file.exists())
```

```python
# ---------------------------------------------------------
# VALIDATE HOUSEHOLD CATEGORY TOTALS
# ---------------------------------------------------------

# Convert household count columns to numeric
for col in households.columns[9:]:
    households[col] = pd.to_numeric(
        households[col],
        errors="coerce"
    )

# Define category groups
water_columns = [
    "water_protected_well",
    "water_semi_protected_well",
    "water_unprotected_well",
    "water_tube_well",
    "water_spring_fountain",
    "water_national_water_supply",
    "water_local_authority",
    "water_community_organization",
    "water_private_supply",
    "water_tank_river_stream",
    "water_rain_water",
    "water_bottled",
    "water_filter_ro_plant",
    "water_bowser",
    "water_other"
]

cooking_columns = [
    "cooking_firewood",
    "cooking_kerosene",
    "cooking_gas",
    "cooking_electricity",
    "cooking_sawdust_paddy_husk",
    "cooking_biogas",
    "cooking_other",
    "cooking_not_relevant"
]

lighting_columns = [
    "lighting_national_grid_hydro",
    "lighting_kerosene_lamp",
    "lighting_solar_grid",
    "lighting_solar_standalone",
    "lighting_biogas",
    "lighting_generator",
    "lighting_other"
]

toilet_columns = [
    "toilet_within_unit_exclusive",
    "toilet_within_unit_shared",
    "toilet_within_premises_exclusive",
    "toilet_within_premises_shared",
    "toilet_no_toilet_shared_elsewhere",
    "toilet_common_public",
    "toilet_no_toilet"
]

# Calculate category totals
households["water_sum"] = households[water_columns].sum(axis=1)
households["cooking_sum"] = households[cooking_columns].sum(axis=1)
households["lighting_sum"] = households[lighting_columns].sum(axis=1)
households["toilet_sum"] = households[toilet_columns].sum(axis=1)

# Compare against total households
households["water_difference"] = (
    households["water_sum"] - households["number_of_households"]
)

households["cooking_difference"] = (
    households["cooking_sum"] - households["number_of_households"]
)

households["lighting_difference"] = (
    households["lighting_sum"] - households["number_of_households"]
)

households["toilet_difference"] = (
    households["toilet_sum"] - households["number_of_households"]
)

# Results
print("Water mismatches:",
      (households["water_difference"] != 0).sum())

print("Cooking mismatches:",
      (households["cooking_difference"] != 0).sum())

print("Lighting mismatches:",
      (households["lighting_difference"] != 0).sum())

print("Toilet mismatches:",
      (households["toilet_difference"] != 0).sum())

print("\nDifference statistics:")

print("\nWater:")
print(households["water_difference"].describe())

print("\nCooking:")
print(households["cooking_difference"].describe())

print("\nLighting:")
print(households["lighting_difference"].describe())

print("\nToilet:")
print(households["toilet_difference"].describe())
```

```python
# ---------------------------------------------------------
# FINALIZE GN-LEVEL HOUSEHOLD DATA
# ---------------------------------------------------------

# Remove temporary validation columns
validation_columns = [
    "water_sum",
    "cooking_sum",
    "lighting_sum",
    "toilet_sum",
    "water_difference",
    "cooking_difference",
    "lighting_difference",
    "toilet_difference"
]

households_clean = households.drop(
    columns=validation_columns
).copy()

# ---------------------------------------------------------
# Convert administrative identifiers
# ---------------------------------------------------------

numeric_columns = [
    "province_code",
    "district_code",
    "ds_code",
    "gn_code",
    "number_of_households"
]

for col in numeric_columns:
    households_clean[col] = pd.to_numeric(
        households_clean[col],
        errors="coerce"
    )

# GN number remains text
households_clean["gn_number"] = (
    households_clean["gn_number"].astype("string")
)

# Administrative names
text_columns = [
    "province_name",
    "district_name",
    "ds_name",
    "gn_name"
]

for col in text_columns:
    households_clean[col] = (
        households_clean[col].astype("string")
    )

# ---------------------------------------------------------
# Household condition variables
# ---------------------------------------------------------

condition_columns = [
    col for col in households_clean.columns
    if col not in [
        "province_code",
        "province_name",
        "district_code",
        "district_name",
        "ds_code",
        "ds_name",
        "gn_code",
        "gn_name",
        "gn_number"
    ]
]

for col in condition_columns:
    households_clean[col] = pd.to_numeric(
        households_clean[col],
        errors="coerce"
    )

# ---------------------------------------------------------
# Create unique GN key
# ---------------------------------------------------------

households_clean["gn_key"] = (
    households_clean["district_code"].astype("string").str.strip()
    + "_"
    + households_clean["ds_code"].astype("string").str.strip()
    + "_"
    + households_clean["gn_code"].astype("string").str.strip()
)

# ---------------------------------------------------------
# Final validation
# ---------------------------------------------------------

print("Shape:", households_clean.shape)

print("\nUnique GN keys:",
      households_clean["gn_key"].nunique())

print("Duplicate GN keys:",
      households_clean["gn_key"].duplicated().sum())

print("\nMissing values:")
print(households_clean.isna().sum().sum())

print("\nData types:")
print(households_clean.dtypes)

# ---------------------------------------------------------
# Save
# ---------------------------------------------------------

output_dir = PROJECT_ROOT / "data" / "interim" / "census"
output_dir.mkdir(parents=True, exist_ok=True)

output_file = output_dir / "households_gn_clean.parquet"

households_clean.to_parquet(
    output_file,
    index=False
)

print("\nSaved successfully:", output_file)
print("File exists:", output_file.exists())
```

```python
# ---------------------------------------------------------
# MERGE THE 3 CLEAN CENSUS DATASETS
# ---------------------------------------------------------

population_clean = pd.read_parquet(
    PROJECT_ROOT / "data/interim/census/population_gn_clean.parquet"
)

housing_clean = pd.read_parquet(
    PROJECT_ROOT / "data/interim/census/housing_gn_clean.parquet"
)

households_clean = pd.read_parquet(
    PROJECT_ROOT / "data/interim/census/households_gn_clean.parquet"
)

print("Population:", population_clean.shape)
print("Housing:", housing_clean.shape)
print("Households:", households_clean.shape)
```

```python
# ---------------------------------------------------------
# VERIFY GN KEYS BEFORE MERGING
# ---------------------------------------------------------

population_keys = set(population_clean["gn_key"])
housing_keys = set(housing_clean["gn_key"])
household_keys = set(households_clean["gn_key"])

print("Population unique keys:", len(population_keys))
print("Housing unique keys:", len(housing_keys))
print("Household unique keys:", len(household_keys))

print("\nPopulation ↔ Housing")
print("Common:", len(population_keys & housing_keys))
print("Population only:", len(population_keys - housing_keys))
print("Housing only:", len(housing_keys - population_keys))

print("\nPopulation ↔ Households")
print("Common:", len(population_keys & household_keys))
print("Population only:", len(population_keys - household_keys))
print("Households only:", len(household_keys - population_keys))

print("\nHousing ↔ Households")
print("Common:", len(housing_keys & household_keys))
print("Housing only:", len(housing_keys - household_keys))
print("Households only:", len(household_keys - housing_keys))
```

```python
# ---------------------------------------------------------
# CREATE / STANDARDIZE GN KEY FOR ALL 3 CENSUS DATASETS
# ---------------------------------------------------------

def create_gn_key(df):
    df = df.copy()

    df["gn_key"] = (
        df["district_code"].astype("string").str.strip()
        + "_"
        + df["ds_code"].astype("string").str.strip()
        + "_"
        + df["gn_code"].astype("string").str.strip()
    )

    return df


# Create GN keys
population_clean = create_gn_key(population_clean)
housing_clean = create_gn_key(housing_clean)
households_clean = create_gn_key(households_clean)


# ---------------------------------------------------------
# Check
# ---------------------------------------------------------

print("Population:", population_clean.shape)
print("Housing:", housing_clean.shape)
print("Households:", households_clean.shape)

print("\nGN key columns:")
print("Population:", "gn_key" in population_clean.columns)
print("Housing:", "gn_key" in housing_clean.columns)
print("Households:", "gn_key" in households_clean.columns)

print("\nUnique GN keys:")
print("Population:", population_clean["gn_key"].nunique())
print("Housing:", housing_clean["gn_key"].nunique())
print("Households:", households_clean["gn_key"].nunique())

print("\nDuplicate GN keys:")
print("Population:", population_clean["gn_key"].duplicated().sum())
print("Housing:", housing_clean["gn_key"].duplicated().sum())
print("Households:", households_clean["gn_key"].duplicated().sum())
```

```python
# ---------------------------------------------------------
# FINAL GN KEY MATCHING CHECK
# ---------------------------------------------------------

population_keys = set(population_clean["gn_key"])
housing_keys = set(housing_clean["gn_key"])
household_keys = set(households_clean["gn_key"])

print("Population ↔ Housing")
print("Common:", len(population_keys & housing_keys))
print("Population only:", len(population_keys - housing_keys))
print("Housing only:", len(housing_keys - population_keys))

print("\nPopulation ↔ Households")
print("Common:", len(population_keys & household_keys))
print("Population only:", len(population_keys - household_keys))
print("Households only:", len(household_keys - population_keys))

print("\nHousing ↔ Households")
print("Common:", len(housing_keys & household_keys))
print("Housing only:", len(housing_keys - household_keys))
print("Households only:", len(household_keys - housing_keys))

# Overall check
all_match = (
    population_keys == housing_keys == household_keys
)

print("\nAll three datasets match:", all_match)
```

```python
# ---------------------------------------------------------
# MERGE POPULATION + HOUSING + HOUSEHOLDS
# ---------------------------------------------------------

# Start with population data
census_gn = population_clean.copy()

# Remove duplicate administrative columns from housing
housing_merge = housing_clean.drop(
    columns=[
        "province_code",
        "province_name",
        "district_code",
        "district_name",
        "ds_code",
        "ds_name",
        "gn_code",
        "gn_name",
        "gn_number"
    ]
)

# Merge housing
census_gn = census_gn.merge(
    housing_merge,
    on="gn_key",
    how="inner",
    validate="one_to_one"
)

# Remove duplicate administrative columns from households
households_merge = households_clean.drop(
    columns=[
        "province_code",
        "province_name",
        "district_code",
        "district_name",
        "ds_code",
        "ds_name",
        "gn_code",
        "gn_name",
        "gn_number"
    ]
)

# Merge households
census_gn = census_gn.merge(
    households_merge,
    on="gn_key",
    how="inner",
    validate="one_to_one"
)

# ---------------------------------------------------------
# CHECK RESULT
# ---------------------------------------------------------

print("Final Census GN shape:", census_gn.shape)

print("\nUnique GN keys:",
      census_gn["gn_key"].nunique())

print("Duplicate GN keys:",
      census_gn["gn_key"].duplicated().sum())

print("\nMissing values:")
print("Total missing values:", census_gn.isna().sum().sum())

print("\nFirst 5 rows:")
display(census_gn.head())
```

```python
# ---------------------------------------------------------
# INSPECT ALL MISSING VALUES IN THE MERGED CENSUS DATA
# ---------------------------------------------------------

missing_summary = (
    census_gn.isna()
    .sum()
    .sort_values(ascending=False)
)

print("Columns containing missing values:")
display(
    missing_summary[missing_summary > 0]
)

# Show the actual affected GN records
missing_columns = missing_summary[
    missing_summary > 0
].index.tolist()

print("\nAffected records:")
display(
    census_gn[
        census_gn[missing_columns].isna().any(axis=1)
    ][[
        "gn_key",
        "district_name",
        "ds_name",
        "gn_name",
        "total_population",
        "children_ratio",
        "elderly_ratio",
        "female_ratio"
    ]]
)
```

```python
# ---------------------------------------------------------
# FINALIZE CENSUS GN MASTER DATASET
# ---------------------------------------------------------

# Copy merged dataset
census_gn_final = census_gn.copy()

# Remove population validation columns
validation_columns = [
    "age_sum",
    "age_difference",
    "sex_sum",
    "sex_difference"
]

census_gn_final = census_gn_final.drop(
    columns=validation_columns,
    errors="ignore"
)

# ---------------------------------------------------------
# Final structure check
# ---------------------------------------------------------

print("Final Census shape:", census_gn_final.shape)

print("\nUnique GN keys:",
      census_gn_final["gn_key"].nunique())

print("Duplicate GN keys:",
      census_gn_final["gn_key"].duplicated().sum())

print("\nMissing values:")
missing_final = (
    census_gn_final.isna()
    .sum()
    .sort_values(ascending=False)
)

display(
    missing_final[missing_final > 0]
)

# ---------------------------------------------------------
# Save final Census GN dataset
# ---------------------------------------------------------

output_file = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "census_gn_master.parquet"
)

output_file.parent.mkdir(
    parents=True,
    exist_ok=True
)

census_gn_final.to_parquet(
    output_file,
    index=False
)

print("\nSaved successfully:", output_file)
print("File exists:", output_file.exists())
```

```python
# ---------------------------------------------------------
# KALU GANGA - LOAD DATA AND CHECK DATE COVERAGE
# ---------------------------------------------------------

# Load the two measurement sheets
kalu_wl_raw = pd.read_excel(
    kalu_file,
    sheet_name="WL Summery MSL",
    header=None
)

kalu_q_raw = pd.read_excel(
    kalu_file,
    sheet_name="Q",
    header=None
)

# ---------------------------------------------------------
# Function to check date coverage
# ---------------------------------------------------------

def check_date_coverage(df, name):

    dates = pd.to_datetime(
        df.iloc[3:, 0].astype(str) + "-" +
        df.iloc[3:, 1].astype(str) + "-" +
        df.iloc[3:, 2].astype(str),
        errors="coerce"
    )

    dates = dates.dropna()

    print("=" * 60)
    print(name)
    print("=" * 60)

    print("Rows in workbook       :", len(df))
    print("Valid dates            :", len(dates))
    print("Unique dates           :", dates.nunique())
    print("Start date             :", dates.min())
    print("End date               :", dates.max())
    print("Duplicate dates        :", dates.duplicated().sum())


# ---------------------------------------------------------
# Run checks
# ---------------------------------------------------------

check_date_coverage(
    kalu_wl_raw,
    "KALU GANGA - WATER LEVEL"
)

print()

check_date_coverage(
    kalu_q_raw,
    "KALU GANGA - DISCHARGE"
)
```

```python
# ---------------------------------------------------------
# KALU GANGA - FIND DUPLICATE AND MISSING DATES
# ---------------------------------------------------------

def get_dates(df):
    return pd.to_datetime(
        df.iloc[3:, 0].astype(str) + "-" +
        df.iloc[3:, 1].astype(str) + "-" +
        df.iloc[3:, 2].astype(str),
        errors="coerce"
    )


# =========================
# WATER LEVEL
# =========================

wl_dates = get_dates(kalu_wl_raw)

print("=" * 60)
print("WATER LEVEL")
print("=" * 60)

duplicate_wl = wl_dates[wl_dates.duplicated(keep=False)].dropna()

print("Duplicate date:")
print(duplicate_wl.unique())

expected_wl = pd.date_range(
    wl_dates.min(),
    wl_dates.max(),
    freq="D"
)

missing_wl = expected_wl.difference(
    pd.DatetimeIndex(wl_dates.dropna().unique())
)

print("\nMissing date:")
print(missing_wl)


# =========================
# DISCHARGE
# =========================

q_dates = get_dates(kalu_q_raw)

print("\n" + "=" * 60)
print("DISCHARGE")
print("=" * 60)

duplicate_q = q_dates[q_dates.duplicated(keep=False)].dropna()

print("Duplicate date:")
print(duplicate_q.unique())

expected_q = pd.date_range(
    q_dates.min(),
    q_dates.max(),
    freq="D"
)

missing_q = expected_q.difference(
    pd.DatetimeIndex(q_dates.dropna().unique())
)

print("\nMissing date:")
print(missing_q)
```

```python
# ---------------------------------------------------------
# KALU GANGA - INSPECT DUPLICATE 2020-02-28
# ---------------------------------------------------------

duplicate_date = pd.Timestamp("2020-02-28")


# =========================
# WATER LEVEL
# =========================

wl_data = kalu_wl_raw.iloc[3:].copy()

wl_data["date"] = pd.to_datetime(
    wl_data.iloc[:, 0].astype(str) + "-" +
    wl_data.iloc[:, 1].astype(str) + "-" +
    wl_data.iloc[:, 2].astype(str),
    errors="coerce"
)

print("=" * 70)
print("WATER LEVEL - DUPLICATE 2020-02-28")
print("=" * 70)

display(
    wl_data[
        wl_data["date"] == duplicate_date
    ]
)


# =========================
# DISCHARGE
# =========================

q_data = kalu_q_raw.iloc[3:].copy()

q_data["date"] = pd.to_datetime(
    q_data.iloc[:, 0].astype(str) + "-" +
    q_data.iloc[:, 1].astype(str) + "-" +
    q_data.iloc[:, 2].astype(str),
    errors="coerce"
)

print("=" * 70)
print("DISCHARGE - DUPLICATE 2020-02-28")
print("=" * 70)

display(
    q_data[
        q_data["date"] == duplicate_date
    ]
)
```

```python
# ---------------------------------------------------------
# KALU GANGA - CLEAN WATER LEVEL AND DISCHARGE
# ---------------------------------------------------------

# =========================================================
# WATER LEVEL
# =========================================================

wl = kalu_wl_raw.iloc[3:].copy()

wl["date"] = pd.to_datetime(
    wl.iloc[:, 0].astype(str) + "-" +
    wl.iloc[:, 1].astype(str) + "-" +
    wl.iloc[:, 2].astype(str),
    errors="coerce"
)

# Keep only valid dates
wl = wl[wl["date"].notna()].copy()

# Rename hourly columns
wl_columns = (
    ["year", "month", "day"]
    + [f"hour_{i:02d}" for i in range(1, 25)]
    + ["avg", "max", "min", "date"]
)

wl.columns = wl_columns

# Convert measurements to numeric
measurement_columns = [
    f"hour_{i:02d}" for i in range(1, 25)
] + ["avg", "max", "min"]

for col in measurement_columns:
    wl[col] = pd.to_numeric(wl[col], errors="coerce")

# Remove duplicate date - keep first authentic record
wl = wl.drop_duplicates(
    subset="date",
    keep="first"
).copy()

# Sort chronologically
wl = wl.sort_values("date").reset_index(drop=True)


# =========================================================
# DISCHARGE
# =========================================================

q = kalu_q_raw.iloc[3:].copy()

q["date"] = pd.to_datetime(
    q.iloc[:, 0].astype(str) + "-" +
    q.iloc[:, 1].astype(str) + "-" +
    q.iloc[:, 2].astype(str),
    errors="coerce"
)

# Keep only valid dates
q = q[q["date"].notna()].copy()

# Rename columns
q_columns = (
    ["year", "month", "day"]
    + [f"hour_{i:02d}" for i in range(1, 25)]
    + ["avg", "max", "min", "date"]
)

q.columns = q_columns

# Convert measurements to numeric
for col in measurement_columns:
    q[col] = pd.to_numeric(q[col], errors="coerce")

# Remove duplicate date - keep first authentic record
q = q.drop_duplicates(
    subset="date",
    keep="first"
).copy()

# Sort chronologically
q = q.sort_values("date").reset_index(drop=True)


# =========================================================
# VALIDATION
# =========================================================

print("=" * 70)
print("KALU GANGA CLEANING VALIDATION")
print("=" * 70)

print("\nWATER LEVEL")
print("Rows:", len(wl))
print("Unique dates:", wl["date"].nunique())
print("Start:", wl["date"].min())
print("End:", wl["date"].max())
print("Duplicate dates:", wl["date"].duplicated().sum())

print("\nDISCHARGE")
print("Rows:", len(q))
print("Unique dates:", q["date"].nunique())
print("Start:", q["date"].min())
print("End:", q["date"].max())
print("Duplicate dates:", q["date"].duplicated().sum())

print("\nExpected observations:")
print("2015-01-01 to 2025-12-31 = 4018 calendar days")

print("\nMissing dates remain:")
expected_dates = pd.date_range(
    "2015-01-01",
    "2025-12-31",
    freq="D"
)

wl_missing = expected_dates.difference(
    pd.DatetimeIndex(wl["date"])
)

q_missing = expected_dates.difference(
    pd.DatetimeIndex(q["date"])
)

print("Water Level:", list(wl_missing))
print("Discharge:", list(q_missing))
```

```python
# ---------------------------------------------------------
# KALU GANGA - CREATE DAILY HYDROLOGICAL FEATURES
# ---------------------------------------------------------

# =========================================================
# SELECT WATER LEVEL FEATURES
# =========================================================

wl_features = wl[
    [
        "date",
        "avg",
        "max",
        "min"
    ]
].copy()

wl_features = wl_features.rename(
    columns={
        "avg": "wl_avg",
        "max": "wl_max",
        "min": "wl_min"
    }
)


# =========================================================
# SELECT DISCHARGE FEATURES
# =========================================================

q_features = q[
    [
        "date",
        "avg",
        "max",
        "min"
    ]
].copy()

q_features = q_features.rename(
    columns={
        "avg": "discharge_avg",
        "max": "discharge_max",
        "min": "discharge_min"
    }
)


# =========================================================
# MERGE WATER LEVEL + DISCHARGE
# =========================================================

kalu_daily = wl_features.merge(
    q_features,
    on="date",
    how="outer",
    validate="one_to_one"
)

kalu_daily = kalu_daily.sort_values(
    "date"
).reset_index(drop=True)


# =========================================================
# OFFICIAL RATHNAPURA FLOOD THRESHOLDS
# =========================================================

ALERT_LEVEL = 17.14
MINOR_FLOOD_LEVEL = 19.44
MAJOR_FLOOD_LEVEL = 21.44


# =========================================================
# THRESHOLD FEATURES
# =========================================================

kalu_daily["alert_exceeded"] = (
    kalu_daily["wl_max"] >= ALERT_LEVEL
)

kalu_daily["minor_flood_exceeded"] = (
    kalu_daily["wl_max"] >= MINOR_FLOOD_LEVEL
)

kalu_daily["major_flood_exceeded"] = (
    kalu_daily["wl_max"] >= MAJOR_FLOOD_LEVEL
)


# =========================================================
# VALIDATION
# =========================================================

print("=" * 70)
print("KALU GANGA DAILY HYDROLOGICAL DATASET")
print("=" * 70)

print("\nShape:", kalu_daily.shape)

print("\nDate range:")
print("Start:", kalu_daily["date"].min())
print("End  :", kalu_daily["date"].max())

print("\nColumns:")
print(kalu_daily.columns.tolist())

print("\nMissing values:")
print(kalu_daily.isna().sum())

print("\nThreshold exceedances:")
print(
    "Alert days        :",
    kalu_daily["alert_exceeded"].sum()
)

print(
    "Minor flood days  :",
    kalu_daily["minor_flood_exceeded"].sum()
)

print(
    "Major flood days  :",
    kalu_daily["major_flood_exceeded"].sum()
)

print("\nFirst 5 rows:")
display(kalu_daily.head())

print("\nHighest water levels:")
display(
    kalu_daily.nlargest(
        10,
        "wl_max"
    )[
        [
            "date",
            "wl_avg",
            "wl_max",
            "wl_min",
            "discharge_avg",
            "discharge_max",
            "alert_exceeded",
            "minor_flood_exceeded",
            "major_flood_exceeded"
        ]
    ]
)
```

```python
# ---------------------------------------------------------
# KALU GANGA - CORRECT THRESHOLD FLAGS FOR MISSING DATA
# ---------------------------------------------------------

measurement_missing = (
    kalu_daily[
        [
            "wl_max",
            "discharge_max"
        ]
    ].isna().any(axis=1)
)

threshold_columns = [
    "alert_exceeded",
    "minor_flood_exceeded",
    "major_flood_exceeded"
]

# Set threshold indicators to missing where
# the required hydrological observation is missing
for col in threshold_columns:
    kalu_daily.loc[
        measurement_missing,
        col
    ] = pd.NA


# ---------------------------------------------------------
# VALIDATION
# ---------------------------------------------------------

print("=" * 70)
print("KALU GANGA - THRESHOLD FLAG VALIDATION")
print("=" * 70)

print("\nRows with missing hydrological measurements:")
display(
    kalu_daily[
        measurement_missing
    ]
)

print("\nThreshold missing counts:")
print(
    kalu_daily[
        threshold_columns
    ].isna().sum()
)

print("\nThreshold counts excluding missing observations:")

for col in threshold_columns:
    print(
        col,
        ":",
        kalu_daily[col].eq(True).sum()
    )
```

```python
# Make copies so the original cleaned datasets remain unchanged

rf = rainfall.copy()
wl = wl_2015_2025.copy()
q = discharge.copy()

# Make sure dates are datetime
rf["date"] = pd.to_datetime(rf["date"])
wl["date"] = pd.to_datetime(wl["date"])
q["date"] = pd.to_datetime(q["date"])

# Sort correctly
rf = rf.sort_values(["station_id", "date"])
wl = wl.sort_values(["station_id", "date"])
q = q.sort_values(["station_id", "date"])

# ---------------------------------------------------------
# 1. Rainfall rolling features
# ---------------------------------------------------------

rf["rainfall_3d"] = (
    rf.groupby("station_id")["rainfall_mm"]
      .transform(lambda x: x.rolling(3, min_periods=3).sum())
)

rf["rainfall_7d"] = (
    rf.groupby("station_id")["rainfall_mm"]
      .transform(lambda x: x.rolling(7, min_periods=7).sum())
)

# ---------------------------------------------------------
# 2. Station-normalized water-level anomaly
# ---------------------------------------------------------

wl["wl_mean"] = (
    wl.groupby("station_id")["water_level"]
      .transform("mean")
)

wl["wl_std"] = (
    wl.groupby("station_id")["water_level"]
      .transform("std")
)

wl["wl_zscore"] = (
    (wl["water_level"] - wl["wl_mean"]) /
    wl["wl_std"]
)

# ---------------------------------------------------------
# 3. Station-normalized discharge anomaly
# ---------------------------------------------------------

q["q_mean"] = (
    q.groupby("station_id")["discharge_m3s"]
      .transform("mean")
)

q["q_std"] = (
    q.groupby("station_id")["discharge_m3s"]
      .transform("std")
)

q["q_zscore"] = (
    (q["discharge_m3s"] - q["q_mean"]) /
    q["q_std"]
)

print("Rainfall features:", rf.shape)
print("Water-level features:", wl.shape)
print("Discharge features:", q.shape)
```

```python
print("=" * 70)
print("RAINFALL FEATURES")
print("=" * 70)

display(
    rf[
        [
            "station_id",
            "date",
            "rainfall_mm",
            "rainfall_3d",
            "rainfall_7d"
        ]
    ].head(15)
)

print("=" * 70)
print("WATER-LEVEL FEATURES")
print("=" * 70)

display(
    wl[
        [
            "station_id",
            "date",
            "water_level",
            "wl_mean",
            "wl_std",
            "wl_zscore"
        ]
    ].head(15)
)

print("=" * 70)
print("DISCHARGE FEATURES")
print("=" * 70)

display(
    q[
        [
            "station_id",
            "date",
            "discharge_m3s",
            "q_mean",
            "q_std",
            "q_zscore"
        ]
    ].head(15)
)
```

```python
# ============================================================
# INSPECT STATION-NORMALIZED HYDROLOGICAL FEATURES
# ============================================================

print("=" * 70)
print("RAINFALL EXTREMES")
print("=" * 70)

display(
    rf[
        ["station_id", "station_name", "date",
         "rainfall_mm", "rainfall_3d", "rainfall_7d"]
    ]
    .sort_values("rainfall_3d", ascending=False)
    .head(20)
)

print("=" * 70)
print("WATER-LEVEL ANOMALIES")
print("=" * 70)

display(
    wl[
        ["station_id", "station_name", "date",
         "water_level", "wl_zscore"]
    ]
    .sort_values("wl_zscore", ascending=False)
    .head(20)
)

print("=" * 70)
print("DISCHARGE ANOMALIES")
print("=" * 70)

display(
    q[
        ["station_id", "station_name", "date",
         "discharge_m3s", "q_zscore"]
    ]
    .sort_values("q_zscore", ascending=False)
    .head(20)
)
```

```python
# ============================================================
# COMBINE RAINFALL + WATER LEVEL + DISCHARGE
# SIX KELANI STATIONS ONLY
# ============================================================

# Keep only stations that have rainfall + water level + discharge
six_stations = sorted(
    set(rf["station_id"])
    & set(wl["station_id"])
    & set(q["station_id"])
)

print("Stations with all three datasets:")
print(six_stations)

# Select useful columns
rf_features = rf[
    [
        "station_id",
        "station_name",
        "date",
        "rainfall_mm",
        "rainfall_3d",
        "rainfall_7d"
    ]
].copy()

wl_features = wl[
    [
        "station_id",
        "date",
        "water_level",
        "wl_zscore"
    ]
].copy()

q_features = q[
    [
        "station_id",
        "date",
        "discharge_m3s",
        "q_zscore"
    ]
].copy()

# Merge rainfall + water level
hydro = rf_features.merge(
    wl_features,
    on=["station_id", "date"],
    how="left",
    validate="one_to_one"
)

# Merge discharge
hydro = hydro.merge(
    q_features,
    on=["station_id", "date"],
    how="left",
    validate="one_to_one"
)

# Keep only the six complete stations
hydro = hydro[
    hydro["station_id"].isin(six_stations)
].copy()

# Sort
hydro = hydro.sort_values(
    ["station_id", "date"]
).reset_index(drop=True)

print()
print("Combined hydrological dataset shape:", hydro.shape)
print()
print("Columns:")
print(hydro.columns.tolist())
print()
print("Missing values:")
print(hydro.isna().sum())
```

```python
# ============================================================
# STATION-SPECIFIC EXTREME THRESHOLD INSPECTION
# ============================================================

print("=" * 70)
print("RAINFALL 3-DAY / 7-DAY EXTREMES")
print("=" * 70)

rainfall_percentiles = (
    hydro.groupby("station_id")
    .agg(
        rainfall_3d_p90=("rainfall_3d", lambda x: x.quantile(0.90)),
        rainfall_3d_p95=("rainfall_3d", lambda x: x.quantile(0.95)),
        rainfall_3d_p99=("rainfall_3d", lambda x: x.quantile(0.99)),
        rainfall_7d_p90=("rainfall_7d", lambda x: x.quantile(0.90)),
        rainfall_7d_p95=("rainfall_7d", lambda x: x.quantile(0.95)),
        rainfall_7d_p99=("rainfall_7d", lambda x: x.quantile(0.99))
    )
)

display(rainfall_percentiles)

print("=" * 70)
print("WATER-LEVEL Z-SCORE DISTRIBUTION")
print("=" * 70)

wl_percentiles = (
    hydro.groupby("station_id")["wl_zscore"]
    .quantile([0.90, 0.95, 0.99])
    .unstack()
)

wl_percentiles.columns = [
    "wl_z_p90",
    "wl_z_p95",
    "wl_z_p99"
]

display(wl_percentiles)

print("=" * 70)
print("DISCHARGE Z-SCORE DISTRIBUTION")
print("=" * 70)

q_percentiles = (
    hydro.groupby("station_id")["q_zscore"]
    .quantile([0.90, 0.95, 0.99])
    .unstack()
)

q_percentiles.columns = [
    "q_z_p90",
    "q_z_p95",
    "q_z_p99"
]

display(q_percentiles)
```

```python
# ============================================================
# CREATE STATION-SPECIFIC P95 THRESHOLDS
# ============================================================

thresholds = (
    hydro.groupby("station_id")
    .agg(
        rainfall_3d_p95=("rainfall_3d", lambda x: x.quantile(0.95)),
        rainfall_7d_p95=("rainfall_7d", lambda x: x.quantile(0.95)),
        wl_z_p95=("wl_zscore", lambda x: x.quantile(0.95)),
        q_z_p95=("q_zscore", lambda x: x.quantile(0.95))
    )
    .reset_index()
)

print("Station-specific P95 thresholds:")
display(thresholds)
```

```python
# ============================================================
# ATTACH STATION-SPECIFIC THRESHOLDS
# ============================================================

hydro_thresholded = hydro.merge(
    thresholds,
    on="station_id",
    how="left",
    validate="many_to_one"
)

print("Shape:", hydro_thresholded.shape)

display(
    hydro_thresholded[
        [
            "station_id",
            "date",
            "rainfall_3d",
            "rainfall_3d_p95",
            "rainfall_7d",
            "rainfall_7d_p95",
            "wl_zscore",
            "wl_z_p95",
            "q_zscore",
            "q_z_p95"
        ]
    ].head(15)
)
```

```python
# ============================================================
# IDENTIFY CANDIDATE FLOOD DAYS
# ============================================================

# High accumulated rainfall
hydro_thresholded["rainfall_extreme"] = (
    (hydro_thresholded["rainfall_3d"] >= hydro_thresholded["rainfall_3d_p95"])
    |
    (hydro_thresholded["rainfall_7d"] >= hydro_thresholded["rainfall_7d_p95"])
)

# Strong river response
hydro_thresholded["river_response_extreme"] = (
    (hydro_thresholded["wl_zscore"] >= hydro_thresholded["wl_z_p95"])
    |
    (hydro_thresholded["q_zscore"] >= hydro_thresholded["q_z_p95"])
)

# Candidate flood condition
hydro_thresholded["candidate_flood_day"] = (
    hydro_thresholded["rainfall_extreme"]
    &
    hydro_thresholded["river_response_extreme"]
)

print("Candidate flood days:")
print(
    hydro_thresholded["candidate_flood_day"]
    .value_counts(dropna=False)
)

print()
print("Candidate days by station:")
display(
    hydro_thresholded
    .groupby("station_id")["candidate_flood_day"]
    .agg(["count", "sum"])
)
```

```python
# ============================================================
# INSPECT CANDIDATE FLOOD DAYS
# ============================================================

candidate_days = (
    hydro_thresholded[
        hydro_thresholded["candidate_flood_day"] == True
    ]
    .sort_values(["date", "station_id"])
)

display(
    candidate_days[
        [
            "station_id",
            "station_name",
            "date",
            "rainfall_mm",
            "rainfall_3d",
            "rainfall_7d",
            "wl_zscore",
            "q_zscore"
        ]
    ].head(100)
)
```

```python
import pandas as pd

candidate = (
    hydro_thresholded[
        hydro_thresholded["candidate_flood_day"] == True
    ]
    .copy()
)

candidate["date"] = pd.to_datetime(candidate["date"])

candidate = candidate.sort_values(
    ["station_id", "date"]
).reset_index(drop=True)

# Calculate the gap between consecutive candidate days
candidate["gap_days"] = (
    candidate.groupby("station_id")["date"]
    .diff()
    .dt.days
)

# Start a new episode if the gap is greater than 2 days.
# This allows a 1-day non-candidate gap inside an episode.
candidate["new_event"] = (
    candidate["gap_days"].isna()
    | (candidate["gap_days"] > 2)
)

# Create station-specific event IDs
candidate["event_number"] = (
    candidate.groupby("station_id")["new_event"]
    .cumsum()
)

candidate["event_id"] = (
    candidate["station_id"]
    + "_E"
    + candidate["event_number"].astype(str).str.zfill(3)
)

candidate.head(20)
```

```python
print("Candidate station-days:", len(candidate))

print(
    "Candidate flood episodes:",
    candidate["event_id"].nunique()
)

print("\nEvents by station:")
print(
    candidate.groupby("station_id")["event_id"]
    .nunique()
)
```

```python
# Create a compact event table
station_events = (
    candidate
    .groupby(["station_id", "event_id"])
    .agg(
        start_date=("date", "min"),
        end_date=("date", "max"),
        candidate_days=("date", "count"),
        max_rainfall_3d=("rainfall_3d", "max"),
        max_rainfall_7d=("rainfall_7d", "max"),
        max_wl_zscore=("wl_zscore", "max"),
        max_q_zscore=("q_zscore", "max")
    )
    .reset_index()
)

station_events["duration_days"] = (
    station_events["end_date"]
    - station_events["start_date"]
).dt.days + 1

station_events.sort_values("start_date").head(30)
```

```python
import pandas as pd
from pathlib import Path

project_root = Path("/Users/ohaseelanka/disaster_analytics_hub")

rf_path = project_root / "data/interim/flood/rainfall_daily_long.parquet"
wl_path = project_root / "data/interim/flood/water_level_daily.parquet"
q_path = project_root / "data/interim/flood/discharge_daily.parquet"

rf = pd.read_parquet(rf_path)
wl = pd.read_parquet(wl_path)
q = pd.read_parquet(q_path)

print("Rainfall:", rf.shape)
print("Water level:", wl.shape)
print("Discharge:", q.shape)

print("\nRainfall columns:")
print(rf.columns.tolist())

print("\nWater level columns:")
print(wl.columns.tolist())

print("\nDischarge columns:")
print(q.columns.tolist())
```

```python
# Make copies so the original cleaned datasets remain unchanged

rf = rf.copy()
wl = wl.copy()
q = q.copy()

# Make sure dates are datetime
rf["date"] = pd.to_datetime(rf["date"])
wl["date"] = pd.to_datetime(wl["date"])
q["date"] = pd.to_datetime(q["date"])

# Sort chronologically within each station
rf = rf.sort_values(["station_id", "date"]).reset_index(drop=True)
wl = wl.sort_values(["station_id", "date"]).reset_index(drop=True)
q = q.sort_values(["station_id", "date"]).reset_index(drop=True)

# --------------------------------------------------
# RAINFALL: 3-day and 7-day accumulated rainfall
# --------------------------------------------------

rf["rainfall_3d"] = (
    rf.groupby("station_id")["rainfall_mm"]
      .transform(lambda x: x.rolling(3, min_periods=3).sum())
)

rf["rainfall_7d"] = (
    rf.groupby("station_id")["rainfall_mm"]
      .transform(lambda x: x.rolling(7, min_periods=7).sum())
)

# --------------------------------------------------
# WATER LEVEL: station-relative z-score
# --------------------------------------------------

wl["wl_mean"] = (
    wl.groupby("station_id")["water_level"]
      .transform("mean")
)

wl["wl_std"] = (
    wl.groupby("station_id")["water_level"]
      .transform("std")
)

wl["wl_zscore"] = (
    (wl["water_level"] - wl["wl_mean"]) / wl["wl_std"]
)

# --------------------------------------------------
# DISCHARGE: station-relative z-score
# --------------------------------------------------

q["q_mean"] = (
    q.groupby("station_id")["discharge_m3s"]
      .transform("mean")
)

q["q_std"] = (
    q.groupby("station_id")["discharge_m3s"]
      .transform("std")
)

q["q_zscore"] = (
    (q["discharge_m3s"] - q["q_mean"]) / q["q_std"]
)

# --------------------------------------------------
# Display results
# --------------------------------------------------

print("Rainfall features:", rf.shape)
print("Water-level features:", wl.shape)
print("Discharge features:", q.shape)

print("\nRainfall columns:")
print(rf.columns.tolist())

print("\nWater-level columns:")
print(wl.columns.tolist())

print("\nDischarge columns:")
print(q.columns.tolist())
```

```python
# Calculate station-specific P95 thresholds

rainfall_thresholds = (
    rf.groupby("station_id")[["rainfall_3d", "rainfall_7d"]]
      .quantile(0.95)
      .reset_index()
      .rename(columns={
          "rainfall_3d": "rainfall_3d_p95",
          "rainfall_7d": "rainfall_7d_p95"
      })
)

wl_thresholds = (
    wl[wl["station_id"].isin(
        ["KN02", "KN07", "KN11", "KN16", "KN18", "KN21"]
    )]
    .groupby("station_id")["wl_zscore"]
    .quantile(0.95)
    .reset_index()
    .rename(columns={"wl_zscore": "wl_zscore_p95"})
)

q_thresholds = (
    q.groupby("station_id")["q_zscore"]
     .quantile(0.95)
     .reset_index()
     .rename(columns={"q_zscore": "q_zscore_p95"})
)

# Display the thresholds

print("RAINFALL P95 THRESHOLDS")
display(rainfall_thresholds)

print("\nWATER LEVEL P95 THRESHOLDS")
display(wl_thresholds)

print("\nDISCHARGE P95 THRESHOLDS")
display(q_thresholds)
```

```python
# Use only the six Kelani stations that have
# rainfall + water level + discharge data

six_stations = sorted(
    set(rf["station_id"])
    & set(wl["station_id"])
    & set(q["station_id"])
)

print("Six common stations:")
print(six_stations)

# Keep only the six common stations
rf_six = rf[rf["station_id"].isin(six_stations)].copy()
wl_six = wl[wl["station_id"].isin(six_stations)].copy()
q_six = q[q["station_id"].isin(six_stations)].copy()

# Select only the required columns
rf_six = rf_six[
    [
        "station_id",
        "station_name",
        "date",
        "rainfall_mm",
        "rainfall_3d",
        "rainfall_7d"
    ]
]

wl_six = wl_six[
    [
        "station_id",
        "date",
        "water_level",
        "wl_zscore"
    ]
]

q_six = q_six[
    [
        "station_id",
        "date",
        "discharge_m3s",
        "q_zscore"
    ]
]

# Merge rainfall + water level
hydro = rf_six.merge(
    wl_six,
    on=["station_id", "date"],
    how="inner",
    validate="one_to_one"
)

# Add discharge
hydro = hydro.merge(
    q_six,
    on=["station_id", "date"],
    how="inner",
    validate="one_to_one"
)

hydro = hydro.sort_values(
    ["station_id", "date"]
).reset_index(drop=True)

print("\nCombined hydrology shape:", hydro.shape)

print("\nColumns:")
print(hydro.columns.tolist())

print("\nMissing values:")
display(hydro.isna().sum())

print("\nFirst 10 rows:")
display(hydro.head(10))
```

```python
# --------------------------------------------------
# STEP 7: DETECT CANDIDATE FLOOD DAYS
# --------------------------------------------------

# Merge station-specific thresholds into the hydrology dataset

hydro = hydro.merge(
    rainfall_thresholds,
    on="station_id",
    how="left",
    validate="many_to_one"
)

hydro = hydro.merge(
    wl_thresholds,
    on="station_id",
    how="left",
    validate="many_to_one"
)

hydro = hydro.merge(
    q_thresholds,
    on="station_id",
    how="left",
    validate="many_to_one"
)

# --------------------------------------------------
# Rainfall extreme
# --------------------------------------------------

hydro["rainfall_extreme"] = (
    (hydro["rainfall_3d"] >= hydro["rainfall_3d_p95"]) |
    (hydro["rainfall_7d"] >= hydro["rainfall_7d_p95"])
)

# --------------------------------------------------
# River-response extreme
# --------------------------------------------------

hydro["river_response_extreme"] = (
    (hydro["wl_zscore"] >= hydro["wl_zscore_p95"]) |
    (hydro["q_zscore"] >= hydro["q_zscore_p95"])
)

# --------------------------------------------------
# Candidate flood day
# --------------------------------------------------

hydro["candidate_flood_day"] = (
    hydro["rainfall_extreme"] &
    hydro["river_response_extreme"]
)

# --------------------------------------------------
# Basic results
# --------------------------------------------------

print("Hydrology shape:", hydro.shape)

print("\nCandidate flood-day counts:")
print(
    hydro["candidate_flood_day"]
    .value_counts(dropna=False)
)

print("\nNumber of candidate station-days:")
print(
    hydro["candidate_flood_day"].sum()
)

print("\nCandidate station-days by station:")
display(
    hydro[hydro["candidate_flood_day"]]
    .groupby("station_id")
    .size()
    .reset_index(name="candidate_days")
)

print("\nSample candidate days:")
display(
    hydro[hydro["candidate_flood_day"]]
    [
        [
            "station_id",
            "station_name",
            "date",
            "rainfall_3d",
            "rainfall_7d",
            "wl_zscore",
            "q_zscore"
        ]
    ]
    .head(20)
)
```

```python
# --------------------------------------------------
# STEP 9: LOAD AUTHENTIC DESINVENTAR FLOOD RECORDS
# --------------------------------------------------

desinventar_path = (
    project_root
    / "data/raw/disaster_records/"
    / "DesInventar_Ratnapura_Kegalle_Floods_pages1-2.xlsx"
)

print("File exists:", desinventar_path.exists())

xls = pd.ExcelFile(desinventar_path)

print("\nSheets:")
print(xls.sheet_names)

# Load the main flood sheet
desinventar_flood = pd.read_excel(
    desinventar_path,
    sheet_name="Ratnapura-Kegalle Floods"
)

print("\nShape:", desinventar_flood.shape)

print("\nColumns:")
print(desinventar_flood.columns.tolist())

display(desinventar_flood.head(10))
```

```python
# --------------------------------------------------
# STEP 9A: STANDARDIZE DESINVENTAR FLOOD RECORDS
# --------------------------------------------------

des = desinventar_flood.copy()

# Convert date to proper datetime
des["date"] = pd.to_datetime(
    des["Date"],
    format="%Y/%m/%d",
    errors="coerce"
)

# Clean district and DS division names
des["district"] = (
    des["District"]
    .astype("string")
    .str.strip()
)

des["ds_division"] = (
    des["Division"]
    .astype("string")
    .str.strip()
)

# Keep only records overlapping our hydrological observation period
des_overlap = des[
    (des["date"] >= "2015-01-01") &
    (des["date"] <= "2019-12-31")
].copy()

# --------------------------------------------------
# Basic validation
# --------------------------------------------------

print("Total DesInventar records:", len(des))
print("2015–2019 overlapping records:", len(des_overlap))

print("\nRecords by year:")
display(
    des_overlap
    .assign(year=des_overlap["date"].dt.year)
    .groupby("year")
    .size()
    .reset_index(name="record_count")
)

print("\nRecords by district:")
display(
    des_overlap
    .groupby("district")
    .size()
    .reset_index(name="record_count")
)

print("\nDS divisions represented:")
display(
    des_overlap
    .groupby(["district", "ds_division"])
    .size()
    .reset_index(name="record_count")
    .sort_values(["district", "ds_division"])
)

print("\nDate range:")
print(des_overlap["date"].min(), "to", des_overlap["date"].max())
```

```python
# --------------------------------------------------
# STEP 9B: TEMPORAL MATCHING
# DesInventar records ↔ Hydrological candidate events
# --------------------------------------------------

# Make sure dates are datetime
candidate_events["start_date"] = pd.to_datetime(
    candidate_events["start_date"]
)

candidate_events["end_date"] = pd.to_datetime(
    candidate_events["end_date"]
)

des_overlap["date"] = pd.to_datetime(
    des_overlap["date"]
)

# --------------------------------------------------
# Find candidate station events occurring within
# ±2 days of each DesInventar flood record
# --------------------------------------------------

matches = []

for _, record in des_overlap.iterrows():

    disaster_date = record["date"]

    # Candidate event overlaps the ±2 day window
    matched = candidate_events[
        (candidate_events["start_date"] <= disaster_date + pd.Timedelta(days=2)) &
        (candidate_events["end_date"] >= disaster_date - pd.Timedelta(days=2))
    ].copy()

    if len(matched) > 0:

        for _, event in matched.iterrows():

            # Calculate distance from the DesInventar date
            if disaster_date < event["start_date"]:
                date_distance = (
                    event["start_date"] - disaster_date
                ).days

            elif disaster_date > event["end_date"]:
                date_distance = (
                    disaster_date - event["end_date"]
                ).days

            else:
                date_distance = 0

            matches.append({
                "desinventar_serial": record["Serial"],
                "disaster_date": disaster_date,
                "district": record["district"],
                "ds_division": record["ds_division"],
                "deaths": record["Deaths"],
                "injured": record["Injured"],
                "missing": record["Missing"],
                "houses_destroyed": record["Houses Destroyed"],
                "houses_damaged": record["Houses Damaged"],
                "directly_affected": record["Directly affected"],
                "indirectly_affected": record["Indirectly Affected"],
                "relocated": record["Relocated"],
                "evacuated": record["Evacuated"],
                "station_id": event["station_id"],
                "event_id": event["event_id"],
                "event_start": event["start_date"],
                "event_end": event["end_date"],
                "candidate_days": event["candidate_days"],
                "max_rainfall_3d": event["max_rainfall_3d"],
                "max_rainfall_7d": event["max_rainfall_7d"],
                "max_wl_zscore": event["max_wl_zscore"],
                "max_q_zscore": event["max_q_zscore"],
                "date_distance_days": date_distance
            })

matched_events = pd.DataFrame(matches)

print("Total DesInventar records:", len(des_overlap))
print("Matched DesInventar records:", 
      matched_events["desinventar_serial"].nunique()
      if len(matched_events) > 0 else 0)

print("Total station-event matches:", len(matched_events))

print("\nUnique DesInventar records WITHOUT a match:")

matched_serials = (
    set(matched_events["desinventar_serial"])
    if len(matched_events) > 0
    else set()
)

unmatched = des_overlap[
    ~des_overlap["Serial"].isin(matched_serials)
].copy()

print(len(unmatched))

print("\nUnmatched records:")
display(
    unmatched[
        ["Serial", "date", "district", "ds_division",
         "Deaths", "Houses Destroyed",
         "Houses Damaged", "Indirectly Affected"]
    ].sort_values("date")
)

print("\nSample matches:")
display(
    matched_events.sort_values(
        ["disaster_date", "station_id"]
    ).head(30)
)
```

```python
# --------------------------------------------------
# STEP 9C: INSPECT REGIONAL CLUSTERING OF
# HYDROLOGICAL CANDIDATE EVENTS
# --------------------------------------------------

historical_events = candidate_events[
    (candidate_events["start_date"] >= "2015-01-01") &
    (candidate_events["end_date"] <= "2019-12-31")
].copy()

print("Station-level candidate events, 2015–2019:",
      len(historical_events))

print("\nEvents by year:")
display(
    historical_events
    .assign(year=historical_events["start_date"].dt.year)
    .groupby("year")
    .size()
    .reset_index(name="station_event_count")
)

print("\nHistorical station events:")
display(
    historical_events[
        [
            "station_id",
            "event_id",
            "start_date",
            "end_date",
            "candidate_days",
            "max_rainfall_3d",
            "max_rainfall_7d",
            "max_wl_zscore",
            "max_q_zscore"
        ]
    ].sort_values(["start_date", "station_id"])
)
```

```python
# --------------------------------------------------
# STEP 10: BUILD REGIONAL FLOOD EPISODES
# --------------------------------------------------

events = historical_events.copy()

events = events.sort_values(
    ["start_date", "end_date", "station_id"]
).reset_index(drop=True)

# --------------------------------------------------
# Determine whether each event starts a new regional
# episode.
#
# If the next event starts more than 2 days after the
# latest end date of the current episode, start a new
# episode.
# --------------------------------------------------

regional_episode_numbers = []

current_episode = 0
current_episode_end = None

for _, row in events.iterrows():

    start = row["start_date"]
    end = row["end_date"]

    if current_episode_end is None:
        current_episode = 1
        current_episode_end = end

    elif start > current_episode_end + pd.Timedelta(days=2):
        current_episode += 1
        current_episode_end = end

    else:
        current_episode_end = max(current_episode_end, end)

    regional_episode_numbers.append(current_episode)

events["regional_episode_number"] = regional_episode_numbers

# Create readable episode IDs
events["regional_episode_id"] = (
    "KELANI_EP"
    + events["regional_episode_number"]
      .astype(str)
      .str.zfill(3)
)

# --------------------------------------------------
# Summarise each regional episode
# --------------------------------------------------

regional_episodes = (
    events
    .groupby("regional_episode_id", as_index=False)
    .agg(
        start_date=("start_date", "min"),
        end_date=("end_date", "max"),
        station_count=("station_id", "nunique"),
        station_event_count=("event_id", "count"),
        stations=("station_id", lambda x: ", ".join(sorted(set(x)))),
        max_rainfall_3d=("max_rainfall_3d", "max"),
        max_rainfall_7d=("max_rainfall_7d", "max"),
        max_wl_zscore=("max_wl_zscore", "max"),
        max_q_zscore=("max_q_zscore", "max")
    )
)

regional_episodes["duration_days"] = (
    regional_episodes["end_date"]
    - regional_episodes["start_date"]
).dt.days + 1

regional_episodes = regional_episodes.sort_values(
    "start_date"
).reset_index(drop=True)

# --------------------------------------------------
# Save
# --------------------------------------------------

regional_output_path = (
    project_root
    / "data/interim/flood/kelani/"
    / "flood_regional_episodes.parquet"
)

regional_episodes.to_parquet(
    regional_output_path,
    index=False
)

print("Regional flood episodes:", len(regional_episodes))
print("Saved to:")
print(regional_output_path)

print("\nRegional episodes:")
display(regional_episodes)
```

```python
# --------------------------------------------------
# STEP 9: VALIDATE REGIONAL EPISODES AGAINST DESINVENTAR
# --------------------------------------------------

# Load regional episodes
regional_episodes = pd.read_parquet(
    project_root
    / "data/interim/flood/kelani/"
    / "flood_regional_episodes.parquet"
)

print("Regional episodes:", len(regional_episodes))

# --------------------------------------------------
# Prepare DesInventar data
# --------------------------------------------------

desi = desinventar.copy()

desi["date"] = pd.to_datetime(desi["date"])

# Create a unique record identifier
desi["desinventar_id"] = desi["Serial"].astype(str)

# Keep the main impact fields
impact_columns = [
    "Deaths",
    "Injured",
    "Missing",
    "Houses Destroyed",
    "Houses Damaged",
    "Directly Affected",
    "Indirectly Affected",
    "Relocated",
    "Evacuated"
]

# Make sure numeric fields are numeric
for col in impact_columns:
    if col in desi.columns:
        desi[col] = pd.to_numeric(
            desi[col],
            errors="coerce"
        )

# --------------------------------------------------
# Match each DesInventar record to regional episodes
# --------------------------------------------------

validation_rows = []

for _, d in desi.iterrows():

    # Only consider regional episodes within ±7 days
    candidates = regional_episodes[
        (
            regional_episodes["start_date"]
            .between(
                d["date"] - pd.Timedelta(days=7),
                d["date"] + pd.Timedelta(days=7)
            )
        )
        |
        (
            regional_episodes["end_date"]
            .between(
                d["date"] - pd.Timedelta(days=7),
                d["date"] + pd.Timedelta(days=7)
            )
        )
        |
        (
            (regional_episodes["start_date"] <= d["date"])
            &
            (regional_episodes["end_date"] >= d["date"])
        )
    ].copy()

    if len(candidates) == 0:
        validation_rows.append({
            "desinventar_id": d["desinventar_id"],
            "disaster_date": d["date"],
            "district": d["district"],
            "ds_division": d["ds_division"],
            "deaths": d["Deaths"],
            "houses_destroyed": d["Houses Destroyed"],
            "houses_damaged": d["Houses Damaged"],
            "indirectly_affected": d["Indirectly Affected"],
            "regional_episode_id": pd.NA,
            "episode_start": pd.NaT,
            "episode_end": pd.NaT,
            "station_count": pd.NA,
            "station_event_count": pd.NA,
            "date_distance_days": pd.NA
        })

    else:

        # Calculate distance from DesInventar date
        candidates["date_distance_days"] = candidates.apply(
            lambda row: min(
                abs((d["date"] - row["start_date"]).days),
                abs((d["date"] - row["end_date"]).days)
            ),
            axis=1
        )

        # Choose closest regional episode
        best = candidates.sort_values(
            ["date_distance_days", "station_count"],
            ascending=[True, False]
        ).iloc[0]

        validation_rows.append({
            "desinventar_id": d["desinventar_id"],
            "disaster_date": d["date"],
            "district": d["district"],
            "ds_division": d["ds_division"],
            "deaths": d["Deaths"],
            "houses_destroyed": d["Houses Destroyed"],
            "houses_damaged": d["Houses Damaged"],
            "indirectly_affected": d["Indirectly Affected"],
            "regional_episode_id": best["regional_episode_id"],
            "episode_start": best["start_date"],
            "episode_end": best["end_date"],
            "station_count": best["station_count"],
            "station_event_count": best["station_event_count"],
            "date_distance_days": best["date_distance_days"]
        })

regional_validation = pd.DataFrame(validation_rows)

# --------------------------------------------------
# Validation summary
# --------------------------------------------------

regional_validation["matched"] = (
    regional_validation["regional_episode_id"].notna()
)

print("Total DesInventar records:", len(regional_validation))
print(
    "Matched to regional episode:",
    regional_validation["matched"].sum()
)
print(
    "Unmatched:",
    (~regional_validation["matched"]).sum()
)

print("\nMatch percentage:")
print(
    round(
        regional_validation["matched"].mean() * 100,
        2
    ),
    "%"
)

# --------------------------------------------------
# Display
# --------------------------------------------------

display(
    regional_validation[
        [
            "desinventar_id",
            "disaster_date",
            "district",
            "ds_division",
            "indirectly_affected",
            "regional_episode_id",
            "episode_start",
            "episode_end",
            "station_count",
            "date_distance_days",
            "matched"
        ]
    ].sort_values("disaster_date")
)
```

```python
# --------------------------------------------------
# VALIDATE REGIONAL EPISODES AGAINST DESINVENTAR
# --------------------------------------------------

# Load regional episodes
regional_episodes = pd.read_parquet(
    project_root
    / "data/interim/flood/kelani/"
    / "flood_regional_episodes.parquet"
)

print("Regional episodes:", len(regional_episodes))


# --------------------------------------------------
# Load DesInventar
# --------------------------------------------------

desi = desinventar_flood.copy()

# Convert date
desi["date"] = pd.to_datetime(
    desi["Date"],
    format="%Y/%m/%d",
    errors="coerce"
)

# Clean location fields
desi["district"] = (
    desi["District"]
    .astype("string")
    .str.strip()
)

desi["ds_division"] = (
    desi["Division"]
    .astype("string")
    .str.strip()
)

# Unique DesInventar record
desi["desinventar_id"] = (
    desi["Serial"].astype(str)
)


# --------------------------------------------------
# Convert impact columns to numeric
# --------------------------------------------------

impact_columns = [
    "Deaths",
    "Injured",
    "Missing",
    "Houses Destroyed",
    "Houses Damaged",
    "Directly affected",
    "Indirectly Affected",
    "Relocated",
    "Evacuated"
]

for col in impact_columns:
    desi[col] = pd.to_numeric(
        desi[col],
        errors="coerce"
    )


# --------------------------------------------------
# Keep 2015–2019 overlap
# --------------------------------------------------

desi_overlap = desi[
    (desi["date"] >= "2015-01-01") &
    (desi["date"] <= "2019-12-31")
].copy()

print(
    "DesInventar records in 2015–2019:",
    len(desi_overlap)
)


# --------------------------------------------------
# Match each DesInventar record to a regional episode
# --------------------------------------------------

validation_rows = []

for _, d in desi_overlap.iterrows():

    candidates = regional_episodes[
        (
            regional_episodes["start_date"]
            <= d["date"] + pd.Timedelta(days=7)
        )
        &
        (
            regional_episodes["end_date"]
            >= d["date"] - pd.Timedelta(days=7)
        )
    ].copy()

    # ----------------------------------------------
    # No regional episode found
    # ----------------------------------------------

    if len(candidates) == 0:

        validation_rows.append({
            "desinventar_id": d["desinventar_id"],
            "disaster_date": d["date"],
            "district": d["district"],
            "ds_division": d["ds_division"],
            "deaths": d["Deaths"],
            "houses_destroyed": d["Houses Destroyed"],
            "houses_damaged": d["Houses Damaged"],
            "indirectly_affected": d["Indirectly Affected"],
            "regional_episode_id": pd.NA,
            "episode_start": pd.NaT,
            "episode_end": pd.NaT,
            "station_count": pd.NA,
            "station_event_count": pd.NA,
            "date_distance_days": pd.NA
        })

    # ----------------------------------------------
    # Regional episode found
    # ----------------------------------------------

    else:

        # Distance from DesInventar event date
        candidates["date_distance_days"] = candidates.apply(
            lambda row: min(
                abs(
                    (d["date"] - row["start_date"]).days
                ),
                abs(
                    (d["date"] - row["end_date"]).days
                )
            ),
            axis=1
        )

        # Select closest episode.
        # If tied, prefer the episode involving
        # more stations.
        best = candidates.sort_values(
            [
                "date_distance_days",
                "station_count"
            ],
            ascending=[
                True,
                False
            ]
        ).iloc[0]

        validation_rows.append({
            "desinventar_id": d["desinventar_id"],
            "disaster_date": d["date"],
            "district": d["district"],
            "ds_division": d["ds_division"],
            "deaths": d["Deaths"],
            "houses_destroyed": d["Houses Destroyed"],
            "houses_damaged": d["Houses Damaged"],
            "indirectly_affected": d["Indirectly Affected"],
            "regional_episode_id": best["regional_episode_id"],
            "episode_start": best["start_date"],
            "episode_end": best["end_date"],
            "station_count": best["station_count"],
            "station_event_count": best["station_event_count"],
            "date_distance_days": best["date_distance_days"]
        })


# --------------------------------------------------
# Create validation dataframe
# --------------------------------------------------

regional_validation = pd.DataFrame(
    validation_rows
)

regional_validation["matched"] = (
    regional_validation["regional_episode_id"]
    .notna()
)


# --------------------------------------------------
# Summary
# --------------------------------------------------

total_records = len(regional_validation)

matched_records = (
    regional_validation["matched"].sum()
)

unmatched_records = (
    (~regional_validation["matched"]).sum()
)

match_percentage = (
    matched_records / total_records * 100
)


print("\n-----------------------------")
print("REGIONAL EPISODE VALIDATION")
print("-----------------------------")

print(
    "Total DesInventar records:",
    total_records
)

print(
    "Matched to regional episode:",
    matched_records
)

print(
    "Unmatched:",
    unmatched_records
)

print(
    "Match percentage:",
    round(match_percentage, 2),
    "%"
)


# --------------------------------------------------
# Display results
# --------------------------------------------------

display(
    regional_validation[
        [
            "desinventar_id",
            "disaster_date",
            "district",
            "ds_division",
            "indirectly_affected",
            "regional_episode_id",
            "episode_start",
            "episode_end",
            "station_count",
            "date_distance_days",
            "matched"
        ]
    ]
    .sort_values("disaster_date")
    .reset_index(drop=True)
)
```

```python
# --------------------------------------------------
# INSPECT UNMATCHED DESINVENTAR RECORDS
# --------------------------------------------------

unmatched = regional_validation[
    ~regional_validation["matched"]
].copy()

print("Unmatched records:", len(unmatched))

display(
    unmatched[
        [
            "desinventar_id",
            "disaster_date",
            "district",
            "ds_division",
            "deaths",
            "houses_destroyed",
            "houses_damaged",
            "indirectly_affected"
        ]
    ].sort_values("disaster_date")
)
```

```python
# --------------------------------------------------
# REGIONAL EPISODE MATCH SUMMARY
# --------------------------------------------------

episode_match_summary = (
    regional_validation[
        regional_validation["matched"]
    ]
    .groupby("regional_episode_id", as_index=False)
    .agg(
        desinventar_records=("desinventar_id", "count"),
        total_indirectly_affected=("indirectly_affected", "sum"),
        max_indirectly_affected=("indirectly_affected", "max"),
        ds_divisions=("ds_division", "nunique"),
        station_count=("station_count", "first"),
        episode_start=("episode_start", "first"),
        episode_end=("episode_end", "first")
    )
    .sort_values(
        ["desinventar_records", "max_indirectly_affected"],
        ascending=False
    )
)

display(episode_match_summary)
```

```python
# --------------------------------------------------
# INSPECT LONG REGIONAL EPISODES
# --------------------------------------------------

long_episode_ids = regional_episodes[
    regional_episodes["duration_days"] >= 7
]["regional_episode_id"].tolist()

print("Long episodes:")
print(long_episode_ids)

long_episode_details = (
    events[
        events["regional_episode_id"].isin(long_episode_ids)
    ]
    [
        [
            "regional_episode_id",
            "station_id",
            "event_id",
            "start_date",
            "end_date",
            "candidate_days",
            "max_rainfall_3d",
            "max_rainfall_7d",
            "max_wl_zscore",
            "max_q_zscore"
        ]
    ]
    .sort_values(
        [
            "regional_episode_id",
            "start_date",
            "station_id"
        ]
    )
    .reset_index(drop=True)
)

display(long_episode_details)
```

```python
# --------------------------------------------------
# STEP 11: CREATE CANDIDATE EPISODE-IMPACT TABLE
# --------------------------------------------------

# Merge regional episode information into
# the validated DesInventar records.

episode_impact = regional_validation[
    regional_validation["matched"]
].copy()

# Select useful regional episode features
episode_features = regional_episodes[
    [
        "regional_episode_id",
        "start_date",
        "end_date",
        "station_count",
        "station_event_count",
        "stations",
        "max_rainfall_3d",
        "max_rainfall_7d",
        "max_wl_zscore",
        "max_q_zscore",
        "duration_days"
    ]
].copy()

# Rename episode dates to avoid confusion
episode_features = episode_features.rename(
    columns={
        "start_date": "episode_start",
        "end_date": "episode_end"
    }
)

# Merge
episode_impact = episode_impact.merge(
    episode_features,
    on="regional_episode_id",
    how="left"
)

# --------------------------------------------------
# Historical impact fields
# --------------------------------------------------

episode_impact["human_impact_count"] = (
    episode_impact["indirectly_affected"]
)

# Binary indicators for historical impact
episode_impact["death_recorded"] = (
    episode_impact["deaths"].fillna(0) > 0
)

episode_impact["houses_destroyed_recorded"] = (
    episode_impact["houses_destroyed"].fillna(0) > 0
)

episode_impact["houses_damaged_recorded"] = (
    episode_impact["houses_damaged"].fillna(0) > 0
)

# --------------------------------------------------
# Sort
# --------------------------------------------------

episode_impact = episode_impact.sort_values(
    [
        "disaster_date",
        "district",
        "ds_division"
    ]
).reset_index(drop=True)

# --------------------------------------------------
# Summary
# --------------------------------------------------

print(
    "Candidate episode-impact records:",
    len(episode_impact)
)

print(
    "Unique regional episodes:",
    episode_impact["regional_episode_id"].nunique()
)

print(
    "Unique DS divisions:",
    episode_impact["ds_division"].nunique()
)

print("\nColumns:")
print(episode_impact.columns.tolist())

# --------------------------------------------------
# Display
# --------------------------------------------------

display(
    episode_impact[
        [
            "desinventar_id",
            "disaster_date",
            "district",
            "ds_division",
            "regional_episode_id",
            "episode_start",
            "episode_end",
            "station_count",
            "max_rainfall_3d",
            "max_rainfall_7d",
            "max_wl_zscore",
            "max_q_zscore",
            "human_impact_count",
            "death_recorded",
            "houses_destroyed_recorded",
            "houses_damaged_recorded"
        ]
    ]
)
```

```python
# STEP 22 — Aggregate GN-level Census data to DS level

# Count variables that should be summed from GN → DS
sum_columns = [
    "total_population",
    "male",
    "female",
    "age_total",
    "age_0_14",
    "age_15_59",
    "age_60_64",
    "age_65_plus",
    "elderly_60_plus",
    "occupied_housing_units",
    "number_of_households",

    # Water
    "water_protected_well",
    "water_semi_protected_well",
    "water_unprotected_well",
    "water_tube_well",
    "water_spring_fountain",
    "water_national_water_supply",
    "water_local_authority",
    "water_community_organization",
    "water_private_supply",
    "water_tank_river_stream",
    "water_rain_water",
    "water_bottled",
    "water_filter_ro_plant",
    "water_bowser",
    "water_other",

    # Cooking
    "cooking_firewood",
    "cooking_kerosene",
    "cooking_gas",
    "cooking_electricity",
    "cooking_sawdust_paddy_husk",
    "cooking_biogas",
    "cooking_other",
    "cooking_not_relevant",

    # Lighting
    "lighting_national_grid_hydro",
    "lighting_kerosene_lamp",
    "lighting_solar_grid",
    "lighting_solar_standalone",
    "lighting_biogas",
    "lighting_generator",
    "lighting_other",

    # Toilet
    "toilet_within_unit_exclusive",
    "toilet_within_unit_shared",
    "toilet_within_premises_exclusive",
    "toilet_within_premises_shared",
    "toilet_no_toilet_shared_elsewhere",
    "toilet_common_public",
    "toilet_no_toilet"
]

census_ds = (
    census_master
    .groupby(
        ["district_name", "ds_name_join", "ds_code"],
        as_index=False
    )[sum_columns]
    .sum(min_count=1)
)

print("DS-level Census rows:", len(census_ds))
print("DS-level Census columns:", len(census_ds.columns))

display(
    census_ds[
        [
            "district_name",
            "ds_name_join",
            "ds_code",
            "total_population",
            "male",
            "female",
            "age_0_14",
            "age_15_59",
            "age_60_64",
            "age_65_plus",
            "occupied_housing_units",
            "number_of_households"
        ]
    ].head(10)
)
```

```python
# STEP 23 — Create DS-level demographic H-E-S-C features

census_ds = census_ds.copy()

# --------------------------------------------------
# 1. Age-based sensitivity
# --------------------------------------------------

census_ds["children_0_14_ratio"] = (
    census_ds["age_0_14"] /
    census_ds["total_population"].replace(0, pd.NA)
)

census_ds["elderly_60_plus_ratio"] = (
    census_ds["elderly_60_plus"] /
    census_ds["total_population"].replace(0, pd.NA)
)

# --------------------------------------------------
# 2. Female population ratio
# --------------------------------------------------

census_ds["female_ratio"] = (
    census_ds["female"] /
    census_ds["total_population"].replace(0, pd.NA)
)

# --------------------------------------------------
# 3. Household / housing exposure
# --------------------------------------------------

census_ds["households_per_1000_population"] = (
    census_ds["number_of_households"] /
    census_ds["total_population"].replace(0, pd.NA)
) * 1000

census_ds["housing_units_per_1000_population"] = (
    census_ds["occupied_housing_units"] /
    census_ds["total_population"].replace(0, pd.NA)
) * 1000

# --------------------------------------------------
# 4. Basic household service indicators
# --------------------------------------------------

# Water sources that can indicate less reliable / more vulnerable access
census_ds["unprotected_well_households"] = (
    census_ds["water_unprotected_well"]
)

census_ds["river_stream_water_households"] = (
    census_ds["water_tank_river_stream"]
)

# Toilet access
census_ds["no_toilet_households"] = (
    census_ds["toilet_no_toilet"]
)

census_ds["public_toilet_households"] = (
    census_ds["toilet_common_public"]
)

# --------------------------------------------------
# 5. Convert service counts into proportions
# --------------------------------------------------

census_ds["unprotected_well_ratio"] = (
    census_ds["water_unprotected_well"] /
    census_ds["number_of_households"].replace(0, pd.NA)
)

census_ds["river_stream_water_ratio"] = (
    census_ds["water_tank_river_stream"] /
    census_ds["number_of_households"].replace(0, pd.NA)
)

census_ds["no_toilet_ratio"] = (
    census_ds["toilet_no_toilet"] /
    census_ds["number_of_households"].replace(0, pd.NA)
)

census_ds["public_toilet_ratio"] = (
    census_ds["toilet_common_public"] /
    census_ds["number_of_households"].replace(0, pd.NA)
)

# --------------------------------------------------
# Check
# --------------------------------------------------

feature_columns = [
    "total_population",
    "number_of_households",
    "occupied_housing_units",
    "children_0_14_ratio",
    "elderly_60_plus_ratio",
    "female_ratio",
    "households_per_1000_population",
    "housing_units_per_1000_population",
    "unprotected_well_ratio",
    "river_stream_water_ratio",
    "no_toilet_ratio",
    "public_toilet_ratio"
]

print("DS-level Census shape:", census_ds.shape)

display(
    census_ds[
        [
            "district_name",
            "ds_name_join",
            "ds_code"
        ] + feature_columns
    ].head(10)
)
```

```python
# STEP 25 — Combine Census + QGIS features for the 27 flood-study DS divisions

# Start from the QGIS-matched flood DS table
flood_ds_features = flood_ds_qgis.copy()

# Select the Census features we need
census_features = census_ds[
    [
        "district_name",
        "ds_name_join",
        "ds_code",
        "total_population",
        "male",
        "female",
        "age_0_14",
        "age_15_59",
        "age_60_64",
        "age_65_plus",
        "elderly_60_plus",
        "occupied_housing_units",
        "number_of_households",
        "children_0_14_ratio",
        "elderly_60_plus_ratio",
        "female_ratio",
        "households_per_1000_population",
        "housing_units_per_1000_population",
        "unprotected_well_ratio",
        "river_stream_water_ratio",
        "no_toilet_ratio",
        "public_toilet_ratio"
    ]
].copy()

# Merge using the validated standardized DS names
flood_ds_features = flood_ds_features.merge(
    census_features,
    left_on=["district", "ds_division"],
    right_on=["district_name", "ds_name_join"],
    how="left",
    validate="one_to_one"
)

# Check results
print("Flood DS feature table shape:", flood_ds_features.shape)

print(
    "Census population matched:",
    flood_ds_features["total_population"].notna().sum(),
    "/",
    len(flood_ds_features)
)

print(
    "Missing Census matches:",
    flood_ds_features["total_population"].isna().sum()
)

print("\nColumns:")
print(flood_ds_features.columns.tolist())

display(
    flood_ds_features[
        [
            "district",
            "ds_division",
            "adm3_pcode",
            "area_sqkm",
            "total_population",
            "number_of_households",
            "children_0_14_ratio",
            "elderly_60_plus_ratio",
            "female_ratio",
            "road_density",
            "facility_count",
            "facility_density",
            "Hub distance_HubDist"
        ]
    ].sort_values(
        ["district", "ds_division"]
    ).reset_index(drop=True)
)
```

```python
from pathlib import Path
import pandas as pd

landslide_path = Path(
    "/Users/ohaseelanka/disaster_analytics_hub/data/raw/landslide/DesInventar_NuwaraEliya_Landslides_page1.xlsx"
)

landslide_raw = pd.read_excel(landslide_path)

print("Shape:", landslide_raw.shape)
print("\nColumns:")
print(landslide_raw.columns.tolist())

print("\nFirst 10 rows:")
display(landslide_raw.head(10))
```

```python
# Step 2 — Clean landslide DesInventar data

landslide = landslide_raw.copy()

# Standardize column names
landslide.columns = (
    landslide.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

# Convert date
landslide["date"] = pd.to_datetime(
    landslide["date"],
    format="%Y/%m/%d",
    errors="coerce"
)

# Clean text fields
for col in ["event", "province", "district", "division"]:
    landslide[col] = (
        landslide[col]
        .astype("string")
        .str.strip()
    )

# Convert impact fields to numeric
impact_columns = [
    "deaths",
    "injured",
    "missing",
    "houses_destroyed",
    "houses_damaged",
    "directly_affected",
    "indirectly_affected",
    "relocated",
    "evacuated",
    "losses_$usd"
]

for col in impact_columns:
    landslide[col] = pd.to_numeric(
        landslide[col],
        errors="coerce"
    )

print("Cleaned shape:", landslide.shape)

print("\nDate range:")
print(landslide["date"].min(), "to", landslide["date"].max())

print("\nDistricts:")
print(landslide["district"].value_counts())

print("\nDS Divisions:")
print(landslide["division"].value_counts())

print("\nMissing values:")
display(landslide.isna().sum())
```

```python
# Step 3 — Landslide impact summary

impact_summary = pd.DataFrame({
    "records_with_value": [
        landslide["deaths"].notna().sum(),
        landslide["injured"].notna().sum(),
        landslide["missing"].notna().sum(),
        landslide["houses_destroyed"].notna().sum(),
        landslide["houses_damaged"].notna().sum(),
        landslide["indirectly_affected"].notna().sum(),
        landslide["directly_affected"].notna().sum(),
        landslide["relocated"].notna().sum(),
        landslide["evacuated"].notna().sum()
    ],
    "total_records": 100
}, index=[
    "Deaths",
    "Injured",
    "Missing",
    "Houses destroyed",
    "Houses damaged",
    "Indirectly affected",
    "Directly affected",
    "Relocated",
    "Evacuated"
])

impact_summary["missing_records"] = (
    impact_summary["total_records"]
    - impact_summary["records_with_value"]
)

display(impact_summary)

print("\nNumeric impact statistics:")
display(
    landslide[
        [
            "deaths",
            "injured",
            "missing",
            "houses_destroyed",
            "houses_damaged",
            "indirectly_affected",
            "relocated",
            "evacuated"
        ]
    ].describe()
)

print("\nRecords by year:")
display(
    landslide["date"]
    .dt.year
    .value_counts()
    .sort_index()
)

print("\nRecords by DS division:")
display(
    landslide["division"]
    .value_counts()
)
```

```python
# Step 4 — Inspect temporal clustering of landslide records

landslide_dates = (
    landslide
    .groupby("date")
    .agg(
        record_count=("serial", "count"),
        ds_count=("division", "nunique"),
        ds_divisions=("division", lambda x: ", ".join(sorted(x.unique())))
    )
    .reset_index()
    .sort_values("date")
)

print("Unique landslide dates:", len(landslide_dates))
print("Total landslide records:", len(landslide))

print("\nDates with multiple landslide records:")
display(
    landslide_dates[
        landslide_dates["record_count"] > 1
    ]
)

print("\nMaximum records on one date:")
print(landslide_dates["record_count"].max())
```

```python
# Step 5 — Examine gaps between unique landslide dates

landslide_dates = landslide_dates.sort_values("date").reset_index(drop=True)

landslide_dates["gap_days"] = (
    landslide_dates["date"].diff().dt.days
)

print("Gap statistics:")
display(
    landslide_dates["gap_days"]
    .describe()
)

print("\nSmallest gaps:")
display(
    landslide_dates[
        ["date", "record_count", "ds_count", "ds_divisions", "gap_days"]
    ].head(25)
)

print("\nLargest gaps:")
display(
    landslide_dates[
        ["date", "record_count", "ds_count", "ds_divisions", "gap_days"]
    ]
    .sort_values("gap_days", ascending=False)
    .head(20)
)

print("\nGap distribution:")
display(
    landslide_dates["gap_days"]
    .value_counts()
    .sort_index()
)
```

```python
# Step 6 — Construct candidate landslide episodes
# Rule: a gap > 3 days starts a new episode

landslide_events = landslide.copy()

landslide_events = (
    landslide_events
    .sort_values(["date", "division", "serial"])
    .reset_index(drop=True)
)

# Gap between consecutive reported landslide dates
landslide_events["date_gap"] = (
    landslide_events["date"].diff().dt.days
)

# New episode when gap is greater than 3 days
landslide_events["new_episode"] = (
    landslide_events["date_gap"].isna() |
    (landslide_events["date_gap"] > 3)
)

# Assign episode number
landslide_events["episode_number"] = (
    landslide_events["new_episode"]
    .cumsum()
)

# Create readable episode ID
landslide_events["episode_id"] = (
    "NE_LS_EP"
    + landslide_events["episode_number"]
      .astype(int)
      .astype(str)
      .str.zfill(3)
)

print("Total incident records:", len(landslide_events))
print(
    "Candidate landslide episodes:",
    landslide_events["episode_id"].nunique()
)

print("\nRecords per episode:")
display(
    landslide_events
    .groupby("episode_id")
    .size()
    .describe()
)
```

```python
# Step 7 — Inspect candidate landslide episodes

landslide_episode_summary = (
    landslide_events
    .groupby("episode_id", as_index=False)
    .agg(
        start_date=("date", "min"),
        end_date=("date", "max"),
        incident_count=("serial", "count"),
        ds_count=("division", "nunique"),
        ds_divisions=("division", lambda x: ", ".join(sorted(x.unique()))),
        total_deaths=("deaths", "sum", min_count=1),
        total_houses_destroyed=("houses_destroyed", "sum", min_count=1),
        total_houses_damaged=("houses_damaged", "sum", min_count=1),
        total_indirectly_affected=("indirectly_affected", "sum", min_count=1),
    )
)

landslide_episode_summary["duration_days"] = (
    landslide_episode_summary["end_date"]
    - landslide_episode_summary["start_date"]
).dt.days + 1

print("Candidate episodes:", len(landslide_episode_summary))

print("\nEpisodes with multiple incident records:")
display(
    landslide_episode_summary[
        landslide_episode_summary["incident_count"] > 1
    ].sort_values("start_date")
)

print("\nEpisodes involving multiple DS divisions:")
display(
    landslide_episode_summary[
        landslide_episode_summary["ds_count"] > 1
    ].sort_values("start_date")
)

print("\nEpisode duration statistics:")
display(
    landslide_episode_summary["duration_days"].describe()
)
```

```python
from math import radians, sin, cos, sqrt, atan2

# Norwood coordinate from Met_Stations.xlsx
norwood_lat = 6.8300
norwood_lon = 80.6000

ds_coords = ds_gis[
    ds_gis["adm2_name"]
    .astype("string")
    .str.contains("Nuwara", case=False, na=False)
][
    ["adm3_name", "adm3_pcode", "center_lat", "center_lon"]
].drop_duplicates().copy()

def haversine_km(lat1, lon1, lat2, lon2):
    R = 6371.0

    lat1, lon1, lat2, lon2 = map(
        radians,
        [lat1, lon1, lat2, lon2]
    )

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = (
        sin(dlat / 2) ** 2
        + cos(lat1) * cos(lat2) * sin(dlon / 2) ** 2
    )

    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    return R * c

ds_coords["distance_from_norwood_km"] = ds_coords.apply(
    lambda row: haversine_km(
        norwood_lat,
        norwood_lon,
        row["center_lat"],
        row["center_lon"]
    ),
    axis=1
)

print(
    ds_coords
    .sort_values("distance_from_norwood_km")
    .to_string(index=False)
)
```

```python
# Create explicit dekad start and end dates
# for the Nuwara Eliya rainfall observations.

rain_dekad = nuwara_rain.copy()

rain_dekad["dekad_start"] = rain_dekad["date"]

rain_dekad["dekad_end"] = (
    rain_dekad["date"].shift(-1) - pd.Timedelta(days=1)
)

# The final observation has no following rainfall observation,
# so its end date is set to the end of that calendar month.
last_idx = rain_dekad.index[-1]

rain_dekad.loc[last_idx, "dekad_end"] = (
    rain_dekad.loc[last_idx, "date"]
    + pd.offsets.MonthEnd(0)
)

print("Rainfall dekad table shape:", rain_dekad.shape)

print("\nFirst 6 dekads:")
display(
    rain_dekad[
        [
            "date",
            "dekad_start",
            "dekad_end",
            "rfh",
            "r1h",
            "r3h",
            "rfq",
            "r1q",
            "r3q"
        ]
    ].head(6)
)
```

```python
# Match all 62 landslide episodes to overlapping rainfall dekads

rainfall_features = [
    "rfh",
    "r1h",
    "r3h",
    "rfq",
    "r1q",
    "r3q"
]

matched_rows = []

for _, episode in landslide_episode_summary.iterrows():

    episode_start = episode["start_date"]
    episode_end = episode["end_date"]

    # Find rainfall dekads that overlap the landslide episode
    matching_rain = rain_dekad[
        (rain_dekad["dekad_start"] <= episode_end) &
        (rain_dekad["dekad_end"] >= episode_start)
    ].copy()

    row = {
        "episode_id": episode["episode_id"],
        "start_date": episode_start,
        "end_date": episode_end,
        "incident_count": episode["incident_count"],
        "ds_count": episode["ds_count"],
        "ds_divisions": episode["ds_divisions"],
        "total_deaths": episode["total_deaths"],
        "total_houses_destroyed": episode["total_houses_destroyed"],
        "total_houses_damaged": episode["total_houses_damaged"],
        "total_indirectly_affected": episode["total_indirectly_affected"],
        "duration_days": episode["duration_days"],
        "rainfall_dekads_matched": len(matching_rain)
    }

    # Store rainfall features
    for feature in rainfall_features:
        row[f"event_{feature}_max"] = matching_rain[feature].max()

    # Keep the rainfall dates that were matched
    row["matched_rainfall_dates"] = ", ".join(
        matching_rain["date"].dt.strftime("%Y-%m-%d")
    )

    matched_rows.append(row)


landslide_rainfall_matched = pd.DataFrame(matched_rows)

print("Matched dataset shape:", landslide_rainfall_matched.shape)

print("\nEpisodes with rainfall matches:")
print(
    landslide_rainfall_matched["rainfall_dekads_matched"]
    .value_counts()
    .sort_index()
)

print("\nEpisodes without rainfall matches:")
print(
    landslide_rainfall_matched[
        landslide_rainfall_matched["rainfall_dekads_matched"] == 0
    ]["episode_id"].tolist()
)

print("\nFirst 10 matched episodes:")
display(
    landslide_rainfall_matched[
        [
            "episode_id",
            "start_date",
            "end_date",
            "rainfall_dekads_matched",
            "event_rfh_max",
            "event_r1h_max",
            "event_r3h_max",
            "event_rfq_max",
            "event_r1q_max",
            "event_r3q_max",
            "total_indirectly_affected"
        ]
    ].head(10)
)
```

```python
# ---------------------------------------------------------
# Step 4: Create PRE-EVENT rainfall features
# ---------------------------------------------------------

# Make sure rainfall is sorted chronologically
rain_dekad = (
    rain_dekad
    .sort_values("dekad_start")
    .reset_index(drop=True)
)

# Use the dekad containing the START of each landslide episode.
# This prevents rainfall after the initial event from being
# used as a predictor.

pre_event_rows = []

for _, episode in landslide_episode_summary.iterrows():

    event_start = episode["start_date"]

    # Find the rainfall dekad containing the episode start
    current_idx = rain_dekad.index[
        (rain_dekad["dekad_start"] <= event_start) &
        (rain_dekad["dekad_end"] >= event_start)
    ]

    if len(current_idx) == 0:
        pre_event_rows.append({
            "episode_id": episode["episode_id"]
        })
        continue

    current_idx = current_idx[0]

    # Previous rainfall dekads
    previous_1 = (
        rain_dekad.iloc[current_idx - 1]
        if current_idx >= 1 else None
    )

    previous_2 = (
        rain_dekad.iloc[max(0, current_idx - 2):current_idx]
    )

    previous_3 = (
        rain_dekad.iloc[max(0, current_idx - 3):current_idx]
    )

    row = {
        "episode_id": episode["episode_id"],
        "event_start": event_start,

        # Rainfall dekad immediately before the event
        "pre_rfh_1dekad": (
            previous_1["rfh"]
            if previous_1 is not None else None
        ),

        # Rainfall accumulated over previous 2 dekads
        "pre_rfh_2dekad_sum": (
            previous_2["rfh"].sum()
            if len(previous_2) > 0 else None
        ),

        # Rainfall accumulated over previous 3 dekads
        "pre_rfh_3dekad_sum": (
            previous_3["rfh"].sum()
            if len(previous_3) > 0 else None
        ),

        # Conditions immediately before the event
        "pre_r1h": (
            previous_1["r1h"]
            if previous_1 is not None else None
        ),

        "pre_r3h": (
            previous_1["r3h"]
            if previous_1 is not None else None
        ),

        "pre_rfq": (
            previous_1["rfq"]
            if previous_1 is not None else None
        ),

        "pre_r1q": (
            previous_1["r1q"]
            if previous_1 is not None else None
        ),

        "pre_r3q": (
            previous_1["r3q"]
            if previous_1 is not None else None
        ),

        # Keep track of which rainfall dekad was used
        "pre_event_rainfall_date": (
            previous_1["date"]
            if previous_1 is not None else None
        )
    }

    pre_event_rows.append(row)


landslide_pre_event_rainfall = pd.DataFrame(pre_event_rows)

print(
    "Pre-event rainfall table shape:",
    landslide_pre_event_rainfall.shape
)

print("\nMissing values:")
display(
    landslide_pre_event_rainfall.isna().sum()
)

print("\nFirst 10 episodes:")
display(
    landslide_pre_event_rainfall.head(10)
)
```

```python
# ---------------------------------------------------------
# Step 5: Validate pre-event rainfall features
# ---------------------------------------------------------

rainfall_feature_cols = [
    "pre_rfh_1dekad",
    "pre_rfh_2dekad_sum",
    "pre_rfh_3dekad_sum",
    "pre_r1h",
    "pre_r3h",
    "pre_rfq",
    "pre_r1q",
    "pre_r3q"
]

print("Number of episodes:", len(landslide_pre_event_rainfall))

print("\nMissing values:")
display(
    landslide_pre_event_rainfall[rainfall_feature_cols]
    .isna()
    .sum()
)

print("\nInfinite values:")
print(
    landslide_pre_event_rainfall[rainfall_feature_cols]
    .isin([float("inf"), float("-inf")])
    .sum()
)

print("\nDescriptive statistics:")
display(
    landslide_pre_event_rainfall[rainfall_feature_cols]
    .describe()
    .T
)

print("\nHighest pre-event 10-day rainfall:")
display(
    landslide_pre_event_rainfall[
        [
            "episode_id",
            "event_start",
            "pre_event_rainfall_date",
            "pre_rfh_1dekad",
            "pre_rfh_2dekad_sum",
            "pre_rfh_3dekad_sum",
            "pre_rfq",
            "total_indirectly_affected"
        ]
    ]
    .sort_values("pre_rfh_1dekad", ascending=False)
    .head(10)
)
```

```python
# ---------------------------------------------------------
# Step 6E — Load QGIS DS spatial features
# ---------------------------------------------------------

from pathlib import Path
import pandas as pd

qgis_path = Path(
    "/Users/ohaseelanka/Desktop/disaster_project_gis/ds79_feature_matrix.csv"
)

qgis_ds = pd.read_csv(qgis_path)

print("QGIS feature matrix shape:", qgis_ds.shape)

print("\nNuwara Eliya DS records:")

nuwara_qgis = qgis_ds[
    qgis_ds["adm2_name"].astype("string").str.strip() == "Nuwara Eliya"
].copy()

display(
    nuwara_qgis[
        [
            "adm3_name",
            "adm3_pcode",
            "area_sqkm",
            "center_lat",
            "center_lon",
            "total_road_m",
            "road_density",
            "facility_count",
            "facility_density",
            "Hub distance_HubDist"
        ]
    ]
    .sort_values("adm3_name")
    .reset_index(drop=True)
)
```

```python
# ---------------------------------------------------------
# Step 6F — Aggregate QGIS spatial features
# ---------------------------------------------------------

# Exact QGIS → harmonized landslide DS mapping
qgis_landslide_mapping = {
    "Ambagamuwa Korale": "Ambagamuwa",
    "Hanguranketa": "Hanguranketha",
    "Kothmale East": "Kothmale",
    "Kothmale West": "Kothmale",
    "Nuwara Eliya": "Nuwara Eliya",
    "Walapane": "Walapane"
}

# Select only the QGIS records relevant to the landslide study
landslide_qgis = nuwara_qgis[
    nuwara_qgis["adm3_name"].isin(qgis_landslide_mapping.keys())
].copy()

# Create harmonized DS name
landslide_qgis["landslide_ds"] = (
    landslide_qgis["adm3_name"]
    .map(qgis_landslide_mapping)
)

# ---------------------------------------------------------
# Aggregate spatial count/extent variables
# ---------------------------------------------------------

landslide_qgis_ds = (
    landslide_qgis
    .groupby("landslide_ds", as_index=False)
    .agg(
        area_sqkm=("area_sqkm", "sum"),
        total_road_m=("total_road_m", "sum"),
        facility_count=("facility_count", "sum")
    )
)

# ---------------------------------------------------------
# Recalculate densities
# ---------------------------------------------------------

landslide_qgis_ds["road_density"] = (
    landslide_qgis_ds["total_road_m"]
    / landslide_qgis_ds["area_sqkm"]
)

landslide_qgis_ds["facility_density"] = (
    landslide_qgis_ds["facility_count"]
    / landslide_qgis_ds["area_sqkm"]
)

# ---------------------------------------------------------
# Calculate area-weighted geographic centre
# ---------------------------------------------------------

landslide_qgis["area_weight"] = landslide_qgis["area_sqkm"]

centres = []

for ds_name, group in landslide_qgis.groupby("landslide_ds"):

    total_area = group["area_sqkm"].sum()

    weighted_lat = (
        (group["center_lat"] * group["area_sqkm"]).sum()
        / total_area
    )

    weighted_lon = (
        (group["center_lon"] * group["area_sqkm"]).sum()
        / total_area
    )

    centres.append({
        "landslide_ds": ds_name,
        "center_lat": weighted_lat,
        "center_lon": weighted_lon
    })

landslide_centres = pd.DataFrame(centres)

# Combine spatial features with weighted centres
landslide_qgis_ds = landslide_qgis_ds.merge(
    landslide_centres,
    on="landslide_ds",
    how="left"
)

# ---------------------------------------------------------
# Validation
# ---------------------------------------------------------

print("Final spatial DS shape:", landslide_qgis_ds.shape)

print("\nSpatial DS profiles:")

display(
    landslide_qgis_ds
    .sort_values("landslide_ds")
    .reset_index(drop=True)
)
```

```python
# ---------------------------------------------------------
# Step 6G — Combine Census + QGIS DS features
# ---------------------------------------------------------

landslide_ds_features = landslide_census_ds.merge(
    landslide_qgis_ds,
    on="landslide_ds",
    how="inner",
    validate="one_to_one"
)

print("Combined DS feature table shape:")
print(landslide_ds_features.shape)

print("\nDS names:")
print(
    landslide_ds_features["landslide_ds"]
    .sort_values()
    .tolist()
)

print("\nMissing values:")
display(
    landslide_ds_features.isna().sum()
    .sort_values(ascending=False)
    .head(20)
)

print("\nSelected exposure/vulnerability/spatial features:")
display(
    landslide_ds_features[
        [
            "landslide_ds",
            "total_population",
            "number_of_households",
            "children_0_14_ratio",
            "elderly_60_plus_ratio",
            "female_ratio",
            "unprotected_well_ratio",
            "river_stream_water_ratio",
            "no_toilet_ratio",
            "public_toilet_ratio",
            "area_sqkm",
            "total_road_m",
            "road_density",
            "facility_count",
            "facility_density",
            "center_lat",
            "center_lon"
        ]
    ]
    .sort_values("landslide_ds")
    .reset_index(drop=True)
)
```

```python
# ---------------------------------------------------------
# Step 7A — Inspect raw landslide DS names
# ---------------------------------------------------------

print("Raw DS names in landslide event data:\n")

display(
    landslide_events["division"]
    .value_counts(dropna=False)
    .rename_axis("raw_division")
    .reset_index(name="record_count")
)
```

```python
# ---------------------------------------------------------
# Step 7B — Create Event × DS observations
# ---------------------------------------------------------

landslide_event_ds = landslide_events.copy()

# Harmonize historical DS names
landslide_event_ds["landslide_ds"] = (
    landslide_event_ds["division"]
    .replace({
        "Ambagamuwakorale": "Ambagamuwa",
        "Kothmale": "Kothmale",
        "Nuwara Eliya": "Nuwara Eliya",
        "Walapane": "Walapane",
        "Hanguranketha": "Hanguranketha"
    })
)

# Check that every historical record mapped correctly
print("Unmapped DS records:")
print(
    landslide_event_ds[
        ~landslide_event_ds["landslide_ds"].isin(
            landslide_ds_features["landslide_ds"]
        )
    ]["division"]
    .value_counts()
)

# Create event × DS observations
landslide_event_ds_summary = (
    landslide_event_ds
    .groupby(
        ["episode_id", "landslide_ds"],
        as_index=False
    )
    .agg(
        start_date=("date", "min"),
        end_date=("date", "max"),
        incident_count=("serial", "count"),

        deaths=("deaths", "sum"),
        injured=("injured", "sum"),
        missing=("missing", "sum"),
        houses_destroyed=("houses_destroyed", "sum"),
        houses_damaged=("houses_damaged", "sum"),
        indirectly_affected=("indirectly_affected", "sum"),
        evacuated=("evacuated", "sum"),

        target_records=("indirectly_affected", "count")
    )
)

landslide_event_ds_summary["duration_days"] = (
    landslide_event_ds_summary["end_date"]
    - landslide_event_ds_summary["start_date"]
).dt.days + 1

print("\nEvent × DS shape:")
print(landslide_event_ds_summary.shape)

print("\nNumber of unique episodes:")
print(landslide_event_ds_summary["episode_id"].nunique())

print("\nNumber of unique DS divisions:")
print(landslide_event_ds_summary["landslide_ds"].nunique())

print("\nEvent × DS records by DS:")
display(
    landslide_event_ds_summary["landslide_ds"]
    .value_counts()
    .rename_axis("landslide_ds")
    .reset_index(name="event_ds_count")
)

print("\nFirst records:")
display(
    landslide_event_ds_summary
    .sort_values(["start_date", "landslide_ds"])
    .head(15)
)
```

```python
# ---------------------------------------------------------
# Step 7C — Audit human-impact target completeness
# ---------------------------------------------------------

target_audit = (
    landslide_event_ds_summary
    .assign(
        target_missing=lambda df:
            df["target_records"] == 0
    )
)

print("Total event × DS observations:")
print(len(target_audit))

print("\nTarget recorded:")
print(
    (~target_audit["target_missing"]).sum()
)

print("\nTarget missing:")
print(
    target_audit["target_missing"].sum()
)

print("\nTarget completeness percentage:")
print(
    round(
        (~target_audit["target_missing"]).mean() * 100,
        2
    ),
    "%"
)

print("\nTarget status by DS:")
display(
    target_audit
    .groupby("landslide_ds")
    .agg(
        event_ds_count=("episode_id", "count"),
        target_recorded=("target_missing", lambda x: (~x).sum()),
        target_missing=("target_missing", "sum")
    )
    .reset_index()
)

print("\nObservations with missing target:")
display(
    target_audit[
        target_audit["target_missing"]
    ][
        [
            "episode_id",
            "landslide_ds",
            "start_date",
            "end_date",
            "incident_count",
            "deaths",
            "houses_destroyed",
            "houses_damaged",
            "indirectly_affected",
            "target_records"
        ]
    ]
    .sort_values(["start_date", "landslide_ds"])
    .reset_index(drop=True)
)
```

```python
# ---------------------------------------------------------
# Step 7D — Prepare clean landslide target
# ---------------------------------------------------------

landslide_target = landslide_event_ds_summary.copy()

# Target completeness flag
landslide_target["target_complete"] = (
    landslide_target["target_records"] > 0
)

# Keep the original target untouched
landslide_target["human_impact_count"] = (
    landslide_target["indirectly_affected"]
)

# Separate model-usable observations
landslide_target_model = landslide_target[
    landslide_target["target_complete"]
].copy()

print("Full event × DS dataset:")
print(landslide_target.shape)

print("\nTarget-complete observations:")
print(landslide_target_model.shape)

print("\nTarget-incomplete observations:")
print(
    (~landslide_target["target_complete"]).sum()
)

print("\nHuman-impact target statistics:")
display(
    landslide_target_model[
        "human_impact_count"
    ].describe()
)

print("\nSorted human-impact values:")
display(
    landslide_target_model[
        [
            "episode_id",
            "landslide_ds",
            "start_date",
            "incident_count",
            "human_impact_count"
        ]
    ]
    .sort_values(
        "human_impact_count",
        ascending=False
    )
    .reset_index(drop=True)
)
```

```python
# ---------------------------------------------------------
# Step 7E — Inspect human-impact target distribution
# ---------------------------------------------------------

target_values = (
    landslide_target_model["human_impact_count"]
    .sort_values()
    .reset_index(drop=True)
)

# Calculate useful empirical percentiles
target_percentiles = (
    landslide_target_model["human_impact_count"]
    .quantile(
        [0, 0.10, 0.25, 0.50, 0.75, 0.90, 0.95, 1.00]
    )
    .rename("human_impact_count")
    .reset_index()
    .rename(columns={"index": "percentile"})
)

print("Empirical target percentiles:")
display(target_percentiles)

print("\nTarget value frequency groups:")

target_frequency = (
    landslide_target_model
    .groupby("human_impact_count")
    .size()
    .reset_index(name="observation_count")
    .sort_values(
        "human_impact_count",
        ascending=False
    )
)

display(target_frequency.head(20))

print("\nNumber of observations by broad impact range:")

impact_ranges = pd.cut(
    landslide_target_model["human_impact_count"],
    bins=[-1, 10, 50, 100, 500, 1000, float("inf")],
    labels=[
        "0–10",
        "11–50",
        "51–100",
        "101–500",
        "501–1000",
        "1001+"
    ]
)

display(
    impact_ranges
    .value_counts(sort=False)
    .rename_axis("impact_range")
    .reset_index(name="observation_count")
)
```

```python
# ---------------------------------------------------------
# Step 7F — Provisional 3-class human-impact target
# ---------------------------------------------------------

landslide_target_model = landslide_target_model.copy()

landslide_target_model["impact_class"] = pd.cut(
    landslide_target_model["human_impact_count"],
    bins=[
        -float("inf"),
        18,
        75,
        float("inf")
    ],
    labels=[
        "LOW",
        "MODERATE",
        "HIGH"
    ],
    include_lowest=True
)

print("Provisional class distribution:")
display(
    landslide_target_model["impact_class"]
    .value_counts()
    .sort_index()
    .rename_axis("impact_class")
    .reset_index(name="observation_count")
)

print("\nClass percentages:")
display(
    (
        landslide_target_model["impact_class"]
        .value_counts(normalize=True)
        .sort_index()
        .mul(100)
        .round(2)
        .rename_axis("impact_class")
        .reset_index(name="percentage")
    )
)

print("\nImpact range within each class:")
display(
    landslide_target_model
    .groupby("impact_class", observed=False)["human_impact_count"]
    .agg(
        count="count",
        minimum="min",
        median="median",
        maximum="max"
    )
    .reset_index()
)
```

```python
# ---------------------------------------------------------
# Step 7G — Check impact classes within historical episodes
# ---------------------------------------------------------

episode_class_check = (
    landslide_target_model
    .groupby("episode_id")
    .agg(
        ds_count=("landslide_ds", "count"),
        ds_list=("landslide_ds", lambda x: ", ".join(sorted(x.unique()))),
        class_count=("impact_class", "nunique"),
        classes=("impact_class", lambda x: ", ".join(sorted(x.astype(str).unique()))),
        min_impact=("human_impact_count", "min"),
        max_impact=("human_impact_count", "max")
    )
    .reset_index()
)

print("Total target-complete episodes:")
print(len(episode_class_check))

print("\nEpisodes involving multiple DS divisions:")
print(
    (episode_class_check["ds_count"] > 1).sum()
)

print("\nEpisodes containing more than one impact class:")
print(
    (episode_class_check["class_count"] > 1).sum()
)

print("\nEpisodes with multiple DS divisions:")
display(
    episode_class_check[
        episode_class_check["ds_count"] > 1
    ]
    .sort_values("episode_id")
)

print("\nEpisodes containing multiple impact classes:")
display(
    episode_class_check[
        episode_class_check["class_count"] > 1
    ]
    .sort_values("episode_id")
)
```

```python
# ---------------------------------------------------------
# Step 7H — Merge pre-event rainfall features
# ---------------------------------------------------------

rainfall_features = landslide_pre_event_rainfall[
    [
        "episode_id",
        "pre_rfh_1dekad",
        "pre_rfh_2dekad_sum",
        "pre_r1h",
        "pre_r3h",
        "pre_rfq",
        "pre_r1q",
        "pre_r3q",
        "pre_event_rainfall_date"
    ]
].copy()

landslide_model_base = landslide_target_model.merge(
    rainfall_features,
    on="episode_id",
    how="left",
    validate="many_to_one"
)

print("Dataset shape after rainfall merge:")
print(landslide_model_base.shape)

print("\nMissing rainfall features:")
display(
    landslide_model_base[
        [
            "pre_rfh_1dekad",
            "pre_rfh_2dekad_sum",
            "pre_r1h",
            "pre_r3h",
            "pre_rfq",
            "pre_r1q",
            "pre_r3q"
        ]
    ]
    .isna()
    .sum()
)

print("\nRainfall + target preview:")
display(
    landslide_model_base[
        [
            "episode_id",
            "landslide_ds",
            "start_date",
            "human_impact_count",
            "impact_class",
            "pre_rfh_1dekad",
            "pre_rfh_2dekad_sum",
            "pre_r1h",
            "pre_r3h",
            "pre_rfq",
            "pre_r1q",
            "pre_r3q"
        ]
    ]
    .sort_values(["start_date", "landslide_ds"])
    .head(15)
)
```

```python
# ---------------------------------------------------------
# Step 7I — Merge Census + QGIS DS features
# ---------------------------------------------------------

landslide_model_dataset = landslide_model_base.merge(
    landslide_ds_features,
    on="landslide_ds",
    how="left",
    validate="many_to_one"
)

print("Final merged dataset shape:")
print(landslide_model_dataset.shape)

print("\nMissing values after DS feature merge:")

display(
    landslide_model_dataset
    .isna()
    .sum()
    .sort_values(ascending=False)
    .head(25)
)

print("\nDS coverage:")
display(
    landslide_model_dataset[
        "landslide_ds"
    ]
    .value_counts()
    .rename_axis("landslide_ds")
    .reset_index(name="observation_count")
)

print("\nFinal dataset preview:")

display(
    landslide_model_dataset[
        [
            "episode_id",
            "landslide_ds",
            "start_date",
            "human_impact_count",
            "impact_class",

            # Hazard
            "pre_rfh_1dekad",
            "pre_rfh_2dekad_sum",
            "pre_r1h",
            "pre_r3h",
            "pre_rfq",
            "pre_r1q",
            "pre_r3q",

            # Exposure
            "total_population",
            "number_of_households",

            # Demographic sensitivity
            "children_0_14_ratio",
            "elderly_60_plus_ratio",
            "female_ratio",

            # Household sensitivity
            "unprotected_well_ratio",
            "river_stream_water_ratio",
            "no_toilet_ratio",
            "public_toilet_ratio",

            # Spatial / infrastructure
            "area_sqkm",
            "road_density",
            "facility_density"
        ]
    ]
    .sort_values(["start_date", "landslide_ds"])
    .head(15)
)
```

```python
# ---------------------------------------------------------
# Step 7J — Select initial modelling features
# ---------------------------------------------------------

initial_features = [
    # Hazard / rainfall
    "pre_rfh_1dekad",
    "pre_rfh_2dekad_sum",
    "pre_r1h",
    "pre_r3h",
    "pre_rfq",
    "pre_r1q",
    "pre_r3q",

    # Exposure
    "total_population",
    "number_of_households",

    # Demographic sensitivity
    "children_0_14_ratio",
    "elderly_60_plus_ratio",
    "female_ratio",

    # Household sensitivity
    "unprotected_well_ratio",
    "river_stream_water_ratio",
    "no_toilet_ratio",
    "public_toilet_ratio",

    # Spatial / infrastructure
    "area_sqkm",
    "road_density",
    "facility_density"
]

target_columns = [
    "human_impact_count",
    "impact_class"
]

identifier_columns = [
    "episode_id",
    "landslide_ds",
    "start_date",
    "end_date",
    "incident_count",
    "target_complete"
]

landslide_model_features = landslide_model_dataset[
    identifier_columns
    + initial_features
    + target_columns
].copy()

print("Initial modelling dataset shape:")
print(landslide_model_features.shape)

print("\nNumber of predictors:")
print(len(initial_features))

print("\nPredictors:")
for i, feature in enumerate(initial_features, start=1):
    print(f"{i:02d}. {feature}")

print("\nMissing values:")
display(
    landslide_model_features[
        initial_features + target_columns
    ]
    .isna()
    .sum()
    .sort_values(ascending=False)
)

print("\nDataset preview:")
display(
    landslide_model_features
    .sort_values(["start_date", "landslide_ds"])
    .head(10)
)
```

```python
# ---------------------------------------------------------
# Step 7K — Feature variation and correlation audit
# ---------------------------------------------------------

import numpy as np
import pandas as pd

X_audit = landslide_model_features[initial_features].copy()

# ---------------------------------------------------------
# 1. Number of unique values
# ---------------------------------------------------------

feature_variation = pd.DataFrame({
    "feature": X_audit.columns,
    "unique_values": [
        X_audit[col].nunique()
        for col in X_audit.columns
    ],
    "missing_values": [
        X_audit[col].isna().sum()
        for col in X_audit.columns
    ]
})

feature_variation["variation_type"] = np.where(
    feature_variation["unique_values"] <= 5,
    "Very limited",
    "Variable"
)

print("Feature variation:")
display(
    feature_variation
    .sort_values(["unique_values", "feature"])
    .reset_index(drop=True)
)

# ---------------------------------------------------------
# 2. Correlation matrix
# ---------------------------------------------------------

correlation_matrix = X_audit.corr(method="spearman")

print("\nSpearman correlation matrix:")

display(
    correlation_matrix.round(3)
)

# ---------------------------------------------------------
# 3. Strongly correlated feature pairs
# ---------------------------------------------------------

correlation_pairs = []

for i in range(len(correlation_matrix.columns)):
    for j in range(i + 1, len(correlation_matrix.columns)):

        feature_1 = correlation_matrix.columns[i]
        feature_2 = correlation_matrix.columns[j]

        correlation_value = correlation_matrix.iloc[i, j]

        if abs(correlation_value) >= 0.80:

            correlation_pairs.append({
                "feature_1": feature_1,
                "feature_2": feature_2,
                "spearman_correlation": correlation_value
            })

strong_correlations = (
    pd.DataFrame(correlation_pairs)
    .sort_values(
        "spearman_correlation",
        key=lambda x: x.abs(),
        ascending=False
    )
    if correlation_pairs
    else pd.DataFrame(
        columns=[
            "feature_1",
            "feature_2",
            "spearman_correlation"
        ]
    )
)

print("\nStrong correlation pairs (|ρ| >= 0.80):")

display(strong_correlations)

# ---------------------------------------------------------
# 4. Class distribution
# ---------------------------------------------------------

print("\nImpact-class distribution:")

display(
    landslide_model_features["impact_class"]
    .value_counts()
    .rename_axis("impact_class")
    .reset_index(name="count")
)
```

```python
# ---------------------------------------------------------
# Step 7L — Create compact modelling feature set
# ---------------------------------------------------------

core_features = [
    # Hazard
    "pre_rfh_1dekad",
    "pre_r1h",
    "pre_r3h",
    "pre_r1q",

    # Exposure
    "total_population",

    # Demographic sensitivity
    "children_0_14_ratio",
    "elderly_60_plus_ratio",

    # Household sensitivity
    "no_toilet_ratio",

    # Infrastructure
    "road_density",
    "facility_density"
]

landslide_core_dataset = landslide_model_features[
    [
        "episode_id",
        "landslide_ds",
        "start_date",
        "end_date",
        "human_impact_count",
        "impact_class"
    ]
    + core_features
].copy()

print("Core modelling dataset shape:")
print(landslide_core_dataset.shape)

print("\nNumber of predictors:")
print(len(core_features))

print("\nCore predictors:")
for i, feature in enumerate(core_features, start=1):
    print(f"{i:02d}. {feature}")

print("\nMissing values:")
display(
    landslide_core_dataset[
        core_features + ["human_impact_count", "impact_class"]
    ]
    .isna()
    .sum()
)

print("\nCorrelation among core predictors:")

display(
    landslide_core_dataset[
        core_features
    ]
    .corr(method="spearman")
    .round(3)
)

print("\nImpact-class distribution:")

display(
    landslide_core_dataset["impact_class"]
    .value_counts()
    .rename_axis("impact_class")
    .reset_index(name="count")
)

print("\nCore dataset preview:")

display(
    landslide_core_dataset
    .sort_values(["start_date", "landslide_ds"])
    .head(10)
)
```

```python
# ---------------------------------------------------------
# Step 7M — Final primary modelling feature set
# ---------------------------------------------------------

final_features = [
    # Hazard
    "pre_rfh_1dekad",
    "pre_r1h",
    "pre_r3h",

    # Rainfall anomaly
    "pre_r1q",

    # Exposure
    "total_population",

    # Demographic sensitivity
    "children_0_14_ratio",
    "elderly_60_plus_ratio",

    # Household sensitivity
    "no_toilet_ratio",

    # Infrastructure / capacity proxy
    "facility_density"
]

landslide_final_dataset = landslide_core_dataset[
    [
        "episode_id",
        "landslide_ds",
        "start_date",
        "end_date",
        "human_impact_count",
        "impact_class"
    ]
    + final_features
].copy()

print("Final primary dataset shape:")
print(landslide_final_dataset.shape)

print("\nNumber of predictors:")
print(len(final_features))

print("\nFinal predictors:")
for i, feature in enumerate(final_features, start=1):
    print(f"{i:02d}. {feature}")

print("\nMissing values:")
display(
    landslide_final_dataset[
        final_features + [
            "human_impact_count",
            "impact_class"
        ]
    ]
    .isna()
    .sum()
)

print("\nFinal feature correlation:")

display(
    landslide_final_dataset[
        final_features
    ]
    .corr(method="spearman")
    .round(3)
)

print("\nImpact-class distribution:")

display(
    landslide_final_dataset["impact_class"]
    .value_counts()
    .rename_axis("impact_class")
    .reset_index(name="count")
)

print("\nUnique episodes:")
print(
    landslide_final_dataset["episode_id"].nunique()
)

print("\nUnique DS divisions:")
print(
    landslide_final_dataset["landslide_ds"].nunique()
)
```

```python
# ---------------------------------------------------------
# Step 7N — Check grouped stratified cross-validation
# ---------------------------------------------------------

from sklearn.model_selection import StratifiedGroupKFold

X = landslide_final_dataset[final_features].copy()
y = landslide_final_dataset["impact_class"].copy()
groups = landslide_final_dataset["episode_id"].copy()

print("Total observations:", len(X))
print("Unique episodes:", groups.nunique())

print("\nOverall class distribution:")
print(y.value_counts())

# ---------------------------------------------------------
# Create grouped stratified folds
# ---------------------------------------------------------

sgkf = StratifiedGroupKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

fold_summary = []

for fold_number, (train_idx, test_idx) in enumerate(
    sgkf.split(X, y, groups),
    start=1
):

    train_groups = set(groups.iloc[train_idx])
    test_groups = set(groups.iloc[test_idx])

    overlap = train_groups.intersection(test_groups)

    train_classes = y.iloc[train_idx].value_counts()
    test_classes = y.iloc[test_idx].value_counts()

    fold_summary.append({
        "fold": fold_number,
        "train_observations": len(train_idx),
        "test_observations": len(test_idx),
        "train_episodes": len(train_groups),
        "test_episodes": len(test_groups),
        "group_overlap": len(overlap),
        "train_low": train_classes.get("LOW", 0),
        "train_moderate": train_classes.get("MODERATE", 0),
        "train_high": train_classes.get("HIGH", 0),
        "test_low": test_classes.get("LOW", 0),
        "test_moderate": test_classes.get("MODERATE", 0),
        "test_high": test_classes.get("HIGH", 0)
    })

fold_summary = pd.DataFrame(fold_summary)

print("\nGrouped cross-validation summary:")
display(fold_summary)

print("\nMaximum group overlap across folds:")
print(fold_summary["group_overlap"].max())
```

```python
# Reconstruct the Sheet3 report structure.
# This does NOT modify the original Excel file.
# It only creates a working cleaned dataframe.

import pandas as pd
import numpy as np

# Start from the raw Sheet3 data
data = sheet3.iloc[2:].copy()

# Rename important columns
data = data.rename(columns={
    0: "province",
    1: "district",
    2: "ds_division",
    3: "disaster",
    4: "date",
    5: "affected_families",
    6: "affected_people",
    7: "deaths",
    8: "injured",
    9: "missing",
    10: "houses_fully_damaged",
    11: "houses_partially_damaged",
    12: "safety_locations_count",
    14: "safety_families",
    15: "safety_persons",
    16: "safety_locations"
})

# Replace completely blank cells with NaN
data = data.replace(r"^\s*$", np.nan, regex=True)

# Clean text columns
text_columns = [
    "province",
    "district",
    "ds_division",
    "disaster",
    "date"
]

for col in text_columns:
    data[col] = (
        data[col]
        .astype("string")
        .str.strip()
    )

# A report block starts when at least one of the main
# report-identifying fields is present.
# Blank rows separate blocks.

data["is_blank_row"] = data[
    ["province", "district", "ds_division", "disaster", "date"]
].isna().all(axis=1)

# Create a report-block ID.
# Every completely blank row starts a new separator.
data["report_block"] = data["is_blank_row"].cumsum()

# Within each block, inherit report-level information
# from the first non-empty value.
for col in ["province", "district", "disaster", "date"]:
    data[col] = data.groupby("report_block")[col].transform(
        lambda x: x.ffill().bfill()
    )

# Convert numeric impact columns
numeric_columns = [
    "affected_families",
    "affected_people",
    "deaths",
    "injured",
    "missing",
    "houses_fully_damaged",
    "houses_partially_damaged",
    "safety_locations_count",
    "safety_families",
    "safety_persons"
]

for col in numeric_columns:
    data[col] = pd.to_numeric(
        data[col],
        errors="coerce"
    )

# Extract the START date from values such as:
# 2020.05.19 - 2020.05.22
data["start_date"] = pd.to_datetime(
    data["date"]
    .astype("string")
    .str.extract(r"(\d{4}\.\d{2}\.\d{2})")[0],
    format="%Y.%m.%d",
    errors="coerce"
)

# Standardize disaster names
data["disaster_standardized"] = (
    data["disaster"]
    .astype("string")
    .str.replace("\n", " ", regex=False)
    .str.strip()
    .str.lower()
)

data["disaster_standardized"] = data["disaster_standardized"].replace({
    "flash flood": "flash flood",
    "land slide": "landslide",
    "landslide risk": "landslide risk",
    "flood risk": "flood risk",
    "flood and landslide": "flood and landslide",
    "flood": "flood"
})

print("Cleaned Sheet3 shape:", data.shape)

print("\nReport blocks:")
print("Number of report blocks:", data["report_block"].nunique())

print("\nStandardized disaster types:")
print(
    data["disaster_standardized"]
    .value_counts(dropna=False)
)

print("\nRows with DS division after reconstruction:")
print(
    data["ds_division"]
    .notna()
    .sum()
)

print("\nDS-level rows by year after reconstruction:")
print(
    data.loc[data["ds_division"].notna(), "start_date"]
    .dt.year
    .value_counts()
    .sort_index()
)
```

```python
# Inspect the 16 report blocks in Sheet3.
# This lets us distinguish:
# 1. confirmed disaster reports
# 2. risk/status reports
# 3. repeated/update reports
# 4. combined flood + landslide reports

block_summary = (
    data.groupby("report_block", dropna=False)
    .agg(
        province=("province", "first"),
        district=("district", "first"),
        disaster=("disaster", "first"),
        start_date=("start_date", "first"),
        raw_date=("date", "first"),
        total_rows=("report_block", "size"),
        ds_rows=("ds_division", lambda x: x.notna().sum()),
    )
    .reset_index()
)

print(block_summary.to_string(index=False))
```

```python
# Find repeated DS-level observations for the same
# district + DS division + date + disaster.
#
# These may represent multiple situation-report updates
# rather than independent disaster events.

ds_data = data[
    data["ds_division"].notna()
].copy()

# Clean text for comparison
for col in ["district", "ds_division", "disaster_standardized"]:
    ds_data[col] = (
        ds_data[col]
        .astype("string")
        .str.replace("\n", " ", regex=False)
        .str.strip()
    )

# Create a date key
ds_data["event_date"] = ds_data["start_date"]

# Count occurrences of the same
# district + DS + date + disaster
duplicate_counts = (
    ds_data
    .groupby(
        [
            "district",
            "ds_division",
            "event_date",
            "disaster_standardized"
        ],
        dropna=False
    )
    .size()
    .reset_index(name="record_count")
)

duplicates = duplicate_counts[
    duplicate_counts["record_count"] > 1
].copy()

print("Total unique DS/date/disaster combinations:",
      len(duplicate_counts))

print("Repeated combinations:",
      len(duplicates))

print("\nRepeated combinations:")
print(
    duplicates
    .sort_values(
        ["event_date", "district", "ds_division"]
    )
    .to_string(index=False)
)

print("\nTotal DS-level records involved in repeated combinations:",
      duplicates["record_count"].sum())
```

```python
import pandas as pd
from urllib.parse import urljoin

reports = []

# Find every report row on the current page
for row in soup.find_all("tr"):

    cells = row.find_all("td")

    # A valid report row has at least 4 cells
    if len(cells) >= 4:

        report_type = cells[0].get_text(" ", strip=True)
        report_date = cells[1].get_text(" ", strip=True)
        report_time = cells[2].get_text(" ", strip=True)

        download_link = cells[3].find("a", href=True)

        if download_link and ".pdf" in download_link["href"].lower():

            pdf_url = urljoin(
                base_url,
                download_link["href"]
            )

            reports.append({
                "report_type": report_type,
                "report_date": report_date,
                "report_time": report_time,
                "pdf_url": pdf_url
            })

reports_df = pd.DataFrame(reports)

print("Reports extracted:", len(reports_df))

print("\nFirst 10 reports:")
display(reports_df.head(10))
```

```python
import time
import requests
import pandas as pd
from bs4 import BeautifulSoup
from urllib.parse import urljoin
import urllib3

urllib3.disable_warnings(urllib3.exceptions.InsecureRequestWarning)

BASE_URL = "https://www.drrweb.dmc.gov.lk/index.php"

all_reports = []

# Start from the first page
limit = 30
start = 0

session = requests.Session()
session.headers.update({
    "User-Agent": "Mozilla/5.0"
})

while True:

    page_url = (
        f"{BASE_URL}"
        f"?Itemid=273"
        f"&lang=en"
        f"&limit={limit}"
        f"&limitstart={start}"
        f"&option=com_dmcreports"
        f"&report_type_id=1"
        f"&view=reports"
    )

    try:
        page_response = session.get(
            page_url,
            timeout=30,
            verify=False
        )

        print(
            f"Page start={start} | "
            f"Status={page_response.status_code}"
        )

        if page_response.status_code != 200:
            print("Stopping because the page request failed.")
            break

        page_soup = BeautifulSoup(
            page_response.text,
            "html.parser"
        )

        page_count = 0

        for row in page_soup.find_all("tr"):

            cells = row.find_all("td")

            if len(cells) < 4:
                continue

            report_type = cells[0].get_text(
                " ",
                strip=True
            )

            report_date = cells[1].get_text(
                " ",
                strip=True
            )

            report_time = cells[2].get_text(
                " ",
                strip=True
            )

            download_link = cells[3].find(
                "a",
                href=True
            )

            if (
                download_link
                and ".pdf" in download_link["href"].lower()
            ):

                pdf_url = urljoin(
                    BASE_URL,
                    download_link["href"]
                )

                all_reports.append({
                    "report_type": report_type,
                    "report_date": report_date,
                    "report_time": report_time,
                    "pdf_url": pdf_url
                })

                page_count += 1

        print(
            f"  Reports found on page: {page_count}"
        )

        # Stop when no report rows are found
        if page_count == 0:
            print("No more reports found.")
            break

        # Move to next page
        start += limit

        # Small delay to avoid sending requests too quickly
        time.sleep(0.5)

        # Safety limit — prevents accidental endless crawling
        if start > 6000:
            print("Safety limit reached.")
            break

    except Exception as e:
        print("Error:", repr(e))
        break


# Convert to DataFrame
all_reports_df = pd.DataFrame(all_reports)

# Clean dates
all_reports_df["report_date"] = pd.to_datetime(
    all_reports_df["report_date"],
    errors="coerce"
)

# Remove exact duplicate PDF URLs
all_reports_df = (
    all_reports_df
    .drop_duplicates(subset=["pdf_url"])
    .reset_index(drop=True)
)

print("\n" + "=" * 60)
print("CRAWLING COMPLETE")
print("=" * 60)

print("Total unique reports:", len(all_reports_df))

print(
    "Latest report:",
    all_reports_df["report_date"].max()
)

print(
    "Earliest report:",
    all_reports_df["report_date"].min()
)

print("\nReport types:")
print(
    all_reports_df["report_type"]
    .str.strip()
    .str.lower()
    .value_counts()
)

print("\nReports by year:")
print(
    all_reports_df["report_date"]
    .dt.year
    .value_counts()
    .sort_index()
)
```

```python
# Inspect DMC report types by year
# This cell only analyzes the metadata index.
# It does NOT download any PDFs.

type_by_year = (
    all_reports_df
    .assign(
        year=all_reports_df["report_date"].dt.year,
        report_type_clean=(
            all_reports_df["report_type"]
            .astype("string")
            .str.strip()
            .str.lower()
        )
    )
    .groupby(
        ["year", "report_type_clean"],
        dropna=False
    )
    .size()
    .reset_index(name="count")
    .sort_values(
        ["year", "count"],
        ascending=[True, False]
    )
)

display(type_by_year)

print("\nNumber of distinct report types:", 
      type_by_year["report_type_clean"].nunique())

print("\nReport types containing 'flood':")
display(
    type_by_year[
        type_by_year["report_type_clean"].str.contains(
            "flood",
            na=False
        )
    ]
)

print("\nReport types containing 'landslide':")
display(
    type_by_year[
        type_by_year["report_type_clean"].str.contains(
            "landslide",
            na=False
        )
    ]
)
```

```python
# Identify potentially relevant DMC reports for flood/landslide extraction.
# No PDFs are downloaded in this cell.

dmc = all_reports_df.copy()

# Clean metadata
dmc["report_type_clean"] = (
    dmc["report_type"]
    .astype("string")
    .str.strip()
    .str.lower()
)

dmc["pdf_filename"] = (
    dmc["pdf_url"]
    .astype("string")
    .str.split("/")
    .str[-1]
    .str.lower()
)

dmc["year"] = dmc["report_date"].dt.year

# Restrict to years available for the historical study
dmc_historical = dmc[
    dmc["year"].between(2018, 2025)
].copy()

# Terms that indicate situation reports rather than weather/drought products
situation_terms = [
    "situation report",
    "situation reports",
    "situation summary report",
    "situation summary",
    "daily situation report",
    "summary situation update",
    "special situation report"
]

situation_mask = (
    dmc_historical["report_type_clean"]
    .str.contains(
        "|".join(situation_terms),
        regex=True,
        na=False
    )
)

# Explicit flood/landslide terms in report type or PDF filename
hazard_mask = (
    dmc_historical["report_type_clean"]
    .str.contains(
        "flood|landslide|land slide",
        regex=True,
        na=False
    )
    |
    dmc_historical["pdf_filename"]
    .str.contains(
        "flood|landslide|land_slide|land-slide",
        regex=True,
        na=False
    )
)

# Candidate set:
# - situation-report type, OR
# - explicitly named flood/landslide report
candidate_reports = dmc_historical[
    situation_mask | hazard_mask
].copy()

candidate_reports = (
    candidate_reports
    .drop_duplicates(subset=["pdf_url"])
    .sort_values(
        ["report_date", "report_time"],
        ascending=[False, False]
    )
    .reset_index(drop=True)
)

print("=" * 70)
print("DMC HISTORICAL CANDIDATE REPORTS")
print("=" * 70)

print("\nHistorical reports available:", len(dmc_historical))
print("Potentially relevant reports:", len(candidate_reports))

print("\nCandidate reports by year:")
display(
    candidate_reports["year"]
    .value_counts()
    .sort_index()
)

print("\nCandidate reports by report type:")
display(
    candidate_reports["report_type_clean"]
    .value_counts()
)

print("\nExplicit flood/landslide matches:")
display(
    candidate_reports[
        candidate_reports["report_type_clean"].str.contains(
            "flood|landslide|land slide",
            regex=True,
            na=False
        )
        |
        candidate_reports["pdf_filename"].str.contains(
            "flood|landslide|land_slide|land-slide",
            regex=True,
            na=False
        )
    ][
        [
            "report_date",
            "report_time",
            "report_type",
            "pdf_filename"
        ]
    ].head(50)
)
```

```python
# Select a small representative sample of DMC PDFs for extraction testing.
# No PDFs are downloaded in this cell.

dmc = all_reports_df.copy()

dmc["report_type_clean"] = (
    dmc["report_type"]
    .astype("string")
    .str.strip()
    .str.lower()
)

dmc["pdf_filename"] = (
    dmc["pdf_url"]
    .astype("string")
    .str.split("/")
    .str[-1]
)

dmc["pdf_filename_clean"] = (
    dmc["pdf_filename"]
    .astype("string")
    .str.lower()
)

dmc["year"] = dmc["report_date"].dt.year

# ---------------------------------------------------------
# 1. Explicit flood report
# ---------------------------------------------------------

flood_reports = dmc[
    dmc["report_type_clean"].str.contains(
        "flood",
        na=False
    )
    |
    dmc["pdf_filename_clean"].str.contains(
        "flood",
        na=False
    )
].copy()

# ---------------------------------------------------------
# 2. Explicit landslide report
# ---------------------------------------------------------

landslide_reports = dmc[
    dmc["report_type_clean"].str.contains(
        "landslide|land slide",
        regex=True,
        na=False
    )
    |
    dmc["pdf_filename_clean"].str.contains(
        "landslide|land_slide|land-slide",
        regex=True,
        na=False
    )
].copy()

# ---------------------------------------------------------
# 3. Generic situation reports
#    Exclude obvious drought/weather reports
# ---------------------------------------------------------

generic_situation = dmc[
    dmc["report_type_clean"].str.contains(
        "situation",
        na=False
    )
    &
    ~dmc["report_type_clean"].str.contains(
        "drought|dry weather|weather forecast|advisory|warning",
        regex=True,
        na=False
    )
].copy()

# ---------------------------------------------------------
# Select representative reports
# ---------------------------------------------------------

samples = []

# One explicit flood report from 2018
if len(flood_reports) > 0:
    samples.append(
        flood_reports
        .sort_values("report_date")
        .iloc[0]
    )

# One explicit landslide report
if len(landslide_reports) > 0:
    samples.append(
        landslide_reports
        .sort_values("report_date")
        .iloc[0]
    )

# One generic report from 2019
generic_2019 = generic_situation[
    generic_situation["year"] == 2019
]

if len(generic_2019) > 0:
    samples.append(
        generic_2019
        .sort_values("report_date")
        .iloc[0]
    )

# One generic report from 2020
generic_2020 = generic_situation[
    generic_situation["year"] == 2020
]

if len(generic_2020) > 0:
    samples.append(
        generic_2020
        .sort_values("report_date")
        .iloc[0]
    )

# One generic report from 2024
generic_2024 = generic_situation[
    generic_situation["year"] == 2024
]

if len(generic_2024) > 0:
    samples.append(
        generic_2024
        .sort_values("report_date")
        .iloc[0]
    )

# One generic report from 2025
generic_2025 = generic_situation[
    generic_situation["year"] == 2025
]

if len(generic_2025) > 0:
    samples.append(
        generic_2025
        .sort_values("report_date")
        .iloc[0]
    )

sample_reports = (
    pd.DataFrame(samples)
    .drop_duplicates(subset=["pdf_url"])
    .reset_index(drop=True)
)

print("=" * 70)
print("REPRESENTATIVE DMC PDFS FOR EXTRACTION TESTING")
print("=" * 70)

print("\nNumber of explicit flood-related reports:", len(flood_reports))
print("Number of explicit landslide-related reports:", len(landslide_reports))
print("Number of generic situation reports after exclusions:", len(generic_situation))

print("\nSelected sample PDFs:", len(sample_reports))

display(
    sample_reports[
        [
            "report_date",
            "report_time",
            "report_type",
            "pdf_filename",
            "pdf_url"
        ]
    ]
)
```

```python
# Download only the 6 representative PDFs for extraction testing.
# This is a small test batch — NOT the full DMC dataset.

from pathlib import Path
import requests
import time
import urllib3

urllib3.disable_warnings(
    urllib3.exceptions.InsecureRequestWarning
)

# Create test download directory
test_pdf_dir = (
    Path("../data/interim/disaster_records/dmc_test_pdfs")
    .resolve()
)

test_pdf_dir.mkdir(
    parents=True,
    exist_ok=True
)

session = requests.Session()

session.headers.update({
    "User-Agent": "Mozilla/5.0"
})

download_results = []

for i, row in sample_reports.iterrows():

    pdf_url = row["pdf_url"]

    filename = str(row["pdf_filename"])

    output_path = test_pdf_dir / filename

    print(f"\n[{i + 1}/{len(sample_reports)}]")
    print("Downloading:", filename)

    try:

        response = session.get(
            pdf_url,
            timeout=60,
            verify=False
        )

        content_type = response.headers.get(
            "Content-Type",
            ""
        )

        file_size = len(response.content)

        if response.status_code == 200 and file_size > 1000:

            output_path.write_bytes(
                response.content
            )

            status = "downloaded"

        else:

            status = "failed"

        download_results.append({
            "report_date": row["report_date"],
            "report_time": row["report_time"],
            "report_type": row["report_type"],
            "filename": filename,
            "status": status,
            "status_code": response.status_code,
            "content_type": content_type,
            "size_bytes": file_size,
            "path": str(output_path)
        })

        print("Status code:", response.status_code)
        print("Content-Type:", content_type)
        print("Size:", file_size, "bytes")
        print("Result:", status)

    except Exception as e:

        download_results.append({
            "report_date": row["report_date"],
            "report_time": row["report_time"],
            "report_type": row["report_type"],
            "filename": filename,
            "status": "error",
            "status_code": None,
            "content_type": None,
            "size_bytes": 0,
            "path": str(output_path),
            "error": repr(e)
        })

        print("ERROR:", repr(e))

    time.sleep(1)

download_results_df = pd.DataFrame(
    download_results
)

print("\n" + "=" * 70)
print("DOWNLOAD TEST COMPLETE")
print("=" * 70)

display(
    download_results_df[
        [
            "report_date",
            "report_type",
            "filename",
            "status",
            "status_code",
            "content_type",
            "size_bytes"
        ]
    ]
)

print("\nFiles saved to:")
print(test_pdf_dir)
```

```python
# Test whether the downloaded DMC PDFs contain extractable text.
# This does NOT modify the PDF files.

from pathlib import Path
import fitz
import pandas as pd

pdf_files = sorted(
    test_pdf_dir.glob("*.pdf")
)

pdf_text_test = []

for pdf_path in pdf_files:

    try:

        document = fitz.open(pdf_path)

        page_count = document.page_count

        page_text_lengths = []
        total_text = ""

        for page in document:

            text = page.get_text("text")

            page_text_lengths.append(
                len(text.strip())
            )

            total_text += text

        document.close()

        total_characters = len(
            total_text.strip()
        )

        pages_with_text = sum(
            length > 20
            for length in page_text_lengths
        )

        pdf_text_test.append({
            "filename": pdf_path.name,
            "pages": page_count,
            "pages_with_text": pages_with_text,
            "total_text_characters": total_characters,
            "average_chars_per_page": round(
                total_characters / page_count,
                1
            ) if page_count else 0
        })

    except Exception as e:

        pdf_text_test.append({
            "filename": pdf_path.name,
            "pages": None,
            "pages_with_text": None,
            "total_text_characters": None,
            "average_chars_per_page": None,
            "error": repr(e)
        })

pdf_text_test_df = pd.DataFrame(
    pdf_text_test
)

print("=" * 70)
print("PDF TEXT EXTRACTION TEST")
print("=" * 70)

display(pdf_text_test_df)
```

```python
# Search the 2018 flood report for pages containing impact-related information.
# This helps us locate the actual historical impact tables.

import pymupdf
import pandas as pd

document = pymupdf.open(flood_pdf)

keywords = [
    "affected",
    "famil",
    "people",
    "death",
    "injur",
    "missing",
    "house",
    "damaged",
    "destroyed",
    "evacuat",
    "relocat",
    "district",
    "division",
    "DS Division"
]

matches = []

for page_number, page in enumerate(document):

    text = page.get_text("text")

    text_lower = text.lower()

    matched_keywords = [
        keyword
        for keyword in keywords
        if keyword.lower() in text_lower
    ]

    if matched_keywords:

        matches.append({
            "page": page_number + 1,
            "matched_keywords": ", ".join(matched_keywords),
            "characters": len(text.strip())
        })

document.close()

matches_df = pd.DataFrame(matches)

print("=" * 80)
print("IMPACT-RELATED CONTENT FOUND IN 2018 FLOOD REPORT")
print("=" * 80)

display(matches_df)
```

```python
# Find impact-related sections in pages 2–5 of the 2018 flood report.
# This avoids printing the entire PDF.

import pymupdf

document = pymupdf.open(flood_pdf)

impact_terms = [
    "affected",
    "families",
    "people",
    "death",
    "injured",
    "missing",
    "houses",
    "damaged",
    "destroyed",
    "evacuated",
    "relocated"
]

for page_number in [2, 3, 4, 5]:

    page = document[page_number - 1]

    lines = [
        line.strip()
        for line in page.get_text("text").splitlines()
        if line.strip()
    ]

    print("\n")
    print("=" * 90)
    print(f"PAGE {page_number} — IMPACT-RELATED LINES")
    print("=" * 90)

    found = False

    for i, line in enumerate(lines):

        line_lower = line.lower()

        if any(
            term in line_lower
            for term in impact_terms
        ):

            found = True

            start = max(0, i - 2)
            end = min(len(lines), i + 4)

            print(
                "\n".join(lines[start:end])
            )
            print("-" * 70)

    if not found:
        print("No impact-related terms found.")

document.close()
```

```python
import pandas as pd
import numpy as np

# Load the two authentic datasets
census_ds = pd.read_parquet(
    "/Users/ohaseelanka/disaster_analytics_hub/data/processed/census_ds_features.parquet"
)

qgis_ds = pd.read_csv(
    "/Users/ohaseelanka/Desktop/disaster_project_gis/ds79_feature_matrix.csv"
)

# ------------------------------------------------------------
# Standardize names for matching
# ------------------------------------------------------------

def clean_name(value):
    if pd.isna(value):
        return np.nan

    return (
        str(value)
        .strip()
        .lower()
        .replace(" ", "")
        .replace("-", "")
        .replace("_", "")
    )

census_ds["ds_match"] = census_ds["ds_name_join"].apply(clean_name)
qgis_ds["ds_match"] = qgis_ds["adm3_name"].apply(clean_name)

# ------------------------------------------------------------
# Focus on the flood-study districts
# ------------------------------------------------------------

flood_districts = [
    "Colombo",
    "Gampaha",
    "Kegalle",
    "Ratnapura",
    "Nuwara Eliya"
]

census_flood = census_ds[
    census_ds["district_name"].isin(flood_districts)
].copy()

qgis_flood = qgis_ds[
    qgis_ds["adm2_name"].isin(flood_districts)
].copy()

# ------------------------------------------------------------
# Match
# ------------------------------------------------------------

matched = census_flood.merge(
    qgis_flood,
    on="ds_match",
    how="inner",
    suffixes=("_census", "_qgis")
)

# ------------------------------------------------------------
# Summary
# ------------------------------------------------------------

print("=" * 90)
print("AUTHENTIC FLOOD DS MATCHING")
print("=" * 90)

print("\nCensus flood DS records:", len(census_flood))
print("QGIS flood DS records:", len(qgis_flood))
print("Matched DS records:", len(matched))

print("\nMatched DS by district:")
print(
    matched["district_name"]
    .value_counts()
    .sort_index()
    .to_string()
)

print("\n" + "-" * 90)
print("MATCHED DS DIVISIONS")
print("-" * 90)

display(
    matched[
        [
            "district_name",
            "ds_name_join",
            "adm3_name",
            "adm3_pcode",
            "area_sqkm",
            "total_road_m",
            "road_density",
            "facility_count",
            "facility_density"
        ]
    ]
    .sort_values(
        ["district_name", "ds_name_join"]
    )
    .reset_index(drop=True)
)

print("\n" + "=" * 90)
print("MATCHING CHECK COMPLETE")
print("=" * 90)
```

```python
# Identify DS divisions that exist in one authentic source
# but not the other.

census_names = set(census_flood["ds_match"].dropna())
qgis_names = set(qgis_flood["ds_match"].dropna())

census_only = census_flood[
    ~census_flood["ds_match"].isin(qgis_names)
].copy()

qgis_only = qgis_flood[
    ~qgis_flood["ds_match"].isin(census_names)
].copy()

print("=" * 90)
print("UNMATCHED AUTHENTIC DS DIVISIONS")
print("=" * 90)

print("\nCENSUS DS → NOT FOUND IN QGIS")
print("-" * 90)

if len(census_only) > 0:
    display(
        census_only[
            [
                "district_name",
                "ds_name_join",
                "ds_code",
                "total_population"
            ]
        ]
        .sort_values(
            ["district_name", "ds_name_join"]
        )
        .reset_index(drop=True)
    )
else:
    print("None")

print("\nQGIS DS → NOT FOUND IN CENSUS")
print("-" * 90)

if len(qgis_only) > 0:
    display(
        qgis_only[
            [
                "adm2_name",
                "adm3_name",
                "adm3_pcode",
                "area_sqkm"
            ]
        ]
        .sort_values(
            ["adm2_name", "adm3_name"]
        )
        .reset_index(drop=True)
    )
else:
    print("None")

print("\n" + "=" * 90)
print("COUNTS")
print("=" * 90)

print("Census-only:", len(census_only))
print("QGIS-only:", len(qgis_only))
print("Total unmatched:", len(census_only) + len(qgis_only))
```

```python
# ============================================================
# CELL 5 — STANDARDIZE FLOOD DS NAMES
# ============================================================

# Copy the existing tables so the original data remains unchanged
census_flood_std = census_flood.copy()
qgis_flood_std = qgis_flood.copy()

# ------------------------------------------------------------
# Standardization mapping
# Census spelling → common standardized spelling
# ------------------------------------------------------------

ds_name_mapping = {
    "Aranayaka": "Aranayake",
    "Bulathkohupitiya": "Bulathkohipitiya",
    "Dehiovita": "Dehiowita",
    "Yatiyanthota": "Yatiyantota",
    "Ambagamuwa Koralaya": "Ambagamuwa Korale",
    "Hanguranketha": "Hanguranketa",
    "Thalawakelle": "Thalawakele",
    "Kahawatta": "Kahawattha",
    "Kaltota": "Kalthota",
    "Opanayaka": "Opanayake"
}

# Apply mapping only to the Census side
census_flood_std["ds_match_std"] = (
    census_flood_std["ds_name_join"]
    .replace(ds_name_mapping)
)

# QGIS already contains the desired standardized names
qgis_flood_std["ds_match_std"] = (
    qgis_flood_std["adm3_name"]
    .astype("string")
    .str.strip()
)

# ------------------------------------------------------------
# Check matching after standardization
# ------------------------------------------------------------

census_names_std = set(
    census_flood_std["ds_match_std"].dropna()
)

qgis_names_std = set(
    qgis_flood_std["ds_match_std"].dropna()
)

census_only_std = census_flood_std[
    ~census_flood_std["ds_match_std"].isin(qgis_names_std)
].copy()

qgis_only_std = qgis_flood_std[
    ~qgis_flood_std["ds_match_std"].isin(census_names_std)
].copy()

matched_std = census_flood_std[
    census_flood_std["ds_match_std"].isin(qgis_names_std)
].copy()

print("=" * 90)
print("FLOOD DS MATCHING AFTER NAME STANDARDIZATION")
print("=" * 90)

print("\nCensus flood DS:", len(census_flood_std))
print("QGIS flood DS:", len(qgis_flood_std))
print("Matched Census DS:", len(matched_std))
print("Census-only:", len(census_only_std))
print("QGIS-only:", len(qgis_only_std))

print("\n" + "=" * 90)
print("REMAINING UNMATCHED RECORDS")
print("=" * 90)

if len(census_only_std) > 0:
    print("\nCENSUS ONLY")
    display(
        census_only_std[
            [
                "district_name",
                "ds_name_join",
                "ds_match_std"
            ]
        ]
        .sort_values(
            ["district_name", "ds_match_std"]
        )
        .reset_index(drop=True)
    )
else:
    print("\nCensus-only: NONE")

if len(qgis_only_std) > 0:
    print("\nQGIS ONLY")
    display(
        qgis_only_std[
            [
                "adm2_name",
                "adm3_name",
                "ds_match_std",
                "adm3_pcode"
            ]
        ]
        .sort_values(
            ["adm2_name", "ds_match_std"]
        )
        .reset_index(drop=True)
    )
else:
    print("\nQGIS-only: NONE")

print("\n" + "=" * 90)
```

```python
# ============================================================
# CELL 6 — FINAL DS NAME STANDARDIZATION
# ============================================================

# Standardize Ja-Ela / Ja Ela
census_flood_std["ds_match_std"] = (
    census_flood_std["ds_match_std"]
    .replace({
        "Ja-Ela": "Ja Ela"
    })
)

# Recalculate matching
census_names_std = set(
    census_flood_std["ds_match_std"].dropna()
)

qgis_names_std = set(
    qgis_flood_std["ds_match_std"].dropna()
)

census_only_std = census_flood_std[
    ~census_flood_std["ds_match_std"].isin(qgis_names_std)
].copy()

qgis_only_std = qgis_flood_std[
    ~qgis_flood_std["ds_match_std"].isin(census_names_std)
].copy()

matched_std = census_flood_std[
    census_flood_std["ds_match_std"].isin(qgis_names_std)
].copy()

print("=" * 90)
print("FINAL FLOOD DS MATCHING")
print("=" * 90)

print("Census flood DS :", len(census_flood_std))
print("QGIS flood DS   :", len(qgis_flood_std))
print("Matched DS      :", len(matched_std))
print("Census-only     :", len(census_only_std))
print("QGIS-only       :", len(qgis_only_std))

print("\n" + "=" * 90)

if len(census_only_std) == 0 and len(qgis_only_std) == 0:
    print("SUCCESS: ALL FLOOD DS DIVISIONS MATCH")
else:
    print("WARNING: SOME DS DIVISIONS ARE STILL UNMATCHED")

print("=" * 90)
```

```python
# ============================================================
# CELL 10 — SEARCH ORIGINAL POPULATION WORKBOOK FOR DISABILITY
# ============================================================

import pandas as pd
import os

population_file = "/Users/ohaseelanka/Downloads/GN_population_excel.xlsx"

# Open workbook structure
population_xls = pd.ExcelFile(population_file)

print("=" * 90)
print("ORIGINAL POPULATION WORKBOOK")
print("=" * 90)

print("\nFile:")
print(population_file)

print("\nAvailable sheets:")
for i, sheet in enumerate(population_xls.sheet_names, start=1):
    print(f"{i:02d}. {sheet}")

# ------------------------------------------------------------
# Search every sheet for disability-related words
# ------------------------------------------------------------

keywords = [
    "disab",
    "disabled",
    "disability",
    "impair",
    "difficulty"
]

print("\n" + "=" * 90)
print("DISABILITY SEARCH")
print("=" * 90)

found = []

for sheet in population_xls.sheet_names:

    try:
        temp = pd.read_excel(
            population_file,
            sheet_name=sheet,
            header=None
        )

        # Convert everything to lowercase strings
        text = temp.astype("string").fillna("")

        matches = text.apply(
            lambda col: col.str.contains(
                "|".join(keywords),
                case=False,
                regex=True,
                na=False
            ).any()
        )

        if matches.any():

            matched_columns = list(
                temp.columns[matches]
            )

            found.append({
                "sheet": sheet,
                "matched_columns": matched_columns
            })

            print(f"\nFOUND in sheet: {sheet}")
            print("Possible columns/locations:", matched_columns)

    except Exception as e:
        print(f"\nCould not read sheet '{sheet}': {e}")

if len(found) == 0:
    print("\nNO DISABILITY-RELATED CONTENT FOUND.")
else:
    print("\nDisability-related content was found.")
```

```python
# ============================================================
# CELL 13 — CREATE CLEAN FLOOD DS FEATURE LIBRARY
# ============================================================

# ------------------------------------------------------------
# 1. Select Census features
# ------------------------------------------------------------

census_features = census_flood_std[
    [
        "district_name",
        "ds_name_join",
        "ds_code",
        "ds_match_std",
        "total_population",
        "number_of_households",
        "children_0_14_ratio",
        "elderly_60_plus_ratio",
        "female_ratio",
        "unprotected_well_ratio",
        "river_stream_water_ratio",
        "no_toilet_ratio",
        "public_toilet_ratio"
    ]
].copy()


# ------------------------------------------------------------
# 2. Select QGIS features
# ------------------------------------------------------------

qgis_features = qgis_flood_std[
    [
        "adm3_name",
        "adm3_pcode",
        "area_sqkm",
        "center_lat",
        "center_lon",
        "total_road_m",
        "road_density",
        "facility_count",
        "facility_density"
    ]
].copy()


# ------------------------------------------------------------
# 3. Merge Census + QGIS
# ------------------------------------------------------------

flood_ds_features = census_features.merge(
    qgis_features,
    left_on="ds_match_std",
    right_on="adm3_name",
    how="inner",
    validate="one_to_one"
)


# ------------------------------------------------------------
# 4. Rename identifiers clearly
# ------------------------------------------------------------

flood_ds_features = flood_ds_features.rename(
    columns={
        "ds_name_join": "census_ds_name",
        "ds_match_std": "ds_name",
        "adm3_name": "qgis_ds_name"
    }
)


# ------------------------------------------------------------
# 5. Arrange columns
# ------------------------------------------------------------

flood_ds_features = flood_ds_features[
    [
        "district_name",
        "ds_name",
        "census_ds_name",
        "qgis_ds_name",
        "ds_code",
        "adm3_pcode",

        # Exposure
        "total_population",
        "number_of_households",

        # Sensitivity
        "children_0_14_ratio",
        "elderly_60_plus_ratio",
        "female_ratio",
        "unprotected_well_ratio",
        "river_stream_water_ratio",
        "no_toilet_ratio",
        "public_toilet_ratio",

        # Infrastructure / spatial context
        "area_sqkm",
        "total_road_m",
        "road_density",
        "facility_count",
        "facility_density",

        # Mapping
        "center_lat",
        "center_lon"
    ]
].copy()


# ------------------------------------------------------------
# 6. Check result
# ------------------------------------------------------------

print("=" * 90)
print("FLOOD DS FEATURE LIBRARY")
print("=" * 90)

print("Shape:", flood_ds_features.shape)

print("\nUnique DS:", flood_ds_features["ds_name"].nunique())

print("\nMissing values:")
display(
    flood_ds_features.isna()
    .sum()
    .to_frame("missing")
    .query("missing > 0")
)

print("\nDuplicate DS names:",
      flood_ds_features["ds_name"].duplicated().sum())

print("\nDuplicate PCODEs:",
      flood_ds_features["adm3_pcode"].duplicated().sum())

print("\nDistrict distribution:")
display(
    flood_ds_features["district_name"]
    .value_counts()
    .sort_index()
    .to_frame("DS_count")
)

print("\nFirst 10 rows:")
display(
    flood_ds_features.head(10)
)

print("\n" + "=" * 90)
```

```python
# ============================================================
# CELL 14 — FLOOD DS FEATURE CORRELATION ANALYSIS
# ============================================================

import numpy as np
import pandas as pd

# Candidate numerical variables
candidate_features = [
    "total_population",
    "number_of_households",
    "children_0_14_ratio",
    "elderly_60_plus_ratio",
    "female_ratio",
    "unprotected_well_ratio",
    "river_stream_water_ratio",
    "no_toilet_ratio",
    "public_toilet_ratio",
    "area_sqkm",
    "total_road_m",
    "road_density",
    "facility_count",
    "facility_density"
]

corr = flood_ds_features[candidate_features].corr(
    method="pearson"
)

print("=" * 90)
print("CORRELATION MATRIX")
print("=" * 90)

display(corr.round(3))


# ------------------------------------------------------------
# Find highly correlated pairs
# ------------------------------------------------------------

print("\n" + "=" * 90)
print("HIGHLY CORRELATED FEATURE PAIRS (|r| >= 0.80)")
print("=" * 90)

high_corr_pairs = []

for i in range(len(candidate_features)):
    for j in range(i + 1, len(candidate_features)):

        feature_a = candidate_features[i]
        feature_b = candidate_features[j]

        r = corr.loc[feature_a, feature_b]

        if abs(r) >= 0.80:
            high_corr_pairs.append({
                "feature_1": feature_a,
                "feature_2": feature_b,
                "correlation": round(r, 3)
            })

high_corr_df = pd.DataFrame(high_corr_pairs)

if len(high_corr_df) > 0:
    display(
        high_corr_df.sort_values(
            "correlation",
            key=lambda x: x.abs(),
            ascending=False
        ).reset_index(drop=True)
    )
else:
    print("No feature pairs reached |r| >= 0.80.")

print("\n" + "=" * 90)
```

```python
# ============================================================
# CELL 17 — INSPECT FLOOD EPISODE STATION PARTICIPATION
# ============================================================

print("=" * 90)
print("STATION PARTICIPATION IN REGIONAL FLOOD EPISODES")
print("=" * 90)

# Count how many regional episodes involved each station
station_counts = {}

for stations in flood_episodes["stations"].dropna():

    for station in stations.split(","):

        station = station.strip()

        station_counts[station] = (
            station_counts.get(station, 0) + 1
        )

station_counts_df = (
    pd.DataFrame(
        list(station_counts.items()),
        columns=["station", "episode_count"]
    )
    .sort_values(
        "episode_count",
        ascending=False
    )
    .reset_index(drop=True)
)

display(station_counts_df)

print("\nNumber of stations represented:")
print(len(station_counts_df))

print("\nStation list:")
print(
    station_counts_df["station"].tolist()
)

print("\n" + "=" * 90)
print("EPISODES BY STATION COUNT")
print("=" * 90)

display(
    flood_episodes[
        [
            "regional_episode_id",
            "start_date",
            "end_date",
            "station_count",
            "stations",
            "max_rainfall_3d",
            "max_rainfall_7d",
            "max_wl_zscore",
            "max_q_zscore"
        ]
    ]
    .sort_values(
        ["station_count", "start_date"],
        ascending=[False, True]
    )
    .reset_index(drop=True)
)

print("=" * 90)
```

```python
# ============================================================
# CELL 31 — FIND POSSIBLE STATION FILES
# ============================================================

from pathlib import Path

search_roots = [
    Path("/Users/ohaseelanka/Desktop/disaster_project_gis"),
    Path("/Users/ohaseelanka/Downloads"),
    Path("/Users/ohaseelanka/Desktop")
]

keywords = [
    "station",
    "stations",
    "kelani",
    "gauge",
    "gauges",
    "hydro",
    "water"
]

extensions = [
    ".shp", ".gpkg", ".geojson", ".csv", ".xlsx", ".xls"
]

matches = []

for root in search_roots:
    if root.exists():
        for path in root.rglob("*"):
            if path.is_file() and path.suffix.lower() in extensions:
                name = path.name.lower()

                if any(keyword in name for keyword in keywords):
                    matches.append(path)

print("=" * 100)
print("POSSIBLE STATION / HYDROLOGICAL FILES")
print("=" * 100)

if matches:
    for path in sorted(set(matches)):
        print(path)
else:
    print("No possible station/hydrological files found.")

print("\nTotal files found:", len(set(matches)))
print("=" * 100)
```

```python
# ============================================================
# CELL 33 — SEARCH FOR KELANI HYDROLOGICAL STATIONS
# ============================================================

import pandas as pd

station_file = "/Users/ohaseelanka/Downloads/Met_Stations.xlsx"

target_ids = ["KN02", "KN07", "KN11", "KN16", "KN18", "KN21"]

print("=" * 100)
print("SEARCHING FOR KELANI HYDROLOGICAL STATIONS")
print("=" * 100)

found = []

xls = pd.ExcelFile(station_file)

for sheet in xls.sheet_names:
    df = pd.read_excel(station_file, sheet_name=sheet)

    # Search every cell for our station IDs
    for target in target_ids:
        mask = df.astype(str).apply(
            lambda col: col.str.strip().str.upper().eq(target)
        ).any(axis=1)

        matches = df.loc[mask]

        if not matches.empty:
            print(f"\nFOUND {target} in sheet: {sheet}")
            display(matches)
            found.append((sheet, target))

print("\n" + "=" * 100)
print("SUMMARY")
print("=" * 100)

if found:
    for sheet, station in found:
        print(f"{station} → {sheet}")
else:
    print("None of the six KN station IDs were found in Met_Stations.xlsx.")

print("=" * 100)
```

```python
# ============================================================
# CELL 34 — SEARCH FOR KELANI GAUGE STATION NAMES
# ============================================================

import pandas as pd
from pathlib import Path

search_files = [
    Path("/Users/ohaseelanka/Downloads/Research_Study_Areas_Kalu_Kelani_Nuwara_Eliya.xlsx"),
    Path("/Users/ohaseelanka/Downloads/rain_gauge_shortlist.xlsx"),
]

target_names = [
    "NORWOOD",
    "KITULGALA",
    "HOLOMBUWA",
    "DERANIYAGALA",
    "GLENCOURSE",
    "HANWELLA"
]

print("=" * 100)
print("SEARCHING FOR KELANI STATION NAMES")
print("=" * 100)

found = []

for file in search_files:

    print(f"\nFILE: {file.name}")

    try:
        xls = pd.ExcelFile(file)

        for sheet in xls.sheet_names:

            df = pd.read_excel(file, sheet_name=sheet)

            for target in target_names:

                mask = df.astype(str).apply(
                    lambda col: col.str.strip().str.upper().str.contains(
                        target,
                        na=False,
                        regex=False
                    )
                ).any(axis=1)

                matches = df.loc[mask]

                if not matches.empty:
                    print(f"\nFOUND: {target}")
                    print(f"Sheet: {sheet}")
                    display(matches)

                    found.append(
                        (file.name, sheet, target)
                    )

    except Exception as e:
        print("Could not read:", file)
        print("Error:", e)

print("\n" + "=" * 100)
print("SUMMARY")
print("=" * 100)

if found:
    for item in found:
        print(item)
else:
    print("No Kelani station names were found.")

print("=" * 100)
```
