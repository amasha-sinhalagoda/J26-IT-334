```python
import sys
import pandas as pd

from pathlib import Path

BASE = Path(
    "/Users/ohaseelanka/disaster_analytics_hub"
)

if str(BASE) not in sys.path:
    sys.path.insert(0, str(BASE))

from src.human_impact_predictor import HumanImpactPredictor

for hazard in ["flood", "landslide"]:

    print("\n" + "=" * 55)
    print(hazard.upper(), "PREDICTION TEST")
    print("=" * 55)

    predictor = HumanImpactPredictor(hazard)

    test = pd.read_csv(
        BASE / f"data/final/{hazard}_test.csv"
    )

    sample = test.iloc[[0]]

    result = predictor.predict(sample)

    print("Predicted impact:", result["predicted_impact"])
    print("Predicted probability:", round(result["confidence"], 4))

    print("\nClass probabilities:")
    for label, probability in result[
        "class_probabilities"
    ].items():
        print(f"  {label}: {probability:.4f}")

    print("\nTop SHAP contributors:")
    for item in result["shap_explanation"][:3]:
        print(
            f"  {item['feature_name']}: "
            f"{item['shap_value']:+.4f}"
        )
```
