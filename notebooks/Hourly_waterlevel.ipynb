```python

import sys
from pathlib import Path

import pandas as pd
import numpy as np

# Display environment information
print("Python version:", sys.version)
print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)

# Find the project root directory
PROJECT_DIR = Path.cwd().resolve()

while not (PROJECT_DIR / "data" / "raw").is_dir():
    if PROJECT_DIR.parent == PROJECT_DIR:
        raise FileNotFoundError("Project directory not found.")
    PROJECT_DIR = PROJECT_DIR.parent

# Input: original hourly water-level Excel file
input_file = (
    PROJECT_DIR
    / "data"
    / "raw"
    / "water_level"
    / "kelani"
    / "2015 - 2025 hourly waterlevel.xlsx"
)

# Output: processed Kelani water-level folder
output_dir = (
    PROJECT_DIR
    / "data"
    / "processed"
    / "kelani"
)

output_dir.mkdir(parents=True, exist_ok=True)

print("\nProject directory:", PROJECT_DIR)
print("Input file:", input_file)
print("Input file exists:", input_file.exists())
print("Output directory:", output_dir)
```

```python
# Output file
output_file = output_dir / "kelani_hourly_waterlevel.xlsx"

with pd.ExcelFile(input_file, engine="openpyxl") as workbook:
    with pd.ExcelWriter(output_file, engine="openpyxl") as writer:

        for sheet_name in workbook.sheet_names:
            # Read actual data; Excel row 4 contains the column headers
            df = pd.read_excel(
                workbook,
                sheet_name=sheet_name,
                header=3
            )

            # Remove completely empty rows and columns
            df = df.dropna(axis=1, how="all")
            df = df.dropna(how="all").reset_index(drop=True)

            # Clean column names
            df.columns = df.columns.astype(str).str.strip()

            # Add Location as the first column
            df.insert(0, "Location", sheet_name)

            # Save the worksheet
            df.to_excel(
                writer,
                sheet_name=sheet_name[:31],
                index=False
            )

            print(
                f"Processed: {sheet_name} | "
                f"Rows: {len(df)}"
            )

print("\nAll worksheets processed successfully.")
print("Saved file:", output_file)
```

```python
from pathlib import Path
import pandas as pd

# Project directory
PROJECT_DIR = Path.cwd().resolve()

while not (PROJECT_DIR / "data" / "processed" / "kelani").is_dir():
    if PROJECT_DIR.parent == PROJECT_DIR:
        raise FileNotFoundError("Project directory not found")
    PROJECT_DIR = PROJECT_DIR.parent

# File paths
processed_dir = PROJECT_DIR / "data" / "processed" / "kelani"

input_file = processed_dir / "kelani_hourly_waterlevel.xlsx"
output_file = processed_dir / "kelani_hourly_waterlevel_processed.xlsx"

# Read the existing Excel file and save a new copy
with pd.ExcelFile(input_file, engine="openpyxl") as workbook:
    with pd.ExcelWriter(output_file, engine="openpyxl") as writer:
        for sheet_name in workbook.sheet_names:
            df = pd.read_excel(
                workbook,
                sheet_name=sheet_name
            )

            df.to_excel(
                writer,
                sheet_name=sheet_name,
                index=False
            )

print("File saved successfully:", output_file)
```

```python
# Remove Unnamed columns from the Norwood sheet only

with pd.ExcelFile(output_file, engine="openpyxl") as workbook:
    sheets = {
        sheet_name: pd.read_excel(workbook, sheet_name=sheet_name)
        for sheet_name in workbook.sheet_names
    }

# Update only Norwood
norwood_sheet = next(
    name for name in sheets
    if name.strip().lower() == "norwood"
)

df = sheets[norwood_sheet]

# Remove columns whose names start with "Unnamed"
df = df.loc[
    :,
    ~df.columns.astype(str).str.startswith("Unnamed:")
]

sheets[norwood_sheet] = df

# Save all sheets back to the same file
with pd.ExcelWriter(output_file, engine="openpyxl") as writer:
    for sheet_name, sheet_df in sheets.items():
        sheet_df.to_excel(
            writer,
            sheet_name=sheet_name,
            index=False
        )

print("Updated Norwood sheet.")
print("Columns:", df.columns.tolist())
print("\nFirst 5 rows of Norwood:")
display(df.head(5))
```

```python
import pandas as pd

# File path
input_file = processed_dir / "kelani_hourly_waterlevel_final.xlsx"

# Read all worksheets
with pd.ExcelFile(input_file, engine="openpyxl") as workbook:
    sheets = {
        sheet_name: pd.read_excel(workbook, sheet_name=sheet_name)
        for sheet_name in workbook.sheet_names
    }

# Process each worksheet
for sheet_name, df in sheets.items():

    # Convert numeric month values to integers, preserving text months
    month = pd.to_numeric(df["Month"], errors="coerce")
    day = pd.to_numeric(df["Date"], errors="coerce")
    year = pd.to_numeric(df["Year"], errors="coerce")

    # Build the actual date from Year, Month and day
    dates = pd.to_datetime(
        {"year": year, "month": month, "day": day},
        errors="coerce"
    )

    # Add the new Date column at the end
    df["Date"] = dates.dt.strftime("%Y-%m-%d")

    sheets[sheet_name] = df

# Update the same Excel file
with pd.ExcelWriter(input_file, engine="openpyxl") as writer:
    for sheet_name, df in sheets.items():
        df.to_excel(writer, sheet_name=sheet_name, index=False)

print("Date column added and file updated successfully.")

# Verify the first 5 rows of every worksheet
with pd.ExcelFile(input_file, engine="openpyxl") as workbook:
    for sheet_name in workbook.sheet_names:
        print(f"\nWorksheet: {sheet_name}")
        display(pd.read_excel(workbook, sheet_name=sheet_name).head(5))
```

```python
# Rename the first column to Location in every worksheet

with pd.ExcelFile(input_file, engine="openpyxl") as workbook:
    sheets = {
        sheet_name: pd.read_excel(workbook, sheet_name=sheet_name)
        for sheet_name in workbook.sheet_names
    }

for sheet_name, df in sheets.items():
    df.rename(columns={df.columns[0]: "Location"}, inplace=True)
    sheets[sheet_name] = df

# Save changes to the same file
with pd.ExcelWriter(input_file, engine="openpyxl") as writer:
    for sheet_name, df in sheets.items():
        df.to_excel(writer, sheet_name=sheet_name, index=False)

print("First column renamed to Location in all worksheets.")

# Verify the first 5 rows of each sheet
with pd.ExcelFile(input_file, engine="openpyxl") as workbook:
    for sheet_name in workbook.sheet_names:
        print(f"\nWorksheet: {sheet_name}")
        display(pd.read_excel(workbook, sheet_name=sheet_name).head(5))
```

```python
# Input and output file paths
input_file = Path("../data/processed/kelani/kelani_hourly_waterlevel_final.xlsx")
output_file = Path("../data/processed/kelani/kelani_hourly_waterlevel_combined.xlsx")

# Read and combine all worksheets
all_sheets = pd.read_excel(input_file, sheet_name=None, engine="openpyxl")

combined_df = pd.concat(
    all_sheets.values(),
    ignore_index=True
)

# Save as one worksheet
combined_df.to_excel(
    output_file,
    sheet_name="Combined",
    index=False,
    engine="openpyxl"
)

# Verify the output
print("Combined file saved successfully:", output_file.resolve())
print("Total rows:", len(combined_df))
print("Total columns:", len(combined_df.columns))
print("\nFirst 5 rows:")
display(combined_df.head())

print("\nLocations included:")
print(combined_df["Location"].unique())
```

```python
# File path
file_path = Path("../data/processed/kelani/kelani_hourly_waterlevel_combined.xlsx")

# Read the combined dataset
df = pd.read_excel(file_path, engine="openpyxl")

# 1. Dataset overview
print("Dataset shape:", df.shape)
print("\nFirst 5 rows:")
display(df.head())

# 2. Count missing values in each column
missing_summary = pd.DataFrame({
    "Column": df.columns,
    "Missing_Count": df.isnull().sum().values,
    "Missing_Percentage": (df.isnull().sum().values / len(df) * 100).round(2)
})

missing_summary = missing_summary.sort_values(
    by="Missing_Count", ascending=False
)

print("\nMissing values by column:")
display(missing_summary)

# 3. Total missing values
print("\nTotal missing cells:", df.isnull().sum().sum())
print("Total rows containing missing values:", df.isnull().any(axis=1).sum())

# 4. Display rows containing missing values
print("\nRows containing missing values:")
display(df[df.isnull().any(axis=1)].head(10))
```
